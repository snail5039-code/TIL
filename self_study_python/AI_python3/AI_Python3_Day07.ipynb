**함수 매핑**

```python
import pandas as pd
import seaborn as sns

# seaborn 라이브러리의 mpg 데이터셋 불러오기
# mpg : 1970~1982년 미국, 유럽, 일본 자동차의 연비 및 제원 데이터
df = sns.load_dataset("mpg")

df.head()
```

**1)map()**

```python
# 딕셔너리 매핑으로 origin열의 값을 한국어로 변환

df['origin_kr'] = df['origin'].map({'usa' : '미국', 'japan' : '일본', 'europe' : '유럽'})

# 변환 전/후 비교
# - 변환하기 전 국가별 데이터 개수 확인
print(df['origin'].value_counts())
print()
# - 변환 후 국가별 데이터 개수 확인
print(df['origin_kr'].value_counts())

# 데이터 확인
df[['origin', 'origin_kr']].head(10)
```

**2) 함수 lambda 매핑**

```python
# lambad 매개변수 : 반환값

(lambda x : x + 10)(10)
```

```python
# lambda로 model_year열의 각 원소에 1900을 더해 실제 연도로 변환

# lambda x : x + 1900
# -> x : 시리즈(열)의 각 원소
# -> x + 1900 : 각 원소에 1900을 더한 결과

df['year'] = df['model_year'].map(lambda x : x + 1900)

# 변환 전/후 비교
df[['model_year', 'year']].head(10)
```

**3) apply()**

```python
# mpg(마일 단위 연비)를 km/L(킬로미터 단위 연비)로 변환
# - 1 mpg = 0.425 km/L

# -> x : mpg열의 각 원소
# -> x * 0.425 : 마일 단위를 킬로미터 단위로 변환(ex : 18.0 * 0.425)
# -> round(..., 2) : 소수점 둘째자리까지 반올림
df['mpg_km'] = df['mpg'].apply(lambda x : round(x * 0.425, 2))

# 변환 전/후 비교
df[['mpg', 'mpg_km']].head()
```

```python
# 조건식을 여러개 활용해아 하는 경우
# def로 사용자 정의 함수를 만들어 apply()에 전달해야 한다

# mpg(연비)값을 기준으로 연비 등급(우수/보통/미흡)을 분류

# 1) 연비 등급을 분류하는 함수를 정의
def mpg_grade(x) :
    # x : mpg 열의 각 원소가 차례대로 전달됨
    if x >= 30 :
        # 연비가 30이상이면
        return "우수"
    elif x >= 20 :
        # 연비가 20이상 30미만이면
        return "보통"
    else :
        # 연비가 20미만이면
        return "미흡"

# 2) apply(함수명) : mpg열(시리즈)의 각 원소를 mpg_grade() 넣어 결과를 반환
df['mpg_grade'] = df['mpg'].apply(mpg_grade)

# 등급별 데이터 개수 확인
print(df['mpg_grade'].value_counts())

# 변환 결과 확인
df[['mpg', 'mpg_grade']].head(10)
```

```python
# <함수 매핑 실습>
import pandas as pd
import seaborn as sns

# 타이타닉 데이터 셋 불러오기
df = sns.load_dataset("titanic")

df.head()
```

```python
# 1) 성별 컬럼을 map() + 딕셔너리를 사용하여 한국어로 변환하고
#   새로운 열 'sex_kr'로 추가한 뒤 성별 인원수를 출력
# ('male' -> 남성, 'female' -> 여성)

df['sex_kr'] = df['sex'].map({'male' : '남성', 'female' : '여성'})

# 성별 인원수 확인
print(df['sex_kr'].value_counts())

# 변환 전/후 비교
df[['sex', 'sex_kr']].head()
```

```python
# 2) map() + lambda를 활용하여 fare(요금)를 원하로 변환하고
#   새로운 열 'fare_kr'로 추가한 뒤 출력
#   (1달러 = 1400원)

df['fare_kr'] = df['fare'].map(lambda x : x * 1400)

# 변환 전/후 비교
df[['fare', 'fare_kr']].head()
```

```python
# 3) apply() + def 함수를 사용하여 아래 기준으로 나이(age) 등급을 분류하고
#   새로운 열 'age_grade'로 추가한 뒤 등급별 인원수를 출력
#   - 18세 미만 : '미성년'
#   - 18세 이상 65세 마만 : 성인'
#   - 65세 이상 : '고령'

# (1) 나이 등급을 분류하는 함수 정의
def age_grade(x) :
    # x : age열의 각 원소가 차례대로 입력됨
    if x < 18 :
        return "미성년"
    elif x < 65 :
        return "성인"
    else :
        return "고령"

# apply(함수명)
df['age_grade'] = df['age'].apply(age_grade)

# 나이 등급별 인원수 확인
print(df['age_grade'].value_counts())
```

**그룹화**

```python
import pandas as pd
import seaborn as sns

# 타이타닉 데이터 셋 불러오기
df = sns.load_dataset("titanic")

# 4개 열 선택
df = df[['age', 'class', 'fare', 'survived']]

df.head()
```

```python
# class(객실 등급)별로 그룹화

# 1) class열의 데이터 종류와 개수
print(df['class'].value_counts())

# 2) class별로 그룹화
# df.group_by(열)
groups = df.groupby('class')
# print(groups)

# for group in groups :
    # print(type(group)) # (key, 데이터프레임) 튜플형태
    # print(group)

for key, group in groups :
    print(f"클래스 등급 : {key}")
    print(f"{key}클래스 승객 인원수 : {len(group)}명")
    print(group.head(), "\n")
```

```python
# class별로 평균 나이, 요금, 생존률 확인
groups[['age', 'fare', 'survived']].mean()
```

```python
# 그룹객체에서 특정 그룹 선택
# 그룹객체.get_group(key값)

# Third 클래스만 선택
group_third = groups.get_group("Third")
group_third.head()
```

```python
# 여러 열을 기준으로 그룹화

import pandas as pd
import seaborn as sns

df = sns.load_dataset("titanic")

# 5개 열 선택
df = df[['survived', 'sex', 'fare', 'age', 'class']]

# class열과 sex열로 그룹화
# df.groupby([열 리스트])
groups = df.groupby(['class', 'sex'])                    

for key, group in groups :
    print(f"key : {key}")
    print(f"승객수 : {len(group)}명")
    print(group.head(), "\n")

df.head()
```

```python
# 그룹별 평균 나이, 요금, 생존률

groups[['age', 'fare', 'survived']].mean()
```

```python
# 특정 그룹 선택

# class가 Third인 남성 그룹 선택
group = groups.get_group(('Third', 'male')) # key를 튜플로 작성
group.head()
```

**그룹 연산 집계**

```python
# class열을 기준으로 그룹화

groups = df.groupby('class')

# 그룹별 나이, 요금 최대값과 최소값의 차이
# 그룹객체.agg(매핑 함수)

# 1) 각 그룹마다의 컬럼 열의 최대값의 차이를 계산하는 함수 정의
def max_min(group) :
    # group : class 별 그룹(데이터 프레임)을 전달받을 매개변수
    return group.max() - group.min() # 최대값 - 최소값

result_agg = groups[['age', 'fare']].agg(max_min)

# 람다 활용
result_agg = groups[['age', 'fare']].agg(lambda group : group.max() - group.min())

result_agg
```

```python
# agg()활용 모든 열에 여러 함수를 적용하여 집계
#   : 그룹.agg([함수1,...])

# class별 나이와 요금의 최대값과 최소값 집계 
agg_all = groups[['age', 'fare']].agg(['min', 'max'])
agg_all
```

```python
# agg()활용 각 열마다 다른 함수를 사용하여 집계 
#   : 그룹객체.agg({'열' :함수1,...})
# class별 요금(fare)의 최소값과 최대값, 나이(age)의 평균 집계 

agg_result = groups[['age', 'fare']].agg({'fare': ['min', 'max'], 'age' : 'mean'})
agg_result
```

## 문제 2) apply(lambda)를 활용한 조건부 데이터 생성 (MPG 데이터)

- Seaborn의 mpg 데이터셋을 로드.

- 연비('mpg') 열의 값을 기준으로 자동차의 효율성을 분류하여 'efficiency' 열을 추가.

    - 기준: 연비가 25 이상이면 'High', 그렇지 않으면 'Low'

- apply()와 lambda를 활용하여 처리하고, 결과 출력.

```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("mpg")

df['efficiency'] = df['mpg'].apply(lambda x : 'High' if x >= 25 else 'Low')
df
```

## 문제 4) agg()를 활용한 열별 다중 집계 (Tips 데이터)

- Seaborn의 tips 데이터셋을 로드.

- 요일('day')별로 그룹화한 뒤, 다음의 통계량을 한 번에 확인.

    - 식사 총액('total_bill'): 합계(sum)

- 팁('tip'): 평균(mean)과 최댓값(max)

agg() 메소드에 딕셔너리 형태를 전달하여 처리.

```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("tips")
df.groupby('day').agg({'total_bill' : 'sum', 'tip' : ['mean', 'max']})
```
