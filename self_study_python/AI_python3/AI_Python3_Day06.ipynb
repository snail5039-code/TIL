```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("titanic")

# pd.qcut()으로 요금(fare)을 4개의 구간으로 분할

fare_qcut = pd.qcut(df['fare'],
                    q=4,
                    labels=['저가', '중저가', '중고가', '고가'])

# 구간 분할 결과를 새로운 열로 추가
df['fare_group'] = fare_qcut

# 구간별 데이터 수 확인
print(df['fare_group'].value_counts())

df[['fare', 'fare_group']].head()
```

```python
# 두 방법의 차이
# pd.cut() : 구간의 폭(범위)를 동일하게 나눔
#   -> 요금이 저가에 집중되어 있으면 저가 구간에 데이터가 몰림
# pd.qcut() : 구간의 데이터 수를 동일하게 나눔
#   -> 항상 고르게 분표되지만 구간의 폭이 불균일할 수 있음

# 구간의 의미(기준)가 중요할때 => pd.cut()
# 구간별 균동한 분포가 필요할 때 => pd.qcut()
```

```python
# <구간 분할 실습>
import pandas as pd
import seaborn as sns

df = sns.load_dataset("tips")

df.head()
```

```python
# 1) 식사 금액(total_bill)을 pd.cut()으로 아래 기준에 맞게 구간 분할하고,
#   새로운 열 'bill_group'으로 추가한뒤 구간별 데이터 수를 출력하기
# 0 ~ 10 달러 : '소액'
# 11 ~ 20 달러 : '중액'
# 21 ~ 30달러 : '고액'
# 31달러 이상 : '초고액'
# bins = [0, 10, 20, 30, float('inf')]

df['bill_group'] = pd.cut(df['total_bill'],
                          bins=[0, 10, 20, 30, float('inf')],
                          labels=['소액', '중액', '고액', '초고액'])
# float('inf') : 무한대를 의미하며, 31달러 이상의 모든 값을 마지막 구간에 포함
print(df['bill_group'].value_counts())
```

```python
# 2) 팁 금액(tip)을 pd.qcut()으로 3등분하고
#    새로운 열 'tip_group'으로 추가한뒤 구간별 데이터 개수 확인

df['tipt_group'] = pd.qcut(df['tip'],
                           q=3,
                           labels=['소팁', '중팁', '대팁'])
print(df['tipt_group'].value_counts())
```

```python
# 3) 1)에서 만든 'bill_group'열을 활용하여
#   '초고액' 구간의 손님 중 흡연자(smoker == 'Yes')가 몇 명인지 출력=

# (1) bill_group이 '초고액'인 행만 추출
df_high = df.loc[df['bill_group'] == '초고액']

# (2) 초고액 손님 중 흡연자만 다시 필터링
df_smoker = df_high.loc[df_high['smoker'] == 'Yes']

print(f"초고액 손님 수 : {len(df_high)}명")
print(f"초고액 손님 중 흡연자 수 : {len(df_smoker)}명")

df_smoker
```

## 문제 4) pd.qcut()을 활용한 동일 개수 구간 분할 (Tips 데이터)

- Seaborn의 tips 데이터셋을 로드.

- 식사 총액('total_bill') 데이터를 동일한 데이터 개수를 가지도록 3개 구간으로 나누어 'bill_level' 열을 추가.

- 구간 라벨은 ['Lower', 'Middle', 'Upper']로 지정.

- value_counts()를 통해 각 구간별 데이터 수가 균등하게 배분되었는지 확인.

```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("tips")

df['bill_level'] = pd.qcut(df['total_bill'],
                           q=3,
                           labels=['Lower', 'Middle', 'Upper'])
df['bill_level'].value_counts()
```
