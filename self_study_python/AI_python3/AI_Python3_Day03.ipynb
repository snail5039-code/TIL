```python
day03

0. 복습
데이터 분석 패키지
- 넘파이
- 판다스
- 맷플로립

넘파이
- 수치 연산을 빠르게 할 수 있는 수치 해석 및 계산을 위한 패키지
- 다차원 배열 기반으로 수치 계산에 필요한 다양한 기능 제공 

1) 넘파이 배열 생성
	import numpy as np
	배열명 = np.array(리스트 or 튜플)
2) 넘파이 배열 요소 접근
 : 리스트와 유사하게 인덱스 번호 사용
	배열명[idx]
	배열명[행 idx][열 idx]
3) 넘파이 배열 속성
- .dtype : 배열의 요소 타입
- .shape : 배열의 구조 
	1차원의 경우 : (열,)
	2차원의 경우 : (행, 열)
- .ndim : 차원의 수
- .size : 전체 요소의 개수

4) 배열의 연산
- 넘파이 배열 + 값 => 배열의 모든 요소에 연산 
- 넘파이 배열 + 넘파이 배열 
	배열의 구조가 같은 경우 : 같은 자리의 원소끼리 연산
	배열의 구조가 다를 경우 : 브로드캐스팅 기능 

1. 판다스(pandas)
- 데이터 분석을 위한 효율적인 데이터 구조를 제공(엑셀과 비슷한 형태)
- 서로 다른 여러가지 유형의 데이터를 공통의 포맷으로 정리하는 것 

1) 판다스의 특징
- 데이터를 읽어들이거나 그래프화 등 데이터 분석 작업을 쉽게 가능하게 함 
- 많은 양의 데이터를 처리할 수 있다(엑셀은 데이터가 많아지면 과부화 발생)
- 1차원 형태의 데이터 구조 : Series
- 2차원 형태의 데이터 구조 : DataFrame

2. 시리즈(Series)
- 데이터가 순차적으로 나열된 1차원 형태를 가진다(1열 형태)
- 각 값에 대응되는 인덱스를 부여할 수 있는 구조(인덱스와 데이터가 일대일 대응)
- 데이터, 행의 이름(인덱스), 열의 제목(name)으로 구성되어 있다
- 딕셔너리 구조와 유사(key : value), 시리즈의 구조(index : value)

1) 시리즈의 생성
import pandas as pd
시리즈명 = pd.Series(데이터, index=행이름, name=열이름)
- 시리즈는 모든 유형의 데이터(정수, 실수, 문자열)를 담을 수 있음

2) 시리즈의 속성
- 시리즈.values : 데이터 값(넘파이 배열)
- 시리즈.index : 인덱스(행이름)값
- 시리즈.name : 열제목

3) 딕셔너리 => 시리즈 생성
- 딕셔너리 : {key:value}
- pd.Series(딕셔너리)
- 딕셔너리의 key = 시리즈의 index
  딕셔너리의 value => 시리즈의 value

4) 시리즈의 원소 선택
- 정수형 위치 인덱스를 사용
	- 리스트와 같음, 시리즈명[idx]
	- 시리즈.iloc(idx)

- 인덱스 이름(행 이름, 라벨)을 사용
	- 딕셔너리와 같음, 시리즈명[label]
	- 시리즈명.loc[label]

5) 시리즈의 원소 추가, 수정, 삭제
(1) 원소 추가 : 딕셔너리의 요소 추가 방법과 유사
	sr[추가할 label] = 값
(2) 원소 수정 : 딕셔너리, 리스트 방법과 유사
	sr[idx] = 값
	sr[label] = 값
(3) 원소 삭제 : del 키워드 사용
	del 시리즈[label]

3. 데이터 프레임(Data Frame)
- 판다스의 2차원 데이터 구조(행과 열을 가지는 평면 구조)
- 행 방향 인덱스(index), 열방향 인덱스(column)가 존재함
- 시리즈, 리스트, 딕셔너리 등으로 생성 가능함

1) 기본 구조 및 속성 
- 데이터프레임.values : 몇행 몇열인지에 대한 데이터 값
		(2차원 넘파이 배열)
- 데이터프레임.index : 행 인덱스 
- 데이터프레임.columns : 열 제목
- 행과 열은 지정하지 않으면 0부터 시작하는 정수 값이 들어간다

2) 데이터 프레임 생성
import pandas as pd

데이터프레임명 = pd.DataFrame(데이터, index=행제목, columns=열제목)

3) 딕셔너리 -> 데이터 프레임
- pd.DataFrame(딕셔너리)
- 딕셔너리 => 시리즈 : 딕셔너리의 key가 시리즈의 index로
- 데이터 프레임에서는
	딕셔너리의 key가 데이터프레임의 columns로

4) 데이터 선택
(1) 행 선택(loc, iloc)
	
		df.loc				df.iloc
===============================================================	
선택기준	인덱스 이름(라벨)		정수형 위치 인덱스
여러행선택	[[행이름1, 행이름2]]		[[idx, idx]]
슬라이싱	범위 끝 포함			범위 끝 제외
		ex) ['a':'c']			ex) [1:3] => 1, 2
		   => 'c','b','c'

(2) 열 선택
- df["열이름"]
- df.열이름 (열이름이 문자열일 경우만 사용 가능)
- 여러 열 선택
	df[[열이름1,...]]

(3) 원소 선택(loc, iloc)
- 인덱스 이름 : df.loc[행이름, 열이름]
- 정수 위치 인덱스 : df.iloc[행번호, 열번호]
```

**시리즈의 원소 선택**

```python
import pandas as pd

# 시리즈 생성
sr = pd.Series({"과목명":"python", "수강일":"2026-06-18", "수강여부":True},
               name = "학생1")
# 1) 인덱스 번호 사용
# print(sr[0])
print(sr.iloc[0])

# 2) 인덱스 이름(라벨)을 사용
print(sr["수강일"])
print(sr.loc["수강일"])

# 3) 여러값을 인덱싱
# print(sr[[1, 2]])
print(sr[["수강일", "수강여부"]])
print(sr[:2]) # idx 0 ~ 1까지 슬라이싱 => 마지막 범위는 포함되지 않음
print(sr["수강일":"수강여부"]) # 이름으로 슬라이싱할 경우 => 마지막 범위 포함

sr
```

```python
# +) 조건으로 선택(필터링)
# 관계 연산자(>,<,..)나 논리 연산자 사용

sr = pd.Series({"python":90, "java":80, "c":70}, name="성적표")

# 1) 70점보다 높은 것만 선택
print(sr > 70) # 조건에 맞으면 True, 아니면 False(불린 시리즈)
print(sr[sr > 70])

print("======================")
# 2) 60점 보다 높고 90점보다 낮은 것만 선택
print(sr[(sr > 60) & (sr < 90)])

# 판다스의 논리 연산
# - and 연산 : &
# - or 연산 : |
# - not 연산 : ~

sr
```

**행선택(iloc)**

```python
import pandas as pd

df = pd.DataFrame({"A":[1, 2, 3],
                   "B":[4, 5, 6],
                   "C" :[7, 8, 9]}, index = ['a', 'b', 'c'])
# 1) loc 사용한 행 선택
print("==== loc 사용 행 선택 ====")
print(df.loc["a"], type(df.loc["a"]))
# 데이터 프레임에서 1개의 행, 1개의 열 => 시리즈
# 2) iloc 사용한 행 선택
print("\n==== iloc를 사용한 행 선택 ====")
print(df.iloc[2])

# 3) 여러 행 선택
print("==== loc로 여러 행 선택 ====")
print(df.loc[["a", "b"]])

print("==== iloc로 여러 행 선택 ====")
print(df.iloc[[0, 1]])

df
```

```python
# 4) 슬라이싱
print("==== loc를 사용한 슬라이싱 ====")
print(df.loc["a":"b"]) # 마지막 b행 포함

print("\n==== iloc를 사용한 슬라이싱 ====")
print(df.iloc[0:2]) # 마지막 idx 2번 행 포함x

df
```

```python
import pandas as pd

# 1) 1개의 열 선택
print(df["A"], type(df["A"])) # 시리즈 형태

# 2) 여러 열 선택
print("-----------------------")
print(df[["A", "C"]])

df
```

##문제 1) 딕셔너리를 활용한 시리즈 생성 및 인덱싱

- 아래의 카페 메뉴와 가격 데이터를 사용하여 menu_sr이라는 판다스 시리즈를 생성.

- 생성된 시리즈에서 '카페라떼'의 가격을 라벨 인덱싱(loc)과 정수 인덱싱(iloc)을 각각 사용하여 출력.

- '아메리카노'부터 '바닐라라떼'까지 슬라이싱하여 출력.

        <데이터>
        menu_data = {'아메리카노': 3000, '카페라떼': 4000, '바닐라라떼': 4500, '카푸치노': 4200}

        <출력결과>
        라벨 인덱싱(loc) : 4000
        정수 인덱싱(iloc) : 4000
        --- 메뉴 슬라이싱 ---
        아메리카노    3000
        카페라떼      4000
        바닐라라떼    4500
        dtype: int64

```python
import pandas as pd

menu_data = {'아메리카노': 3000, '카페라떼': 4000, '바닐라라떼': 4500, '카푸치노': 4200}

sr = pd.Series(menu_data, name="메뉴")

print(f"라벨 인덱싱(loc) : {sr.loc["카페라떼"]}")
print(f"정수 인덱싱(iloc) : {sr.iloc[1]}")

print("--- 메뉴 슬라이싱 ---")
print(f"아메리카노 : {sr.iloc[0]}")
print(f"카페라떼 : {sr.iloc[1]}")
print(f"바닐라라떼 : {sr.iloc[2]}")
sr
```

##문제 2) 시리즈 필터링 및 데이터 변경

- 학생 5명의 시험 성적 데이터를 시리즈로 생성한 후, 점수가 85점 이상인 학생들만 추출하여 출력.

- 시리즈에 '학생F'의 점수 92점을 새로 추가하세요.

- '학생C'의 데이터를 시리즈에서 삭제한 후 최종 시리즈를 출력.

        <데이터>
        scores = [80, 95, 70, 88, 60]
        names = ['학생A', '학생B', '학생C', '학생D', '학생E']

        <출력결과>
        --- 85점 이상 학생 ---
        학생B    95
        학생D    88
        dtype: int64
        --- 최종 데이터 ---
        학생A    80
        학생B    95
        학생D    88
        학생E    60
        학생F    92
        dtype: int64

```python
import pandas as pd

scores = [80, 95, 70, 88, 60]
names = ['학생A', '학생B', '학생C', '학생D', '학생E']

sr = pd.Series(scores, index = names, name="점수")
print("--- 85점 이상 학생 ---")
print(sr[sr >= 85])

sr.loc["학생F"] = 92
del sr["학생C"]
print("--- 최종 데이터 ---")
print(sr)
sr
```

##문제 3) 딕셔너리를 활용한 데이터프레임 생성 및 변환

- 아래의 상품 재고 데이터를 사용하여 데이터프레임 inventory_df를 생성.

- 행 인덱스는 'p1', 'p2', 'p3'로 지정.

- iloc를 사용하여 첫 번째(p1)와 세 번째(p3) 행만 선택하여 출력.

- loc를 사용하여 'p2'부터 'p3'까지의 행과 '상품명', '가격' 열을 선택하여 출력.

        <출력결과>
        --- 1번 결과 (iloc) ---
        상품명      가격  재고
        p1  모니터  300000  10
        p3  마우스   30000  50
        --- 2번 결과 (loc) ---
        상품명     가격
        p2  키보드  50000
        p3  마우스  30000

```python
import pandas as pd

product = {"상품명":["모니터", "키보드", "마우스"], "가격":[300000, 50000, 30000], "재고": [10, 50, 30]}

df = pd.DataFrame(product, index=["p1", "p2", "p3"])
print("--- 1번 결과 (iloc) ---")
print(df.iloc[[0,2]])

print("--- 2번 결과 (loc) ---")
print(df.loc["p2" : "p3", ["상품명", "가격"]])
df
```
