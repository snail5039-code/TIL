## 문제 5) 서브플롯 (Subplot) 활용: 복합 시각화

- plt.subplot()을 사용하여 1행 2열 구조의 그래프 시각화.

- 첫 번째 칸(왼쪽)에는 문제 3의 선 그래프를, 두 번째 칸(오른쪽)에는 문제 4의 막대 그래프를 배치.

- 전체 그림의 크기(figsize)는 (12, 5)로 설정.

```python
fig = plt.figure(figsize=(12, 5))
plt.rc("font", family="NanumBarunGothic")

month = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
visitors = [150, 200, 180, 300, 250]

langs = ['Python', 'Java', 'C++', 'JavaScript']
pref = [45, 25, 15, 15]

plt.subplot(1, 2, 1)
plt.plot(month, visitors, marker='s')
plt.title("방문자 추이")

plt.subplot(1, 2, 2)
plt.bar(langs, pref)
plt.title("언어 선호도")

plt.show()
```
