```python
# readline() : 파일 내용을 한줄씩 읽어옴
with open("test1.txt", "r") as f :
    print(f.readline(), end="") # 첫번째 줄
    print(f.readline(), end="") # 두번째 줄
    print(f.readline(), end="") # 세번째 줄
    print(f.readline(), end="") # 네번째 줄(마지막) 
    print(f.readline(), end="") 
    # 읽어오는 데이터가 없으면 ""(빈문자열)을 가져옴
```

```python
# readline()으로 파일 내용 전체 읽어오기
with open("test1.txt", "r") as f :
    text = f.readline()
    print(text, end="")
    while text != "" : # 더 이상 읽어올 데이터가 없으면 빈문자열을 가져옴
        text = f.readline()
        print(text, end="")
```

```python
# readlines() : 각 라인 단위로 저장된 리스트
with open("test1.txt", "r") as f :
    text_li = f.readlines()
    # print(text_li, type(text_li))
    for line in text_li :
        print(line, end="")
    print()

print(text_li)
```

```python
# readlines()와 enumerate()함수 사용 => 줄번호와 내용 출력
with open("test1.txt", "r") as f :
    text_li = f.readlines()
    for idx, line in enumerate(text_li, start = 1) :
        # print(line, type(line)) 
        print(f"{idx}번째 줄 : {line}", end="")
```

```python
# enumerate()
li = ["가위", "바위", "보"]
for i in enumerate(li, start = 1) :
    print(i, type(i))
```

```python
# <텍스트 파일 입출력 실습>
# 학생 3명의 이름을 사용자로부터 입력받아, 리스트에 저장한 후
# 1) 아래와 같이 txt파일을 작성, 파일명은 student.txt
# <예시>
# 1번 학생 : 유리
# 2번 학생 : 철수
# 3번 학생 : 짱구

# 2) txt파일에 저장된 내용을 읽어와서 출력
```

```python
# 1) 학생 3명의 이름을 입력받아 리스트에 저장
# 방법 1) 리스트 메소드 사용  
# name_li = [] # 빈 리스트 생성

# # 3명의 학생의 이름을 입력받고 리스트에 저장
# for i in range(3) :
#     # i : 0, 1, 2
#     name = input(f"{i+1}번째 학생 이름 입력 : ")
#     name_li.append(name)

# 방법 2) 리스트 컴프리헨션
name_li = [input(f"{i+1}번째 학생 이름 입력 : ") for i in range(3)]

print(name_li)
```

```python
# 2) 텍스트 파일 생성 
with open("student.txt", "w") as f :
    # enumerate() 활용 => 학생 번호와 이름을 파일에 작성
    for idx, name in enumerate(name_li, start = 1) :
        # print(name, type(name))
        f.write(f"{idx}번 학생 : {name}\n")
        
print("student.txt파일 작성 완료!")
```

```python
# 3) 파일 읽어와서 출력
with open("student.txt", "r") as f :
    print(f.read()) # 내용 전체 읽어오기
```

**csv 파일 입출력**

```python
import csv
# csv 파일 작성

with open("student.csv", "w") as f :
    # 1) 쓰기 객체 생성
    writer = csv.writer(f)
    # 2) 파일 내용 작성
    # writerow() : 1행 작성, 리스트로 작성
    writer.writerow(["학번", "이름", "나이"])
    writer.writerow([1, "둘리", 10])
    writer.writerow([2, "도우너", 11])

print("student.csv파일 작성 완료!")
```

```python
# csv 파일 읽기 

with open("student.csv", "r") as f :
    # 1) 읽기 객체 생성
    reader = csv.reader(f)
    next(reader) # 헤더 건너뛰기(제목행)
    for i in reader :
        print(i) # 1행을 리스트로 가져옴
```

```python
# writerows() : 여러 행 작성(이중 리스트 사용)
with open("student2.csv", "w") as f :
    # 쓰기 객체 생성
    writer = csv.writer(f)
    writer.writerow(["학번", "이름", "나이"]) # 1행 작성
    writer.writerows([[1, "짱구", 5], [2, "도우너", 11], [3, "철수", 6]])

print("student2.csv파일 작성 완료!")
```

##문제2) 파일 내용 역순 출력하기

- 아래의 내용을 담고 있는 alphabet.txt 파일을 생성한 뒤, 이 파일을 읽어와서 **한 줄씩 역순(마지막 줄부터 첫 줄 순서)**으로 출력하는 프로그램을 작성.
- readlines() 메소드와 리스트 슬라이싱을 활용.

        <파일 내용 (alphabet.txt)>
        Apple
        Banana
        Cherry
        Dream

        <출력결과>
        Dream
        Cherry
        Banana
        Apple

```python
li = ["Apple", "Banana", "Cherry", "Dream"]
with open("alphabet.txt", "w") as f : 
    for i in li:
        f.write(i + "\n")

with open("alphabet.txt", "r") as f :
    lines = f.readlines()
for i in lines[::-1] :
    print(i, end="")
```

##문제3) 도서 목록 CSV 파일 생성 및 읽기

- 도서 정보를 담고 있는 2차원 리스트 book_list를 사용하여 books.csv 파일을 생성.

- csv.writer를 사용하며, 파일 생성 시 빈 줄이 생기지 않도록 newline='' 옵션을 반드시 사용.

- 파일을 생성한 후, 다시 읽어와서 도서명만 한 줄씩 출력.

        <데이터>
        book_list = [['번호', '도서명', '저자'], [1, '파이썬 기초', '강사A'], [2, '데이터 분석 실무', '강사B'], [3, '머신러닝 입문', '강사C']]

        <출력결과>
        --- 도서 목록 ---
        파이썬 기초
        데이터 분석 실무
        머신러닝 입문

```python
import csv
book_list = [['번호', '도서명', '저자'], [1, '파이썬 기초', '강사A'], [2, '데이터 분석 실무', '강사B'], [3, '머신러닝 입문', '강사C']]

with open("books.csv", "w", newline="") as f :
    writer = csv.writer(f)
    writer.writerows(book_list)

with open("books.csv", "r") as f :
    reader = csv.reader(f)
    next(reader)

    for i in reader :
        print(i[1])
```
