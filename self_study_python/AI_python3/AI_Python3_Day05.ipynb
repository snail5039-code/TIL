2. 필터링
 : 데이터 중 특정 조건을 만족하는 행만 골라내는 방법
필터링 방법 : 
	(1) 불린 인덱싱 : 조건식을 적용하면 각 행이 True/False
		로 표시되고, True인 행만 선택
	(2) isin() 메소드 : 특정 값 목록에 해당하는 행을 선택

1) 불린 인덱싱
- 데이터 프레임의 각 열은 시리즈(Series) 객체
- 시리즈에 조건식을 적용하면 각 원소가 True/False 값으로 표시되는
  시리즈가 만들어진다
- 이 시리즈를 데이터 프레임에 적용하면 True인 행들만 선택할 수 있다

2) 복합 조건
- & : and 연산자 -> 두 조건이 모두 True일때 선택
- | : or 연산자 -> 두 존건 중 하나라도 True일때 선택
- ~ : not 연산자 -> 조건이 False인 행을 선택
- 복합 조건을 사용할 때는 반드시 각 조건을 () 소괄호로 감싼다!

3) isin() 메소드
- 시리즈.isin(리스트) : 리스트 안에 있는 값과 일치하는 행을 선택
- 여러 값 중 하나라도 해당되는 행을 추출할 때는 불린 인덱싱보다
  간결하게 사용할 수 있다

**불린 인덱싱**

```python
import pandas as pd
import seaborn as sns

# tips 데이터에서 팁(tip)이 5달러 이상인 손님만 추출

df = sns.load_dataset("tips")

df.head()
```

```python
# 1) 조건식을 적용하면 True/False 값을 가진 시리즈가 반환된다
bol_tip = df["tip"] >= 5
# df['tip'] : 팁 금액 열(시리즈)
# >=5 : 5 이상인 경우 True, 아니면 False

print(type(bol_tip))
bol_tip
```

```python
# 2) True/False 시리즈를 데이터 프레임에 적용하여 조건에 맞는 행만 추출
df_tip = df.loc[bol_tip]
# df.loc[불린 시리즈] : True인 행만 선택하여 데이터 프레임으로 반환

print(f"전체 데이터 개수 : {len(df)}건")
print(f"팁이 5달러 이상인 데이터 개수 : {len(df_tip)}건")

df_tip.head()
```

```python
# tips 데이터에서 흡연자(smoker=='Yes')이면서 저녁 식사(time=='Dinner')
# 인 손님만 추출

# & : 두 조건이 모두 True인 행 선택
bol_sr = (df['smoker'] == 'Yes') & (df['time'] == 'Dinner')
# print(bol_sr) # 두 조건이 모두 True인 경우에만 True => 시리즈

df_result = df.loc[bol_sr]

print(f"흡연자이면서 저녁 식사인 손님 수 : {len(df_result)}명")
df_result.head()
```

**isin() 메소드**

```python
import pandas as pd
import seaborn as sns 

df = sns.load_dataset("tips")

# 요일별 데이터 개수
print(df['day'].value_counts())

df
```

```python
# 방법 1) 불린 인덱싱으로 주말 손님 추출
bol_weekend = (df['day'] == 'Sat') | (df['day'] == 'Sun')
df_week1 = df.loc[bol_weekend]

# 방법2) isin() 메소드 사용
isin_weekend = df["day"].isin(["Sat", "Sun"])
# print(isin_weekend) # day열 값이 Sat 또는 Sun이면 True인 시리즈
df_week2 = df.loc[isin_weekend]

print(f"주말 방문 손님 수 : {len(df_week1)}명")
print(f"주말 방문 손님 수 : {len(df_week2)}명")
# 추출할 값의 종류가 많을 수록 isin()이 훨씬 간결하다

df_week2.head()
```
