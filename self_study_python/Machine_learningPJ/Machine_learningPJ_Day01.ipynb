## 1. Streamlit

- 파이썬으로 애플리케이션을 빠르고 쉽게 만들 수 있는 오픈 소스 프레임워크

- HTML, CSS 지식 없이도 웹 앱을 만들 수 있음

### 1) 특징
- 간단한 문법: 파이썬 코드만으로 웹 앱 구현

- 무료 배포: Streamlit Community Cloud를 통해 무료로 배포 가능

## 2. Streamlit 기본 위젯 및 레이아웃

### 1) 입력 위젯

사용자로부터 입력을 받는 위젯

1. `st.text_input()`: 텍스트 입력
2. `st.number_input()`: 숫자 입력 (`min_value`, `max_value`로 범위 제한, `value`로 기본값 설정)
3. `st.selectbox()`: 여러 옵션 중 하나를 드롭다운으로 선택
4. `st.slider()`: 범위 내 값 선택
5. `st.checkbox()`: `True`/`False` 반환, 조건문과 함께 사용
6. `st.radio()`: 여러 옵션 중 하나만 선택
7. `st.button()`: 클릭 시 `True` 반환, 조건문과 함께 사용해 액션 실행

### 2) 레이아웃 구성

여러 요소를 배치하는 레이아웃 함수

1. `st.sidebar`: 왼쪽 고정 사이드바. 설정이나 입력 위젯 배치에 유용하며 `st.sidebar.위젯()` 형태로 사용
2. `st.columns()`: 여러 컬럼으로 나누어 배치. 숫자로 컬럼 수를 지정하고 `with`문과 함께 사용
3. `st.container()`: 여러 요소를 그룹화하고 레이아웃을 구조화
4. `st.expander()`: 접었다 펼 수 있는 섹션. 긴 내용을 숨기고 보여줄 때 사용

```python
import streamlit as st

# 페이지 제목 설정
st.title("안녕하세요!")

# 텍스트 표시
st.write("첫 번째 Streamlit 실행!")

# 마크다운 사용
st.markdown("""
## Streamlit
- Python으로 웹 앱을 만들 수 있는 프레임워크
- 간단한 문법으로 빠르게 실행 가능
""")
```

```python
import streamlit as st

# 제목 표시
st.title("가장 큰 제목")
st.header("큰 제목")
st.subheader("중간 제목")

# 텍스트 표시
st.write("일반 텍스트를 표시한다")
st.text("고정폭 텍스트를 표시")

# 코드 표시
st.code("print('실행~')", language="python")

# 성공/에러/정보 메시지
st.success("성공 메시지!")
st.error("에러 메시지!")
st.warning("경고 메시지!")
st.info("정보 메시지!")
```

```python
import streamlit as st

# 텍스트 입력
name = st.text_input("이름을 입력하세요", value="짱구")
st.write(f"안녕하세요, {name}님!")

# 숫자 입력
age = st.number_input(
    "나이를 입력하세요", min_value=0, max_value=120, value=20
)
st.write(f"당신의 나이는 {age}세입니다.")

# 선택 박스
option = st.selectbox(
    "좋아하는 프로그래밍 언어를 선택하세요",
    ["Python", "JavaScript", "Java", "C++", "기타"],
)
st.write(f"선택한 언어: {option}")

# 슬라이더
score = st.slider("점수를 선택하세요", 0, 100, 50)
st.write(f"선택한 점수: {score}점")

# 체크박스
agree = st.checkbox("이용약관에 동의합니다")
if agree:
    st.write("동의해 주셔서 감사합니다!")

# 라디오 버튼
gender = st.radio("성별을 선택하세요", ("남성", "여성", "기타"))
st.write(f"선택한 성별: {gender}")

# 버튼(조건문과 함께 사용)
if st.button("클릭하세요"):
    st.balloons()
    st.write("버튼이 클릭되었습니다!")
```

```python
import streamlit as st

# 사이드바
st.sidebar.title("사이드바")
st.sidebar.write("사이드바입니다!")

# 사이드바에 위젯 배치
name = st.sidebar.text_input("이름")
age = st.sidebar.slider("나이", 0, 100, 20)

# 메인 영역
st.title("메인 콘텐츠")
st.write(f"이름: {name}, 나이: {age}")

# 컬럼 레이아웃
col1, col2, col3 = st.columns(3)

with col1:
    st.header("첫 번째 컬럼")
    st.write("컬럼 내용")

with col2:
    st.header("두 번째 컬럼")
    st.write("컬럼 내용")

with col3:
    st.header("세 번째 컬럼")
    st.write("컬럼 내용")

# 컨테이너
with st.container():
    st.write("컨테이너입니다.")
    st.write("여러 요소를 그룹화할 수 있습니다.")

# 확장 가능한 섹션
with st.expander("자세한 정보 보기"):
    st.write("이 내용은 접었다 펼칠 수 있습니다.")
    st.write("긴 내용을 숨기고 보여줄 때 유용합니다.")
```
