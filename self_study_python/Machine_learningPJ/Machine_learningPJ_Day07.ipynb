Streamlit 어플리케이션 배포

1) 애플리케이션 배포
- 개발된 애플리케이션을 사용자가 실제로 사용할 수 있도록
   운영환경에 설치하고 진행하는 과정
- streamlit은 GitHub와 streamlit cloud를 활용하여 
  비교적 쉽게 배포 작업을 할 수 있다

2) Git과 GitHub
(1) Git
- 버전 관리 시스템
- 로컬 컴퓨터에서 작동하는 프로그램

(2) GitHub
- Git이 관리하는 버전 기록을 공유할 수 있는 웹 사이트
- 구글 드라이브처럼 개인 사용자가 이용할 수 있는 저장소를 제공
- 저장소에 파일을 업로드하거나 다운받을 수 있음

3) Git관련 명령어
- git init : 현재 폴더를 Git 저장소로 만들기
- git add . :  모든 파일 스테이징
- git commit -m "메시지" : 변경 사항 저장

# 원격 저장소 관련 명령
git remote add  origin [Github URL]
 : 현재 git저장소에 원격 저장소를 추가하는 명령어

git push origin main 
 : 로컬 변경사항 커밋을 원격에 업로드

git push -f origin main
 : 강제로 push
