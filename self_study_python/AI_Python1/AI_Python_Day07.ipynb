### 반복문(for문)

```python
# for 변수명 in 집합(시퀀스 자료형, 컬렉션,...) :
# (들여쓰기, tab) 반복할 문장

# 1) for문과 문자열
str1 = "안녕하세요"
print(str1, type(str1))

for i in str1 :
    print(i) # i : str1에 저장된 문자를 하나씩 저장함

# 2) for문과 컬렉션(리스트, 튜플, 세트, 딕셔너리)
# (1) 리스트와 for문
li = ["aa", 1, 5.3, True]
print(li, type(li))

for i in li :
    print(i, type(i))

# (2) 튜플과 for문
print("\n==============")
tu = "b", 10, 5.3 # 튜플의 패킹
print(tu, type(tu))
for i in tu :
    print(i, type(i))

# (3) 세트와 for문
print("\n==========")
set1 = {"봄", "여름", "가을", "겨울"} # 세트는 인덱스라는 개념이 없음(비시퀀스 자료형)
print(set1, type(set1))

for i in set1 :
    print(i)

# (4) 딕셔너리와 for문
print("\n==========")
di = {"a" : "apple", "b" : "banana"}
print(di, type(di))

for key in di :
    # print(key) # 반복할때마다 딕셔너리의 key를 가져옴
    # 딕셔너리[key] => key에 해당하는 value
    print(f"key : {key}, value : {di[key]}")

# +)
li = [(1, 2), (3, 4), (4, 5)]
print(li, type(li))

for i in li :
    print(i, type(i))
print("===============")

for tu in li :
    # print(tu, type(tu)) i, j = tu # 언패킹
    print(f"i = {i}, j = {j}")
print("=============")

for i, j in li :
    # 튜플의 언패킹 사용
    print(f"i = {i}, j = {j}")

# range()

print(range(1, 10, 2), type(range(1, 10, 2)))
li = list(range(1, 10, 2))
print(li)

# for와 range()

for i in range(1, 10) : # 1 ~ 9
    print(i, end = " ")
print()

for i in range(5) : # 0 ~ 4
    print(i, end = " ")
print()

for i in range(1, 10, 2) : # 1 ~ 9, 2씩 증가하는 정수
    print(i, end = " ")
print()

for i in range (10, 0, -1) : # 10 ~ 1, 1씩 감소하는 정수
    print(i, end = " ")
print()

# range()는 for문에서 반복횟수 지정할 때 사용
for i in range(4) : # 0, 1, 2, 3
    print("hi~!")

# <for문 실습>
# 1) for문을 사용하여 5 ~ 10 까지 출력
for i in range(5, 11) :
    print(i, end = " ")
print()
# 2) for문을 사용하여 1 ~ 20까지 정수 중 짝수만 출력
# <방법1>
for i in range(2, 21, 2) :
    print(i, end = " ")
print()
# <방법2> if문 사용
for i in range(1, 21) :
    if i % 2 == 0 :
        # i에 저장된 값이 짝수일 경우에만 출력
        print(i, end = " ")

# 3) li = [10, 5, 6, 8, 3]
# 위의 리스트와 for문을 사용하여 리스트 요소 중 짝수만 출력

li = [10, 5, 6, 8, 3]

for i in li :
    if i % 2 == 0 :
        print(i)

# 4) 구구단 3단 출력
# <출력예시>
# 3 x 1 = 3
# 3 x 2 = 6
# ...
# 3 x 9 = 27

for i in range(1, 10) :
    print(f"3 x {i} = {3 * i}")

for i in range(3) : # i : 0, 1, 2
    for j in range(3) : # j : 0, 1, 2
        print(f"i : {i}, j = {j}")

# 두자리 숫자와 이중 반복문
# 10, 11, 12, ... 19
# 20, 21, 22, ... 29
# ...             99
# - 두 자리 숫자에서 일의 자리 숫자가 0 ~ 9까지 나와야
#   십의 자리 숫자가 +1
# - 이중 반복문도 안쪽 for문(일의 자리 수)의 반복이 전부 끝나야
#   바깥 for문(십의 자리 수) 한번 끝난다

# <이중 for문 실습>
# 2단 ~ 9단 구구단 출력

# <출력 예시>
# 2 x 1 = 2
# 2 x 2 = 4
# ...
# 2 x 9 = 18
# 3 x 1 = 3
# ...
# 9 x 9 = 81

for i in range(2, 10) :
    print(f"======= {i}단 =======")
    for j in range(1, 10) :
        print(f"{i} x {j} = {i * j}")
```

### 기타제어문

```python
# 1) break문
# 1 ~ 10까지 중 5까지만 출력
for i in range(1, 11) :
    print(i)
    if i == 5 :
        break # 반복문 강제 종료

# 2) continue문 : 아래 코드를 실행하지 않고 다음 반복으로 넘어간다
# 1 ~ 10까지 중 5만 빽 ㅗ출력
print("================")

for i in range(1, 11) :
    if i == 5 :
        continue # 다음 반복으로 넘어간다
    print(i)

# 3) pass문
if True :
    # print("if문 안의 문장")
    pass
print("프로그램 종료!")

for i in range(3) :
    pass
print("반복 끝!")
```

- 문제1) 별 찍기(for문)
- 사용자로부터 숫자를 입력받아, 해당 숫자만큼 별을 찍는 프로그램을 작성하세요. 예를 들어, 입력이 5이면 다음과 같이 출력되어야 합니다

```python
num = int(input("별 찍을 개수를 입력하세요 : "))

for i in range(1, num + 1) :
    print("*" * i)

for i in range(1, num + 1) :
    for j in range(i) :
        print("*", end="")
    print()
```

- 문제3) 전체 구구단을 다음과 같은 모습으로 출력하는 프로그램 만들기
- 이중 for문으로 작성 이중 while문으로 작성

```python
for j in range(1, 10) :
    for dan in range(2, 10) :
        print(f"{dan} x {j} = {dan * j}", end="\t")
    print()

print("=================")

j = 1
while j <= 9 :
    dan = 2
    while dan <= 9 :
        print(f"{dan} x {j} = {dan * j}", end ="\t")
        dan += 1

    print()
    j += 1
```

- 문제4) 시험합격자 출력하기
- 학생들의 이름과 시험 결과는 다음과 같다
- 짱구 60점, 철수 90점, 유리 85점
- 학생들의 정보를 저장할 수 있는 딕셔너리 생성 후
- key는 이름, value는 점수 저장 반복문과 조건문을 사용하여 80점 이상인 학생들만 이름과 시험 점수를 출력

```python
di = {"짱구" : "60", "철수" : "90", "유리" : "85"}
print(di, type(di))
print("---- for문과 딕셔너리 사용 ----")
for i in di :
    print(f"이름 : {i}, 점수 : {di[i]}점")

print("---- for문과 if문 사용 합격자 출력 ----")
print("**** 합격자 명단 ****")
for i in di :
    if int(di[i]) >= 80 :
        print(f"이름 : {i}, 점수 : {di[i]}점")
```
