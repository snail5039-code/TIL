### 출력함수

```python
print("안녕하세요")
print(10)
print(10 + 20) # 계산 결과 출력
print("오늘은", 1, "일차 수업입니다") # , 콤마로 구분해서 쓰면 여러가지 값을 함계 출력
# print() 옵션
# 1) end 옵션 : 출력 내용의 마지막을 설정하는 옵션
print("반갑습니다", end = "!")
print("end옵션의 기본값은 줄바꿈입니다", end = "\n")
# \n : 줄바꿈
print("내 이름은 박의혁입니다")
# 2) sep 옵션 : 구분자 옵션, 출력 요소 사이마다 넣을 내용을 설정
print("오늘은", 1, "일차 수업입니다", sep = "@")

print("문자열은 따옴표로 감싸야 한다!")
print(4 + 3)
print("4 + 3")
print("4" + "3") # 문자열 + 문자열 => 연결 => 4라는 문자열과 3이라는 문자열 연결
```

### 변수

```python
# 변수명 = 값
# = : 대입 연산자

# 변수의 생성
a = 10 # a라는 변수에 10이라는 숫자를 저장
b = "abc" # b라는 변수에 "abc"라는 문자열 값을 저장

# 변수에 저장된 값을 사용
#  : 저장된 값을 사용할 곳에 변수명을 작성
print(a, type(a)) # a에 저장된 값 10을 출력
print(b, type(b)) # b에 저장된 값 abc를 출력

# type() : () 소괄호 안에 있는 값의 타입을 알려주는 함수
# int : 정수
# str : 문자열
# float : 실수

# <실습> 이름(문자열), 나이(정수), 취미(문자열)를 각각 저장하는 변수를 생성하고
# 변수에 저장된 값 출력

# <출력예시>
# 내 이름은 000입니다
# 나는 00살이고 취미는 0000입니다

# 1) 이름, 나이, 취미를 저장하는 변수 생성
name = "박의혁"
age = 30
hobby = "운동"

print(name, type(name))
print(age, type(age))
print(hobby, type(hobby))

print("내 이름은 " + name + "입니다") # 문자열1 + 문자열2 + 문자열3 => 문자열 연결
print("나는 ", age, "살이고 취미는 " + hobby + "입니다")
```
