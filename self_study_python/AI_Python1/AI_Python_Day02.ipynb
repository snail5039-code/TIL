### 기본 자료형

```python
# 1) 정수형(int)
print(10, type(10))

# 2) 실수형(float)
print(3.14, type(3.14))
print(1.0, type(1.0)) # 소수점이 붙으면 무조건 실수

# 3) 논리형(bool) : True, False
print(True, type(True))
print(False, type(False))

# 4) 문자열(str)
print("Hello", type("Hello"))
print("10", type("10")) # 따옴표가 붙어 있으면 문자열
print("10 + 20", type("10 + 20"))
print("10" + "20") # 문자열 + 문자열 => 연결
```

### 형변환

```python
# 1) 정수형 형변환 : int()

# (1) 실수 => 정수
print(15.19, type(15.19))
print(int(15.19), type(int(15.19))) # 15.19 => 15 (소수부는 버림)

# (2) 문자열 => 정수
print("15", type("15"))
print(int("15"), type(int("15"))) # "15" => 15

# print(int("15.19")) # 문자열 실수는 정수로 형변환 불가능!
print(float("15.19"), type(float("15.19"))) # "15.19" => 15.19
print(int(float("15.19"))) # "15.19" => 15.19 => 15

# print(int("가")) # 문자열 자료형을 정수로 변환할 때는 변환 가능한 값을 넣어줘야함

# (3) 논리형 => 정수
#  True => 1
# False => 0
print(int(True)) # 1
print(int(False)) # 0

print(True + 10) # 1 + 10
print(False * 10) # 0 * 10

# 2) 실수형 형변환 : float()

# (1) 정수 => 실수
print(float(5), type(float(5))) # 5 => 5.0

#  (2) 문자열 => 실수
print(float("3.14"), type(float("3.14"))) # "3.14" => 3.14
print(float("5"), type(float("5"))) # 정수형 문자열은 실수로 형변환 가능

# (3) 논리형 => 실수
print(float(True), type(float(True))) # True => 1.0
print(float(False), type(float(False))) # False => 0.0

# 3) 논리형 형변환 : bool()

print(bool(0), type(bool(0)))
print(bool(123), type(bool(123))) # 0, 0.0을 제외한 숫자는 모두 True로 변환됨
print(bool(""), type(bool(""))) # 빈문자열 => False
print(bool([]), type(bool([]))) # 빈괄호 => False

# 4) 문자열 형변환 : str()

print(str(10), type(str(10))) # 10 => "10"
print(str(3.14), type(str(3.14))) # 3.14 => "3.14"
print(str(True), type(str(True))) # True => "True"
```

### 시퀀스 자료형

```python
#  1) 인덱싱

# idx   01234
data = "hello"
#       -54321

print(data, type(data))
print(data[0], type(data[0])) # h
print(data[-5]) # h

# hlo, 홀수번째 문자들 추출
print(data[0], data[2], data[4]) # 인덱스 사용
print(data[-5] + data[-3] + data[-1]) # 마이너스 인덱스 사용

# <실습> hello를 역순으로 출력 => olleh
# 인덱스 사용
print(data[4] + data[3] + data[2] + data[1] + data[0])
# 마이너스 인덱스 사용
print(data[-1] + data[-2] + data[-3] + data[-4] + data[-5])

# 2) 슬라이싱
# 변수명[시작:종료:간격]
#       01234
data = "hello"
#      -54321

print(data[0:3:1]) # idx 0부터 2까지 1씩 가격으로 슬라이싱
print(data[0:3]) # 간격의 기본값은 1
print(data[:3]) # 0번 인덱스 번호 생략 => 첫문자부터 자를 경우
print(data[0:5:2]) # idx 0부터 4까지 2씩 간격으로 슬라이싱
print(data[:5:2])
print(data[::2]) # 마지막까지 자를 경우 => 종료 인덱스 생략 가능

# +) 간격에 음수 값을 넣으면 뒤에서부터 슬라이싱
print(data[::-1]) # 처음부터 끝까지 뒤에서부터 1씩(-1) 슬라이싱

# 3) 시퀀스 연산자

# + : 문자열 연결
data1 = "hello"
data2 = "python"

print(data1 + data2) # 문자열 + 문자열
# print(data1 + 10) # 문자열 + 다른 자료형 => 오류 발생!

# * : 문자열 반복
print("hello" * 3) # 'hello' 3번 반복
# print("hello" * 3.5)
# print("hello" * 3.0) # 3.0도 float타입!
```
