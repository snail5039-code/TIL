### 형식을 갖춘 문자열
### 1)%연산자

```python
print("%d" %10)
print("%f" %3.14)
print("%.2f" %3.14) # .2f : 실수형 소수점 2자리까지 출력
print("%s" %"python")
print("%s" %True)

print("내 이름은 %s이고, 나이는 %d살 입니다" %("짱구", 5))
```

### 2) str.format()

```python
num1 = 10
num2 = 20

# 10 + 20 = 30 를 출력
#  1) print() 사용 출력
print(num1, "+", num2, "=", num1 + num2)

# 2) %연산자, 형식 지정자 사용
print("%d + %d = %d" %(num1, num2, num1 + num2))

# 3) str.format() 사용
print("{} + {} = {}".format(num1, num2, num1 + num2))

print("{0} + {2} = {1}".format(num1, num2, num1 + num2))
```

### 3) f-string

```python
hour = 12
print(f"나는 {hour}시에 밥을 먹을거야!")

percent = 10.56
print(f"작업 진행률 : {percent : .1f}%") # {실수형 값 : .1f} 소수점 첫번째 자리까지 출력

# 형식을 갖춘 문자열 => 문자열 값

name = "짱구"
age = 5

str1 = "이름 : %s, 나이 %d살" %(name, age)
str2 = "이름 : {}, 나이 {}살" .format(name, age)
str3 = f"이름 : {name}, 나이 {age}살"

print(str1, type(str1))
print(str2, type(str2))
print(str3, type(str3))

# <형식을 갖춘 문자열 실습>
# 사용자로부터 이름(문자열), 나이(정수), 생년월일(실수)을 입력받아,
# 입력받은 이름, 나이, 생년월일 출력
# <출력예시>
# 이름 : 짱구
# 나이 : 20살
# 생년월일 : 2006.1231
# %연산자, str.format(), f-string 각각 출력

# 1) 사용자로부터 입력받기
name = input("이름 : ")
age = int(input("나이 : "))
birt = float(input("생년월일 : "))

# 2) 결과 출력
print("===== % 연산자 =====")
print("이름 : %s\n나이 : %d\n생년월일 : %.4f" %(name, age, birt))

print("===== str.format() =====")
print("이름 : {}\n나이 : {}\n생년월일 : {}" .format(name, age, birt))

print("===== f-string() =====")
print(f"이름 : {name}\n나이 : {age}\n생년월일 : {birt}")
```

### 연산자
### 1)최우선 연산자

```python
print(10 + 2 * 3) # 10 + 6 => 16
print((10 + 2) * 3) # 12 * 3 => 36

# 1) 부호 연산자
print(+10) # 양의 정수
print(-10) # 음의 정수
print(+3.14) # 양의 실수
print(-3.14) # 음의 실수

# 2) not 연산자
print(not True) # False
print(not False) # True
```

### 3) 산술 연산자

```python
# 1) + : 덧셈
print(10 + 3)

# 2) - : 뺄셈
print(10 - 3)

# 3) * : 곱셈
print(10 * 3)

# 4) / : 나눗셈 => 결과는 float
print(10 / 3)
print(10 / 2) # 5.0

# +) 나눗셈 연산 주의사항
# => 0으로 나눌 수 없음
# print(10 / 0)

# 5) 몫 연산자 : //
print(10 // 3) # 10 / 3 => 몫 : 3, 나머지 : 1 => 몫 3

# 6) 나머지 연산자 : %
print(10 % 3) # 나머지 1

# +) 나머지 연산자 활용
# (1) 짝수, 홀수 구분
# 짝수 % 2 = 0
# 홀수 % 2 = 1
print(12 % 2) # 0
print(9 % 2) # 1

# (2) 배수 관계 확인
# 3의 배수(3, 6, 9, ...) % 3 => 0
print(15 % 3) # 0, 15는 3의 배수

# 7) 거듭제곱 연산자 : **
print(10 ** 2) # 10 * 10
print(10 ** 3) # 10 * 10 * 10

# <산술 연산 실습> 코인 노래방 프로그램
# 코인 노래방 1곡당 가격이 500원이다
# 사용자로부터 투입할 금액을 입력받아,
# 투입한 금액으로 부를 수 있는 최대 곡의 개수와 거스름돈 출력

# <출력 예시>
# 금액 입력 : 2300
# 최대 부를 수 있는 곡의 개수는 4곡이고, 거스름돈은 300원 입니다

# 1) 사용자로부터 금액 입력받기
money = int(input("금액 입력 : "))

# 2) 최대 곡의 개수, 거스름돈 개수
# 최대 곡의 개수 = 금액 // 1곡당 가격
# 거스름돈 = 금액 % 1곡당 가격

song = money // 500
change = money % 500
# 산술연산자가 대입 연산자보다 우선 순위가 높음
# => 산술 연산의 결과를 변수에 저장

# 3) 결과 출력
print(f"최대 부를 수 있는 곡의 개수는 {song}곡이고, 거스름돈은 {change}원입니다")
```

- 문제1) 사용자로부터 상품명(문자열), 재고량(정수형), 제작날짜(mm.dd 실수형)을 입력받아 출력하기 %연산자, str.format, f-string 3개 전부 활용할 것! <결과화면>

- 문제4) 산술연산 결과 출력하기

```python
num1 = int(input("첫번째 정수 입력 : "))
num2 = int(input("두번째 정수 입력 : "))

print("==== 산술 연산 결과 ====")
print(f"{num1} + {num2} = {num1 + num2}")
print(f"{num1} - {num2} = {num1 - num2}")
print(f"{num1} * {num2} = {num1 * num2}")
print(f"{num1} / {num2} = {num1 / num2 : .2f}")
print(f"{num1} // {num2} = {num1 // num2}")
print(f"{num1} % {num2} = {num1 % num2}")
print(f"{num1} ** {num2} = {num1 ** num2}")
```
