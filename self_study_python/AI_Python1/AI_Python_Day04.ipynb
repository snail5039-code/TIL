### 연산자

```python
# 쉬프트 연산자
print(10 << 2)
print(10 >> 2)
```

### 1) 관계 연산자

```python
print(10 > 2) # 10은 2보다 크니???? => 응(True)
print(10 < 2) # 10은 2보다 작니???? => 아니(False)

print(10 >= 3) # 10은 3이상이니???? => True
print(10 <= 3) # 10은 3이하니???? => False

print(10 == 10) # 10은 10이랑 같니???? => True
print(10 == 20) # 10은 20이랑 같니???? => False

print(10 != 20) # 10은 20이랑 다르니???? => True

# +) 문자열 비교
print("안녕" == "안녕") # True
print("안녕" != "hello") # True
```

### 2) 논리 연산자

```python
# 1) and : 둘 다 True라면 True
print(True and True) # True
print(True and False) # False

# 2) or : 둘 중 하나라도 True 라면 True
print(True or True) # True
print(True or False) # True
print(False or False) # False

print(10 > 5 and 0 >= 0)
# 관계 연산잔가 논리 연산자보다 우선 순위가 높다
# 관계 True and True
# 논리    True

print(not(10 > 6))
# 관계 : not True
# 단항 : False
# 최우선 연산자() 안에 있는 관계 연산 => 당항 연산

print(10 % 2 == 0) # 10은 짝수니????
# 산술 0 == 0
# 관계  True

# <조건식 실습>
# 사용자로부터 2개의 정수를 입력받아
# 1) 입력받은 2개의 정수가 하나라도 양수라면 True 출력
#    둘다 양수가 아니라면 False출력

# 2) 입력받은 2개의 정수가 모두 짝수라면 True 출력
#    하나라도 짝수가 아니라면 False

# (1) 사용자로부터 2개의 정수 입력
num1 = int(input("첫번째 정수 입력 : "))
num2 = int(input("두번째 정수 입력 : "))

# (2) 하나라도 양수라면 True
# 정수 > 0 or 정수2 > 0
print(num1 > 0 or num2 > 0)

# (3) 모두 짝수라면 True
# 짝수인지 확인 : 정수 %2 == 0
print(num1 % 2 == 0 and num2 % 2 == 0)
```

### 3) 삼항연산자

```python
# 참일때 if 조건식 else 거짓일때

print("True 는 참" if True else "False는 거짓")
print("True 는 참" if False else "False는 거짓")
print("10은 5보다 크다" if 10 > 5 else "10은 5보다 크지 않다")

# <삼항 연산자 예제>
# 사용자로부터 정수를 입력받아,
# 입력받은 정수가 짝수라면 "짝수입니다" 출력
# 그게 아니라면 "홀수입니다" 출력

# 1) 사용자로부터 정수 입력
num = int(input("정수 입력 : "))

# 2) 짝수라면 => "짝수입니다" 출력, 그게 아니라면 "홀수입니다" 출력
result = "짝수입니다" if num % 2 == 0 else "홀수 입니다"
# 대입 연산자는 우선순위가 가장 낮음 => 연산 결과를 변수에 저장
print(f"{num}은/는 {result}")

# <연산자 실습1>
# 정수를 입력받아, 입력받은 정수가 3의 배수라면
# 입력받은 정수를 그대로 출력, 아니라면 0을 출력

num = int(input("정수 입력 : "))
result = f"{num}" if num % 3 == 0 else "0"
print(f"{result}")

# <연산자 실습2>
# 사용자로부터 python 점수와 java 점수를 정수로 입력받아,
# python 점수가 80점 이상하고 java 점수가 60점 이상이라면 "시험 합격" 출력
# 그게 아니라면 "시험 불합격" 출력

python = int(input("파이썬 점수 입력 : "))
java = int(input("자바 점수 입력 : "))

result = "시험 합격" if python >= 80 and java >= 60 else "시험 불합격"
print(f"파이썬 점수 {python}, 자바 점수 {java}점으로 시험 결과는 {result}입니다")

# <연산자 실습3>
# 두개의 정수를 입력받아,
# 두 정수의 차이를 절대값으로 출력
# * 절대값 : 10의 절대값 => 10
#             -10의 절대값 => 10

num1 = int(input("첫번째 정수 입력 : "))
num2 = int(input("두번째 정수 입력 : "))

# 숫자(큰수) - 숫자(작은수) => 양수

result = num1 - num2 if num1 > num2 else num2 - num1

print(f"{num1}와 {num2}의 차이는 {result}")
```

### 복합 대입 연산자

```python
num1 = 1
num1 += 10 # num1 + 10 => num1에 저장된 값 10 증가
print(num1)

num2 = 10
num2 -= 5 # num2 = num2 - 5에 저장된 값 5 감소
print(num2)

num3 = 10
num4 = 2
num3 *= num4 # num3 = num3(10) * num4(2)
print(num3)
```

### 시퀀스 연산자

```python
str1 = "시퀀스 자료형은!"
str2 = "값이 연속적으로 이어진 자료형"

print(str1 + str2) # 문자열 + 문자열 => 연결
print(str1 * 3) # 문자열 * 3 => 3번 반복

# 시퀀스 자료형과 복합 대입 연산자 사용
alpha = "a"
alpha += "b" # alpha("a") + "b"
print(alpha)

alpha *= 3
print(alpha)
```

### 리스트(list)

```python
# idx 0  1  2  3  4
li = [1, 2, 3, 4, 5] # 1 ~ 5가 저장된 리스트
#    -5 -4 -3 -2 -1

print(li, type(li))
# 1) 리스트는 시퀀스 자료형 => 인덱스 번호 사용 가능
print(li[1]) # li의 idx 1번 위치 값 => 2
print(li[1:3]) # idx 1 ~ 2까지 슬라이싱
print(li[1:2], type(li[1:2]))
```

- 문제4) 시험 합격 여부 확인 프로그램 만들기 사용자로부터 수학, 영어, 국어 점수 입력받기
- 입력받은 점수들의 평균이 70점 이상이라면
### '합격입니다' 출력
- 그게 아니라면 "불합격입니다" 출력

```python
num1 = float(input("수학 점수 입력 : "))
num2 = float(input("영어 점수 입력 : "))
num3 = float(input("국어 점수 입력 : "))

averge = (num1 + num2 + num3) / 3

result = "합격입니다" if averge >= 70 else "불합격입니다"

print(f"평군 점수는 {averge : .2f}이고 시험 결과는 {result}!")
```
