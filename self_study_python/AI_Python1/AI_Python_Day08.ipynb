### 리스트(list)

```python
# 리스트 생성
li = [1, 1, 3.14, 'abc']
print(li, type(li))

# 비어있는 리스트 생성
li2 = []
li3 = list()

print(li2)
print(li3)

# 리스트 요소에 접근
li = [10, 20, 30, 40, 50]
print(li, type(li))

print(li[4], type(li[4]))
print(li[1:4], type(li[1:4]))

# 리스트 요소 값 수정
li = [10, 20, 30, 40, 50]
print(li)
print(li[0])
# 리스트명[idx] = 값
li[0] = 100 # idx 0번 위치 값 100으로 수정
print(li)

# +) 리스트에서 사용 가능한 함수
li = [1, 3, 5, 7, 9]

print(f"li의 길이 : {len(li)}") # 요소의 개수
print(f"li의 최대값 : {max(li)}") # 리스트의 요소가 숫자만 있을 때 가능
print(f"li의 최소값 : {min(li)}")
print(f"li의 합계 : {sum(li)}")
print(f"li의 평균 : {sum(li) / len(li)}") # 평균 = 합계 / 개수

# 리스트 메소드
li = [1, 3, 5, 7, 9]
print(li)

# 1) 추가 : append, insert
li.append(10) # 마지막 요소에 10을 추가
print(li)
li.append(11)
print(li)

# insert(idx, 값)
li.insert(1, 2) # idx 1번 위치에 2를 추가
print(li)

# 2) 연결 : extend
li1 = [10, 30]
li2 = [50, 70, 90]
print(li1 + li2) # 리스트 + 리스트 => 하나의 리스트로 연결
print(li1)
li1.extend(li2) # li1의 원본이 변경됨
print(li1)

# 3) 삭제 : remove, pop
li = [1, 2, 3, 5, 7, 1]
print(li)
li.remove(1) # li에서 1을 삭제 => 중복값이 있을 때, 먼저 나온 값이 삭제
print(li)
li.remove(1)
print(li)
# li.remove(1) # 삭제하려는 값이 없으면 오류 발생

# pop(idx)
print(f"삭제할 값 : {li.pop(0)}")
print(li)
li.pop()
print(li) # 인덱스 번호를 지정하지 않으면 마지막 요소가 삭제됨

# 4) 정렬 : sort(), reverse()
li = [1, 3, 2, 5, 4, 6]
print(li)
li.sort() # 리스트 오름차순 정렬
print(li)

li.sort(reverse = True) # 리스트 내림차순 정렬
print(li)

li2 = ["a", "b", 1, 5, 3]
# li2.sort() # 리스트의 요소가 숫자끼리 혹은 문자열끼리 있어야 정렬 가능
print(li2)

li2.reverse() # 역순으로 정렬
print(li2)

# +) in, not in 연산자
li3 = ["봄", "여름", "가을", "겨울"]
print("봄" in li3) # 봄이 li3에 있니??? => True
print("사자" not in li3) # 사자가 li3에 없니??? => True

# 5) clear() : 리스트 전체 요소 삭제 => 빈 리스트
li3.clear()
print(li3)

# del 저장공간명 : 변수 삭제, 컬렉션 ...
del li3
```

### 튜플(tuple)

```python
# 튜플의 생성
tu1 = 10, 20, 30, 40 # 튜플의 패킹
print(tu1, type(tu1))

tu2 = 10,
print(tu2, type(tu2)) # 요소가 1개일때는 , (콤마) 작성

# 비어있는 튜플 생성
tu3 = ()
tu4 = tuple()
print(tu3)
print(tu4)

# 튜플의 요소 접근
tu = 10, 20, 30, 40
print(tu, type(tu))
# 인덱스 번호 사용 가능
print(tu[0])
print(tu[:3])

# 튜플의 요소 수정
# tu[0] = 100 # 튜플은 수정이 안됨!

# 튜플의 값 수정시 리스트로 형변환하여 수정함
li = list(tu) # 튜플을 리스트로 형변환
print(li, type(li)) # 튜플의 요소를 가진 리스트
li[0] = 100 # 리스트는 수정 가능함
print(li)
tu = tuple(li) # 리스트를 다시 튜플로 형변환
print(tu, type(tu))

# 튜플은 추가, 삭제 안됨
print(tu)

# tu.append(50) # 추가 불가

# tu.remove(100) # 삭제 불가

# 튜플의 메소드
tu = (1, 1, 2, 3)
print(tu, type(tu))
print(tu.count(1)) # tu에서 1의 개수
print(tu.index(1)) # tu에서 1의 위치 인덱스 번호(중복이 있다면 맨 앞의 인덱스 번호)
```

### 3) 세트(set)

```python
# 세트의 생성

se1 = {1, 1, 2, 3}
print(se1, type(se1)) # 세트는 중복을 허용하지 않음

# 비어있는 세트 생성
# se2 = {} # dict
se2 = set()

print(se2, type(se2))

# 세트의 메소드
set1 = {"사과", "참외", "파인애플", "멜론"}
print(set1, type(set1))

set1.add("포도")
print(set1)
set1.add("사과") # 중복된 값은 추가되지 않음
print(set1)

set1.remove("멜론")
print(set1)
# set1.remove("망고") # 없는 값을 넣으면 오류 발생
set1.discard("망고")
print(set1)

print(f"삭제할 값 : {set1.pop()}")
print(set1)
```

### 4) 딕셔너리(dict)

```python
# 딕셔너리 생성
cities = {"korea" : "서울", "japan" : "일본", "china" : "베이징"}
print(cities, type(cities))

# 비어있는 딕셔너리 생성
di1 = {}
di2 = dict()
print(di1)
print(di2)

# 1) 딕셔너리 값 수정 : 딕셔너리명[key(기존)] = 값
# * 리스트 요소 수정 : 리스트명[idx] = 값
cities["japan"] = "도쿄"
print(cities)

# 2) 딕셔너리 요소 추가(key와 value 추가)
# : 딕셔너리[key(기존에 없는)] = 값
cities['germany'] = "베를린"
print(cities)

# 3) value에 접근
print(cities['korea'])
print(cities.get("korea"))

# 4) key, value 삭제
# (1) 딕셔너리.pop(key)
print(f"삭제할 값 : {cities.pop("china")}")
print(cities)

# (2) del 키워드 사용
del cities["japan"]
print(cities)

# 딕셔너리 메소드
di = {"a" : "apple", "b" : "banana", "c" : "camera"}
print(di, type(di))

# 1) keys()
print(di.keys())
for key in di.keys() :
    print(key)
print("============")

# 2) values()
print(di.values())
for value in di.values() :
    print(value)
print("============")

# 3) items()
print(di.items())
for item in di.items() :
    print(item, type(item))
print("===== 튜플의 언패킹 활용 ======")
for key, value in di.items() : # 튜플의 언패킹
    print(f"key : {key}, value : {value}")

# <컬렉션 메소드 실습>
# 1) 빈 리스트를 생성하고 for문을 리스트에 1 ~ 10까지 추가 후 출력

li = []
print(li, type(li))

for i in range(1, 11) :
    # print(i)
    li.append(i)
# (3) 반복 종류 후 확인
print(li)

# 2) 음료수 자판기 프로그램
# 버튼 1번은 '콜라' 2번은 '사이다' 3번은 '환타' 이고
# 사용자로 부터 버튼 번호를 입력받아
# 버튼 번호에 해당하는 음료수 명 출력
# 만약 그 외 번호를 입력시에는 '제공하지 않는 메뉴' 출력

# (1) 메뉴 딕셔너리
di = {1 : "콜라", 2 : "사이다", 3 : "환타"}

# (2) 사용자로부터 번호 입력받기
button = int(input("음료수 버튼 번호를 입력(1~3) : "))

# (3) 버튼번호에 해당하는 음료수 출력
if button in di.keys() :
    # print(di[button])
    print(di.get(button))
else :
    print("제공하지 않는 메뉴")
```

```python
di = {"둘리" : "010-1234-5678", "또치" : "010-8765-4321", "도우너" : "010-5555-5555"}
print("*** 전화번호부 프로그램 실행 ***")
print("1. 연락처 검색")
print("2. 연락처 삭제")
print("3. 연락처 추가")
print("4. 전체 연락처 확인")
print("5. 프로그램 종료")
num = int(input("실행하고 싶은 기능 입력(1 ~ 5) : "))

if num == 1 :
    name = input("검색할 이름 입력 : ")
    if name in di :
        print(f"{name}의 연락처는 {di[name]}")
    else :
        print("연락처에 없는 이름입니다")
if num == 2 :
    name = input("삭제할 이름 입력 : ")
    if name in di :
        del di[name]
        print(f"{name}의 연락처를 삭제합니다")
        print()
        print("<삭제 후 결과>")
        print("---- 연락처 목록 ----")
        for k, v in di.items() :
            print(f"{k}, {v}")
    else :
        print("연락처에 없는 이름입니다")
if num == 3 :
    name = input("추가할 이름 입력 : ")
    num1 = input("추가할 연락처 입력 : ")
    di[name] = num1
    print(f"{name}의 연락처가 추가되었습니다")
    print("추가 후 결과")
    print("---- 연락처 목록 ----")
    for k, v in di.items() :
        print(f"{k}, {v}")
if num == 4 :
    print("---- 연락처 목록 ----")
    for k, v in di.items() :
        print(f"{k}, {v}")
if num == 5 :
    print("---- 프로그램을 종료합니다! ----")
```

- 문제 6) 수학 여행을 어디로 갈지 결정하기 위해 학생들이 희망하는 모든 수학여행을 조사하려고 한다.
- 학생들의 인원수와 장소를 입력받아 다음과 같이 출력하는 프로그램 만들기(set 사용할 것)

```python
num = int(input("학생들의 인원수 입력 : "))
str1 = input("1번째 학생의 희망하는 여행지 >> ")
str2 = input("2번째 학생의 희망하는 여행지 >> ")
str3 = input("3번째 학생의 희망하는 여행지 >> ")

se = {str1, str2, str3}

print(f"조사된 수학 여징지는 {se}입니다")

num = int(input("학생들의 인원수 입력 : "))
se = set()

for i in range(num) :
    place = input(f"{i+1}번째 학생의 희망하는 여행지 >> ")
    se.add(place)
print(f"조사된 수학 여행지는 {se}입니다")
```
