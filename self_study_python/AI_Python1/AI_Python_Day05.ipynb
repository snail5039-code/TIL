### 연산자
### 관계 연산자(비교 연산자)
### , <, >=, <=, ==, !=
### 논리 연산자
- A and B : 둘다 True라면 True
- A or B : 둘 중 하나라도 True라면 True
### 조건식
- : 결과가 True, False로 나오는 것
- ex) 관계 연산자, 논리 연산자, ...
### 삼항 연산자
- 참일때 if 조건식 else 거짓일 때
### 대입 연산자(=)
- : 복합대입 산술 연산(산술 연산과 대입 연산을 동시에 사용)
- +=, -=, *=, /=, //=, %/, **=
### 컬렉션
- : 기본 자료형과 달리 여러가지 값을 하나로 묶어서 사용하는 "자료형"
### list 리스트	[]
### tuple 튜플	()
### set	세트	{}
- dict 딕셔너리 {key:value}
### 리스트(list)
- []대괄호로 값들을 묶어주고,(콤마)로 값들을 구분
- 중복값을 허용, 모든 자료형을 허용(리스트도 포함)
- 시퀀스 자료형(인덱스 번호 사용 가능)
- 추가, 수정, 삭제가 언제나 가능 # 나중에 메소드라는 챕터에서 확인
### 튜플(tuple)
- () 소괄호로 묶어주고, (콤마)로 값들을 구분
- 중복값을 허용, 모든 자료형을 허용
### 시퀀스 자료형
- 한번 생성되고 나면 변경이 불가능(추가, 수정, 삭제 불가능)
### => 읽기 전용 리스트
### 세트(set)
- {} 중괄호로 묶어주고 ,(콤마)로 값들을 구분
### 중복값을 허용하지 않음
- 비시퀀스 자료형(인덱스라는 개념잉 벗음)
### => 인덱스 번호 사용 불가
### 순서라는 개념이 없음
### 딕셔너리(dict)
- {key:value}, key와 value가 한쌍으로 관리
- 비시퀀스 자료형(인덱스 번호 사용 불가)
### key를 인덱스 번호 처럼 사용
- 딕셔너리명[key] : key에 해당하는 value
- key는 중복값을 허용하지 않음, value는 중복값을 허용함
- value는 모든 자료형 허용 : 정수, 실수, 문자열, 리스트, 튜플, ...
- ex) {키1 : 값1, 키2 : 값2, ...}
### 컬렉션
### 1)리스트(list)

```python
# idx  0  1  2  3  4
li1 = [1, 2, 3, 4, 5]

print(li1, type(li1))

# 1) 리스트는 시퀀스 자료형
# 인덱싱 : 변수명[idx]
print(li1[1], type(li1[1])) # 인덱싱을 하면 해당 값의 자료형
# 슬라이싱 : 변수명[시작:종료:간격]
print(li1[1:3], type(li1[1:3])) # 슬라이싱을 하면 리스트
print(li1[1:2], type(li1[1:2]))

# 2) 중복값을 허용, 모든 자료형을 허용
li2 = [1, 1, 1, 0.5, True, "안녕"]
print(li2, type(li2))

# 3) in 연산자, not in 연산자 => 맞으면 True, 틀리면 False
# a in 리스트 : a라는 값이 리스트에 있니????
# a not in 리스트 : a라는 값이 리스트에 없니????
print(li2, type(li2))
print(0.5 in li2) # 0.5라는 값이 li2안에 있니??? => True
print(100 not in li2) # 100이라는 값이 li2안에 없니??? => True

# +) 이중 리스트(중첩 리스트) : 리스트 안에 다른 리스트가 있는 구조
#      0  1  2         3
li3 = [1, 2, 3, ["1", "2", "3"]]
#                 0    1    2

# 내부 리스트 : 다른 리스트 안에 있는 리스트
# 외부 리스트 : 내부 리스트를 요소로 가진 리스트
print(li3[3], type(li3[3]))
# 내부 리스트 안에 있는 "2" 가져오기
print(li3[3][1], type(li3[3][1]))
```

### 2) 튜플(tuple)

```python
#      0  1   2     3     4
tu1 = (1, 2, "가", 1.1, True) # 모든 자료형, 중복값 허용
print(tu1, type(tu1))

# 1) 튜플은 시퀀스 자료형
print(tu1[3], type(tu1[3]))
print(tu1[::2],type(tu1[::2]))

# 2) 튜플의 특징 - 패킹, 언패킹
# 패킹(packing) : 여러 데이터를 튜플로 묶는 것
tu2 = 1, 2, 3
print(tu2, type(tu2))

# 언패킹(unpacking) : 튜플의 각 요소를 여러개의 변수에 할당(저장)
num1, num2, num3 = tu2 # (1, 2, 3)
print(num1) # 1
print(num2) # 2
print(num3) # 3

# +) 변수에 저장된 값 서로 바꿀 때
x = 10
y = 20
print("=== 변경 전 ===")
print(f"x = {x}")
print(f"y = {y}")

# 튜플의 패킹, 언패킹 활용 => x, y에 저장된 값 변경
x, y = y, x # y, x => 20, 10 => (20, 10) : 튜플의 패킹
# x, y = (20, 10) : 튜플의 언패킹
print("=== 변경 후 ===")
print(f"x = {x}")
print(f"y = {y}")

# +) 리스트와 튜플은 시퀀스 자료형
# 시퀀스 연산자
# 1) + : 연결
print([1, 2, 3] + [4, 5, 6]) # 리스트 + 리스트 => 하나의 리스트로 연결
print((1, 2, 3) + (4, 5, 6)) # 튜플 + 튜플 => 하나의 튜플로 연결

# 2) * : 반복
print([1, 2, 3] * 3)
print((4, 5, 6) * 2)
print("=== 변경 전 ===")
print(f"x = {x}")
print(f"y = {y}")

# +) 튜플 주의사항! = 튜플의 요소가 1개인 경우
# : 뒤에 반드시,(콤마)를 표시
tu = (2,)
print(tu, type(tu))
```

### 3) 세트(set)

```python
# 1) 세트는 중복값을 허용하지 않음
set1 = {1, 1, 2, "가", 3.5, 1, True} # 세트는 True를 1이나 1.0과 같다고 판단
print(set1, type(set1))

# 2) 세트는 비시퀀스 자료형
# print(set1[0]) # 인덱스 번호 사용 불가
# print(se1 * 2) # 시퀀스 연산자 사용 불가

# 3) 세트 활용 : 리스트의 중복값 제거
li = [1, 2, 3, 1, 2, 3, 4, 5]
print(li, type(li)) #  리스트는 중복값을 허용
# (1) 리스트를 세트로 형변환해서 중복값을 제거
se = set(li) # 리스트를 세트로 형변환 후 변수에 저장
print(se, type(se)) # 리스트 안에 있는 중복값 제거
# (2) 중복값이 제거된 세트를 다시 리스트로 형변환
li = list(se) # 세트를 리스트로 형변환
print(li, type(li))

print(int(True))
print(int(False))
print(True + 20)
print(False * 20)
```

### 4) 딕셔너리(dict)

```python
# 딕셔너리
# {키1 : 값1, 키2 : 값2, ...}

di1 = {"A" : 1, "B" : 2}
print(di1, type(di1)) # key와 value 한쌍의 하나의 값으로 취급됨

# 1) 딕셔너리는 비시퀀스 자료형
# print(di1[0]) # 인덱스 번호 사용 불가
# print(di1 * 2) # 시퀀스 연산자 사용 불가

# 2) 딕셔너리의 key가 인덱스 번호와 비슷한 역할
# 딕셔너리명[key] => 해당하는 value에 접근
print(di1["A"]) # 1
print(di1["B"]) # 2

# 3) value에는 모든 자료형을 허용
di2 = {"A반" : ["짱구", "철수", "유리"], "B반" : ("둘리", "도우너")}
print(di2, type(di2))
print(di2["A반"], type(di2["A반"]))
print(di2["B반"], type(di2["B반"]))

print(di2.keys()) # key값들만 묶어줌
print(di2.values()) # value들만 묶어줌
print(di2.items()) # key와 value를 하나의 튜플로 묶음
```

### +) 비어있는 컬렉션 생성

```python
# 1. 리스트 => [] 이용하거나 list() 함수를 이용해서 비어있는 리스트 생성
li1 = []
li2 = list()
print(li1, type(li1))
print(li2, type(li2))

# 2. 튜플 => () 이용하거나 tuple() 함수를 이용해서 비어있는 튜플 생성
tu1 = ()
tu2 = tuple()

print(tu1, type(tu1))
print(tu2, type(tu2))

# 3. 세트 => 비어있는 세트는 {}를 이용해서 만들지 못함. set()함수로 생성해야함
se1 = {} # 빈 딕셔너리
se2 = set()

print(se1, type(se1))
print(se2, type(se2))

# 4. 딕셔너리 => {}를 이용하거나 dict()를 이용해서 생성함
di1 = {}
di2 = dict()

print(di1, type(di1))
print(di2, type(di2))
```

### +) in, not in 연산자

```python
# a in 컬렉션 : a라는 값이 컬렉션에 있니????
# a not in 컬렉션 : a라는 값이 컬렉션에 없니????

# 1. 리스트
li = [1, 2, 3]
print(li, type(li))
print(10 in li)

# 2. 튜플
tu = 1, 2, 3 # 튜플의 패킹
print(tu, type(tu))
print(3 in tu) # True

# 3. 세트
se = {1, 2, 3}
print(se, type(se))
print(1 not in se) # False

# 4. 딕셔너리
di = {"a" : "apple", "b" : "banana"}
print(di, type(di))
print("c" in di)
print("a" in di) # a라는 key가 딕셔너리에 있니?
print("apple" in di) # "apple"은 key가 아니라 value값

# +) 문자열 : 문자열에서 특정 문자가 있는지 확인
str1 = "hello python"
print("hello" in str1) # "hello"가 str1 안에 있니??
```

- +) 컬렉션의 길이(요소의 갯수) : len()

```python
# 1. 리스트
li = [1, 2, 3, 4, 5]
print(li)
print(f"li의 길이 : {len(li)}") # 5(요소 5개)

# 2. 튜플
tu = (1, 2, 3, 4)
print(tu)
print(f"tu의 길이 : {len(tu)}") # 4

# 3. 세트
se = {10, 20, 20, 30}
print(f"se의 길이 : {len(se)}") # 3(세트는 중복값을 허용하지 않음)

# 4. 딕셔너리
di = {"a" : "apple", "b" : "bus"}
print(di)
print(f"di의 길이 : {len(di)}") # 2(딕셔너리는 key와 value 한쌍이 하나의 값)

# +) 문자열의 길이(문자의 개수)
str1 = "hello"
print(f"str1의 길이 : {len(str1)}") # 5
```

- 문제5) 가방 안의 물건 확인
- 물건(문자열)들을 담고 있는 가방 리스트 생성
- 사용자로부터 물건의 이름(문자열)을 입력받아
- 입력받은 물건이 가방 안에 있다면 "000은/는 가방에 있습니다"
- 없다면 "000은/는 가방 안에 없습니다"를 출력

```python
back = ["핸드폰", "거울", "지갑", "물티슈"]
print(f"가방속의 물건 : {back}")

item = input("가방속에 있는지 확인할 물건 입력 : ")

print(f"{item}은/는 가방안에 {'있습니다' if item in back else '없습니다'}")
```
