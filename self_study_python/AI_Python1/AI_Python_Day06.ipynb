### 조건문(if문)

```python
# if 조건식 :
# (들여쓰기, tab) 참일때 실행

print("--------- 조건식이 참일 때 결과 --------")
if True :
    # 조건식이 참일때 실행할 문장
    # 들여쓰기 중요! => 조건문 안의 실행문이라는 것을 알려줌
    print("if문 안의 문장")
print("if문 밖의 문장") # 들여쓰기 영역 밖에 있는 문장은 조건식과는 아무 상관없음

print("-------- 조건식이 거짓일때 결과 -------")
if False :
    # 조건식이 거짓이므로 들여쓰기 안의 문장이 실행되지 않음
    print("if문 안의 문장")
print("if문 밖의 문장")

# <if문 예제>
# 사용자로부터 숫자를 입력받아, 입력받은 숫자가 100이라면 "100입력!" 출력

num = int(input("숫자 입력 : "))

# if문 사용
if num == 100 :
  # num에 저장된 값이 100일때 실행할 문장
  print("100입력!")
print("프로그램 종료!")

print("------- 조건식이 참일때 -------")
if True :
  print("if문 안의 문장")
else :
  print("else문 안의 문장")
print("if~else문 끝")

print("-------- 조건식이 거짓일때 --------")
if False :
  print("if문 안의 문장")
else :
  print("else문 밖의 문장")
print("if~else문 끝")

# <if~else문 실습>
# 1) 사용자로부터 정수를 입력받아, 입력받은 정수가 짝수라면 "짝수" 출력
#    그게 아니라면 "홀수" 출력

num = int(input("정수 입력 : "))
if num % 2 == 0 :
  print("짝수")
else :
  print("홀수")

# 2) 사용자로부터 정수를 입력받아, 입력받은 정수가 2와 3으로 나누어 떨어진다면
#    "나누어 떨어집니다" 출력 그게 아니라면 "나누어 떨어지지 않습니다" 출력

num = int(input("정수 입력 : "))
if num % 2 == 0 and num % 3 == 0 :
  print("나누어 떨어집니다")
else :
  print("나누어 떨어지지 않습니다")

# if ~ elif ~ else문
# - 조건식이 2개 이상 있을 때 사용
# - 조건식을 잘못 설정하면 원하는 결과가 나오지 않을 수 있음
# - elif는 여러개 사용 가능

num = int(input("정수 입력 : "))
if num > 20 :
    print("20보다 큽니다")
elif num >= 10 :
    print("10이상 20이하 입니다")
elif num >= 5 :
    print("5이상 10미만 입니다")
else :
    print("5보다 작습니다")

# +) if ~ if
# if를 여러개 작성할 경우 서로 다른 조건문
num = 50

if num > 10 :
    print("10보다 큽니다")

if num > 5 :
    print("5보다 큽니다")

# <조건문 실습>
# 1) 나이를 입력받아 입장료를 계산하는 프로그램
#    10세 이하의 입장료는 1000원, 70세 이상의 입장료는 0원
#    기본 입장료는 2000원
#    입장료 계산 후 입장료와 나이 출력

age = int(input("나이 입력 : "))
if age >= 70 :
    print(f"입장료 : 0원, 나이 : {age}세")
elif age <= 10 :
    print(f"입장료 : 1000원, 나이 : {age}세")
else :
    print(f"입장료 : 2000원, 나이 : {age}세")

# 2) 성적을 입력받아 학점을 계산하는 프로그램
#    90점 이상 A학점
#    80 ~ 89 B학점
#    70 ~ 79 C학점
#    60 ~ 69 D학점
#    그 외 F학점

num = int(input("성적 입력 : "))
if num >= 90 :
    print("A학점")
elif num >= 80 :
    print("B학점")
elif num >= 70 :
    print("C학점")
elif num >= 60 :
    print("D학점")
else :
    print("F학점")
```

### 반복문(while문)

```python
# while 조건식 :
# (들여쓰기, tab) 반복할 문장

while True : # 무한루프(무한 반복) => 조건식이 결과가 참일 동안 반복 실행
  print("while문 안의 문장")
  break # 기타제어문 => 반복문 강제 종료
print("while문 끝") # 반복할 문장 아님! 들여쓰기 영역 밖에 있음

while False : # while문의 조건식이 False 이므로 반복문 자체가 실행이 안됨
  print("while문 안의 문장")
print("while문 끝")

# while문은 조건식을 잘 설정해야한다!
# - while문의 조건식을 잘 못 설정하면
# - 의도하지 않은 무한루프가 발생할 수 있고
# - 반복문 자체가 실행되지 않을 수 있다

# <while문 예제> 1 ~ 10을 출력

# 1 ~ 10까지의 정수를 저장할 변수
num = 1

# 반복문 사용

# 반복문 사용
while num < 11 :
  print(num)
  num += 1 # num = num + 1 => 1씩 증가

print(f"반복 종료 후 num : {num}")

# <while문 실습1>
# 1) 1 ~ 30까지 정수 중 짝수만 출력
# <방법1>
num = 2 # 첫번째 짝수
while num < 31 :
    print(num)
    num += 2 # 2씩 증가

# <방법2> if문 사용
num = 1
while num < 31 :
  if num % 2 == 0 :
     print(num)
  num += 1

# 2) 1시부터 시작해서 7시가 되면 알람이 울리는 프로그램
# <출력예시>
# 1시
# 2시
# 3시
# ...
# 7시가 되었습니다 알람이 울립니다

time = 1 # 1시부터 시작

while time < 7 :
    print(f"{time}시")
    time += 1
# 반복문 종료 후 time에 저장된 값 출력
print(f"{time}시가 되었습니다 알람이 울립니다")

# <무한루프 활용>
# 정수를 하나 입력받고, 입력받은 정수가 1 ~ 10 사이의 정수라면 입력받은 정수를 그대로 출력 후
# 반복문 종료, 그게 아니라면 정수를 다시 입력받도록 한다

while True :
  num = int(input("정수 입력 : "))
  if 1 <= num <= 10 :
    print(f"num = {num}")
    break # 반복문 탈출 => 무한 반복 종료
  else :
    print("잘못된 숫자입니다. 다시 입력해주세요")
```

- 문제 4) 팩토리얼 계산 (while문 사용) 사용자로부터 양의 정수를 입력받아, 해당 숫자의 팩토리얼을 계산하는 프로그램을 while문을 사용하여 작성하세요.

```python
num = int(input("양의 정수를 입력하세요 : "))

result = 1

while num > 0 :
  result *= num
  num -= 1
print(f"입력한 숫자의 팩토리얼은 {result}입니다")
```

- 문제5) 정수를 입력받아서 그 횟수 만큼 'Hello'를 출력하는 프로그램 만들기 0이하의 값이 입력되면 '잘못된 입력입니다.' 출력하고 다시 정수 입력받기

```python
while True :
  num = int(input("반복할 횟수 입력 : "))
  if num < 0 :
    print("잘못된 입력입니다")
  else :
    while num > 0 :
      print("Hello")
      num -= 1
    break
```
