## 클래스

```python
# class 키워드로 클래스를 정의한다
class Dog :
    # __init__ : 생성자
    def __init__(self, name, age) :
        print("Dog클래스 객체 생성")
        self.name = name # 전달받은 name을 '이 객체'의 데이터 속성으로 저장
        self.age = age # 전달받은 age를 '이 객체'의 데이터 속성으로 저장

# 클래스명()로 객체를 찍어낸다
# Dog(...)의 소괄호 안의 값이 __init_의 name, age로 전달된다
my_dog = Dog("초코", 3) 
your_dog = Dog("멍멍이", 10)

print(type(my_dog)) # 이 객체가 Dog 클래스로 만들어졌음을 확인
# 데이터 속성
# : 객체명.데이터속성명을 통해 데이터 속성 사용 가능

# 1) 속성 출력 
print(f"이름 : {my_dog.name}, 나이 : {my_dog.age}살")

# 2) 데이터 속성 값 수정
my_dog.age = 5
print(f"수정 된 나이 : {my_dog.age}살")
class Dog :
    def __init__(self, name, age) :
        self.name = name
        self.age = age
    # 메소드 : 첫번째 매개변수는 self를 작성
    def bark(self) :
        print(f"{self.name}는 멍멍!")

    def introduce(self) :
        print(f"저는 {self.age}살 {self.name}입니다")    

# 객체 생성
# 같은 클래스로 서로 다른 객체를 여러개 만든다
my_dog = Dog("초코", 3)
your_dog = Dog("바둑이", 5)

# 객체명.메소드명()로 메소드 사용 가능
my_dog.bark()
my_dog.introduce()
your_dog.introduce()

my_dog.age = 10 # my_dog의 나이를 바꿔도 your_dog는 그대로이다(서로 독립)
print(f"{my_dog.name} : {my_dog.age}살")
print(f"{your_dog.name} : {your_dog.age}살")
# <클래스 실습>
# 1) Cat 클래스를 정의 후 객체 생성
# - 생성자 : name, color 속성을 초기화
# -meow() 메소드 : "00는 야옹!" 출력
# - introduce() 메소드 : "저는 00색 고양이 00입니다" 출력

class Cat :
    def __init__(self, name, color) :
        self.name = name
        self.color = color
    def meow(self) :
        print(f"{self.name}는 야옹!")
    def introduce(self) :
        print(f"저는 {self.color}색 고양이 {self.name}입니다")

# 객체 생성
cat = Cat("나비", "검은")
cat.meow()
cat.introduce()
# 2) Book클래스 정의 후 객체 생성
# - 생성자 : title, price 속성을 초기화
# - discount() 메소드 : 할인율을 전달받아, 가격을 할인률% 할인한 값을 출력

class Book :
    def __init__(self, title, price) :
        self.title = title
        self.price = price

    def discount(self, rate) :
        # rate : 메소드를 호출할때 전달받을 인수를 저장할 매개변수
         # 가격에서 rate%만큼 할인한 새 가격을 계산
        discounted = self.price - self.price * rate // 100
        print(f"{self.title} 할인가 : {discounted}")
book = Book("해리포터", 15000)
book.discount(20)
```

## 상속

```python
# 부모 클래스 : 동물들의 '공통' 속성, 행동을 작성
class Animal : 
    def __init__(self, name) :
        self.name = name

    def eat(self) :
        print(f"{self.name}은/는 밥을 먹는다")

# 자식 클래스 : Animal을 물려받는다 
class Dog(Animal) :
    def bark(self) :
        print(f"{self.name}은/는 멍멍!")

my_dog = Dog("초코") # Dog엔 생성자가 없지만 부모 것을 물려받아 쓴다
my_dog.eat() # 부모에서 물려받은 메소드
my_dog.bark() # 자식이 추가한 새로운 메소드 
# 자식 클래스에서 자신만의 데이터 속성을 추가하려고 하면
# 자식 클래스의 생성자에서, 부모의 생성자를 super().__init__(...)
# 호출한 뒤 추가

class Animal :
    def __init__(self, name) :
        self.name = name

class Dog(Animal) :
    def __init__(self, name, breed) :
        super().__init__(name) # 부모의 생성자를 호출해 name 속성 값 저장
        self.breed = breed # 자식에서 추가할 속성

dog = Dog("멍멍이", "리트리버")
print(f"{dog.name} {dog.breed}")
# 메소드 오버라이딩
class Animal :
    def sound(self) :
        print("동물이 소리를 낸다")

# 부모의 sound()를 자식에서 각자 다시 정의(오버라이딩)
class Dog(Animal) :
    def sound(self) :
        print("강아지는 멍멍!")
class Cat(Animal) :
    def sound(self) :
        super().sound() # 부모의 메소드 호출
        print("고양이는 야옹~!")

# 같은 이름의 메소드지만 객체(클래스)마다 다르게 동작한다
animal = Animal()
animal.sound()
dog = Dog()
dog.sound()
cat = Cat()
cat.sound()
# <상속 실습>
# 1) 부모 클래스 vehicle 클래스(__init__으로 name 속성, move()로 "00가 이동한다")
# 를 만들고, 
# Car 클래스가 이를 상속받아 honk() 메소드(빵빵!)를 추가
# Car 객체를 만들어 move()와 honk()를 모두 호출

class Vehicle :
    def __init__(self, name) :
        self.name = name

    def move(self) :
        print(f"{self.name}가 이동한다")

class Car(Vehicle) :
    def honk(self) :
        print("빵빵!")

car = Car("자동차")
car.move() # 부모에게 물려받은 메소드
car.honk() # 자식이 추가한 메소드
# 2) 부모 person(__init__으로 name)을 만들고,
#    자식 Student가 super().__init__()으로 name을 물려받은 다음,
#    school 속성을 추가하기.  introduce()로 "00학교의 00입니다" 출력

class Person : 
    def __init__(self, name) :
        self.name = name

class Student(Person) :
    def __init__(self, name, school) :
        super().__init__(name) # 부모의 생성자 호출 
        self.school = school # 자식의 추가 데이터 속성

    def introduce(self) :
        print(f"{self.school} 학교의 {self.name}입니다")

stu = Student("해리포터", "호그와트")
stu.introduce()
```

## 예외처리

```python
# 숫자가 아닌 값을 int로 바꾸려 하면 오류가 나며 여기서 프로그램이 멈춘다
num = int("안녕") # ValueError 발생 -> 아래 줄은 실행되지 않음
print("끝")
# try ~ except
try :
    # 오류가 날 수 있는 코드
    num = int("안녕") 
    print("변환 성공")
except :
    # try에서 오류가 발생하면 실행할 코드
    print("정수로 바꿀 수 없습니다")

print("끝!")
# 특정 예외만 잡기(여러 except)
# except 예외명 
# except를 여러개 두면 상황마다 다른 처리를 할 수 있다

try :
    num1 = int(input("첫번째 정수 입력 : "))
    num2 = int(input("두번째 정수 입력 : "))
    print(f"결과 : {num1 / num2}")
except ValueError :
    # 숫자가 아닌 값을 입력한 경우
    print("숫자만 입력하세요")
except ZeroDivisionError as e : # e : 오류 메시지를 담아준다
    # 0으로 나눈 경우
    print(f"오류 메세지 : {e}")
    print("0으로 나눌 수 없습니다")
except Exception : # 모든 예외 클래스의 부모 클래스 Exception
    print("알 수 없는 예외 발생!")
# try ~ except ~ else ~ finally

try :
    num = int(input("숫자 입력 : "))
except ValueError :
    print("숫자가 아닙니다")
else :
    # 오류가 없을때 실행할 문장
    print(f"입력받은 숫자는 {num}입니다")
finally :
    # 오류 발생 유무 상관없이 실행할 문장
    print("프로그램 종료!")
# raise : 강제로 오류 발생

def set_age(age) :
    if age < 0 :
        # 잘못된 값이면 직접 예외를 발생시킨다
        raise ValueError("나이는 0이상이어야 합니다")
    print(f"나이 설정 : {age}")

set_age(20)
# set_age(-5)

# 잘못된 값 -> raise로 예외 발생 -> try ~ except로 잡는다
try :
    set_age(-5)
except ValueError as e :
    print(f"오류 : {e}")
# 사용자 정의 예외

# Exception을 상속받아 나만의 예외클래스를 만들 수 있다 
class NeativeAgeError(Exception) :
    pass

def set_age(age) :
    if age < 0 :
        # 내가 만든 예외 발생
        raise NeativeAgeError("나이는 음수가 될 수 없습니다")
    print(f"나이 : {age}")

try :
    set_age(-1)
except NeativeAgeError as e :
    print(f"사용자 정의 예외 : {e}")
# <예외처리 실습>
# 1) 사용자 정의 예외 BalanceError(잔액 부족)를 만들고,
#    잔액보다 큰 금액을 출금하려 하면 rasie하고 try ~ except로 예외처리

class BalanceError(Exception) :
    pass

# 잔액에서 출금하는 함수 정의
def withdraw(balance, amount) :
    if amount > balance :
        raise BalanceError("잔액이 부족합니다")
    print(f"{amount}원 출금 완료")

withdraw(3000, 1000) # 잔액 3000원에서 1000원 출금
# withdraw(1000, 3000) 

try :
    withdraw(1000, 5000) # 잔액 1000원인데 5000원 출금 시도
except BalanceError as e :
    print(f"출금 실패 : {e}")
```

## 문제 4) 상속과 super() — 부모 물려받기
부모 Employee 클래스: __init__으로 name 속성을 받고, work() 메서드로 "OO — 출근해서 일합니다" 를 출력하세요.
자식 Developer 클래스: Employee를 상속받고, super().__init__()으로 name을 물려받은 뒤 language 속성을 추가하세요. coding() 메서드로 "OO — OO 코딩합니다" 를 출력하세요.
Developer("김개발", "파이썬") 객체로 work()와 coding()을 모두 호출하세요.
  <출력결과>
  김개발 — 출근해서 일합니다
  김개발 — 파이썬 코딩합니다

```python
class Employee :
    def __init__(self, name) :
        self.name = name
    def work(self) :
        print(f"{self.name} - 출근해서 일합니다")
class Develope(Employee) :
    def __init__(self, name, language) :
        super().__init__(name)
        self.language = language
    def coding(self) :
        print(f"{self.name} - {self.language} 코딩합니다")
de = Develope("김개발", "파이썬")
de.work()
de.coding()

class Employee :
    def __init__(self, name) :
        self.name = name
    def work(self) :
        print(f"{self.name} - 출근해서 일합니다")
class Develope(Employee) :
    def __init__(self, name, language) :
        super().__init__(name)
        self.language = language
    def coding(self) :
        super().work()
        print(f"{self.name} - {self.language} 코딩합니다")
de = Develope("김개발", "파이썬")
de.coding()
```

## 문제 7) 여러 except — 상황마다 다른 처리
두 문자열을 정수로 바꿔 나누는 함수 divide(a, b) 를 만드세요.
ValueError(숫자가 아님)와 ZeroDivisionError(0으로 나눔)를 각각 잡아 다른 안내를 출력하세요.
divide("10", "2"), divide("10", "0"), divide("십", "2") 를 차례로 호출하세요.
  <출력결과>
  결과: 5.0
  0으로 나눌 수 없습니다
  숫자만 입력하세요

```python
def divide(a, b) :
    try :
        num1 = int(a)
        num2 = int(b)
        result = num1 / num2
        print(f"결과 : {result}")
    except ValueError :
        print("숫자만 입력하세요")
    except ZeroDivisionError :
        print("0으로 나눌 수 없습니다")
divide("10", "2")
divide("10", "0")
divide("십", "2")
```

## 문제 8) else ~ finally — 성공/실패와 무관하게 마무리
문자열을 정수로 바꾸는 코드를 try에 넣으세요.
오류가 없을 때만 else에서 "변환 성공: OO" 를 출력하세요.
성공하든 실패하든 항상 finally에서 "작업 종료" 를 출력하세요.
text = "42" (성공하는 값)로 실행하세요.
  <데이터>
  text = "42"

  <출력결과>
  변환 성공: 42
  작업 종료

```python
text = "42"

try :
    int(text)
except :
    pass
else :
    print(f"변환 성공 : {text}")
finally :
    print("작업 종료")
```

## 문제 9) raise — 직접 예외 일으키기
할인율을 검사하는 함수 check_rate(rate) 를 만드세요. rate가 0~100 범위를 벗어나면 raise ValueError("할인율은 0~100 사이여야 합니다") 로 예외를 발생시키세요.
check_rate(20) 은 정상 출력하고, check_rate(150) 은 try~except로 잡아 오류 메시지를 출력하세요.
  <출력결과>
  할인율: 20%
  오류: 할인율은 0~100 사이여야 합니다

```python
def check_rat(rate) :
    if rate < 0 or rate > 100 :
        raise ValueError("할인율은 0 ~ 100 사이어야 합니다")
    print(f"할인율 : {rate}%")
check_rat(20)

try :
    check_rat(150)
except ValueError as e :
    print(f"오류 : {e}")
```

## 문제 10) 사용자 정의 예외 — 나만의 예외 만들기
Exception을 상속해 사용자 정의 예외 OutOfStockError(재고 부족)를 만드세요. (앞에서 배운 '상속'이 여기서 쓰입니다!)
order(stock, amount) 함수: 주문 수량이 재고보다 많으면 OutOfStockError("재고가 부족합니다") 를 발생시키고, 아니면 "OO개 주문 완료" 를 출력하세요.
order(10, 3) 은 정상, order(10, 20) 은 try~except로 잡아 메시지를 출력하세요.
  <출력결과>
  3개 주문 완료
  주문 실패: 재고가 부족합니다

```python
class OutOfStockError(Exception) :
    pass
def order(stock, amount) :
    if stock < amount :
        raise OutOfStockError("재고가 부족합니다")
    print(f"{amount}개 주문 완료")
try :
    order(10, 3)
    order(10, 20)
except OutOfStockError as e:
    print(f"주문 실패 : {e}")
```
