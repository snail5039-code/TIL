## 데이터 프레임 활용

````python
import pandas as pd
import seaborn as sns
# seaborn : 시각화 라이브러리, 연습용 데이터를 가지고 있다

# tips 데이터 가져오기
tips = sns.load_dataset("tips")
tips.head()
# 총 결제 금액이 40달러 이상인 손님만 가져오기
tips[tips["total_bill"] >= 40]
# 여러 조건을 동시에 연결
# - &(and)
# - |(or)
# - 각 조건을 반드시 ()로 감싸야 한다

# 결제 금액이 40달러 이상이고 방문 요일이 토요일인 손님만 필터링
tips[(tips["total_bill"] >= 40) & (tips["day"] == "Sat")]
# 토요일 또는 일요일에 방문한 손님 필터링
weekend = tips[(tips["day"] == "Sat") | (tips["day"] == "Sun")]
print(f"주말 손님 수 : {len(weekend )}명")
# len(df) : 행의 개수
weekend
# +) 여러 값 중 하나 => isin([값 목록])
weekend = tips[tips["day"].isin(["Sat", "Sun"])]
print(f"주말 손님 수 : {len(weekend)}")
weekend
# +) ~(not)를 조건의 반대가 된다
# 주말이 아닌 평일 손님만
result = tips[~tips["day"].isin(["Sat", "Sun"])]
print(f"평일 손님 수 : {len(result)}")
result
# <필터링 실습>
# tips 데이터에서
# 1) 팁(tip)이 6달러 이상인 손님만 출력, 인원수도 출력
result = tips[tips["tip"] >= 6]
print(f"팁이 6달러 이상인 인원수 : {len(result)}명")
result.head()
# 2) 저녁(time이 "Dinner")이면서 인원(size)이 5명 이상인 손님과 인원수 출력
result = tips[(tips["time"] == "Dinner") & (tips["size"] >= 5)]
print(f"저녁이면서 인원이 5이상인 손님 : {len(result)}")
result
````
