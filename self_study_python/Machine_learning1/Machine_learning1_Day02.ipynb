1. 컬렉션
 : 여러개의 값을 하나로 묶어서 사용하는 자료형
종류		괄호		특징
===========================================================
리스트 list	[]		순서 o, 수정 o
튜플 tuple	()		순서 o, 수정 x
딕셔너리 dict	{key:value}	key를 사용하여 value를 꺼냄
세트 set	{}		중복 허용 x, 순서 x

1) 리스트 list
- [] 를 사용 ,(콤마)로 값들을 구분
- 시퀀스 자료형 => 인덱스 번호 사용 가능
- 인덱싱, 슬라이싱 가능 

2) 리스트의 메소드
- .append(값) : 맨 뒤에 추가
- .insert(idx, 값) : 지정한 idx 위치에 값 추가
- .remove(값) : 값을 찾아서 삭제
- .pop(idx) : 해당 위치의 값을 반환 후 삭제 
	      (idx 생략 시 맨 뒤가 삭제됨)
- .sort() : 리스트 오름차순 정렬 
	ex) .sort(reverse = True) : 요소 내림차순 정렬 

3) 튜플
- () 를 사용 ,(콤마)로 값들을 구분
- 리스트와 거의 비슷하지만 수정은 불가능 
- 패킹과 언패킹
	- 패킹(packing) : 여러 값을 하나의 튜플로 묶는 것 
	- 언패킹(unpacking) : 튜플 안의 값들을 여러 변수에
			      한번에 나눠 담는 것

4) 딕셔너리
- {키:값} 형태, 키에 값을 짝지어 저장
- 딕셔너리는 키로 값을 꺼내서 사용
  딕셔너리명[key]
- 딕셔너리 메소드
	.get(키) : 값 조회(없으면 None)
	.keys() : 키들만 묶음
	.values() : 값들만 묶음
	.items() : (키, 값) 쌍 묶음

5) 세트
- 중복 허용 x, {}를 사용
- 순서가 없음(인덱스 개념이 없음)
- 가장 흔한 용도 : 중복 제거(리스트를 set()함수에 사용 -> 중복이 제거됨)
+) 세트의 집합 연산(합집합, 교지합, 차집합)
- 세트 & 세트 => 교집합
- 세트 | 세트 => 합집합
- 세트 - 세트 => 차집합

## 컬렉션
### 1) 리스트(list)

```python
# []를 사용하여 값들을 묶어줌

fruits = ["사과", "바나나", "포도", "귤"]
print(fruits, type(fruits))

# 인덱스 번호 사용 가능
print(fruits[0])
print(fruits[-1])
print(fruits[1:3])
print(len(fruits)) # 리스트의 길이(값의 개수)
# 리스트의 메소드
fruits = ["사과", "바나나", "포도", "귤"]

# 1) 추가
fruits.append("수박")
print(fruits)

fruits.insert(1, "멜론") # idx 1번 위치에 추가
print(fruits)

# 2) 수정(인덱스 번호 사용)
fruits[0] = "청사과"
print(fruits)

# 3) 삭제
fruits.remove("포도")
print(fruits)

print(f"삭제할 값 : {fruits.pop(1)}")
print(fruits)

# 4) 정렬
fruits.sort()
print(fruits)

fruits.sort(reverse=True)
print(fruits)
# 리스트와 같이 많이 사용하는 함수
scores = [70, 95, 60, 88]

print(f"리스트의 길이 : {len(scores)}")
print(f"합계 : {sum(scores)}")
print(f"최고점 : {max(scores)}")
print(f"최저점 : {min(scores)}")
```

### 2) 튜플(tuple)

```python
# () 로 튜플을 만든다
tu = (10, 20)
print(tu, type(tu))

# 인덱스 번호 사용 가능
print(tu[0])
# 단, 튜플은 수정 불가능
# tu[0] = 99
# 튜플에 있는 값을 수정(리스트로 변환 후 수정)
li = list(tu)
print(li, type(li))
li[0] = 99
print(li)
tu = tuple(li)
print(tu)
# ** 튜플의 패킹과 언패킹

# 패킹 : ,(콤마)로 값들을 나열하면 하나의 튜플로 묶음
person = "홍길동", 20, "컴공"
print(person, type(person))

# 언패킹 : 튜플의 값을 변수에 한번에 푼다
name, age, major = person
print(name)
print(age)
print(major)
# +) 튜플의 요소가 1개일 때 => 값 뒤에 ,(콤마)를 작성
one_tu = (5,)

print(one_tu, type(one_tu))
```

### 3) 딕셔너리

```python
# {키:값,...}

student = {"이름" : "홍길동", "나이" : 20, "학과" : "컴공"}
print(student, type(student))
print(student["이름"])

# 있는 key에 값을 넣으면 수정
student["나이"] = 21
print(student)

# 없는 키를 넣으면 추가
student["반"] = "A반"
print(student)
# 딕셔너리 메소드
print(student)

print(student.keys()) # key들만 묶음
print(student.values()) # value들만 묶음
print(student.items()) # (key, value) 쌍들 => 각 쌍이 튜플로 묶임

# get : 없는 키를 사용해도 오류 발생 x => None이 나옴
print(student.get("나이"))
print(student.get("전화번호"))
```

### 4) 세트(set)

```python
# 중복이 섞인 리스트
numbers = [1, 2, 2, 3, 3, 3]
print(numbers)

# set로 바꾸면 중복이 사라진다
se = set(numbers)
print(se, type(se))

# 세트의 집합 연산
set_a = {1, 2, 3}
set_b = {2, 3, 4}

# 1) 교집합
print(set_a & set_b)

# 2) 합집합
print(set_a | set_b)

# 3) 차집합
print(set_a - set_b)
```

### 반복문(for문)

```python
# fro + range()

for i in range(3) :
    print(f"i = {i}")

print("=" * 10)

for i in range(1, 4) :
    print(f"i = {i}")

print("=" * 10)

for i in range(0, 10, 2) :
    print(f"i = {i}")

# 반복 횟수 지정
print("=" * 10)

for i in range(3) :
    # 3번 반복할 코드
    print("Hello")
# for + 컬렉션

fruits = ["사과", "바나나", "포도"]

for fruit in fruits :
    print(f"과일 : {fruit}")

print("=" * 10)

scores = {"국어" : 90, "영어" : 85}

# items()로 키와 값을 동시에 사용 (튜플의 언패킹)
for subject, score in scores.items() :
    print(f"{subject} : {score}점")
# <제어문 실습>

# 1) 5명의 학생의 파이썬 점수를 입력받아 리스트에 저장 후
# 학생들의 점수의 총합과 평균 점수 출력

# # 5명의 학생의 점수를 저정할 빈 리스트 생성
# scores = []

# # for문 사용 5명의 학생의 점수를 입력받아 리스트에 저장
# for i in range(5) :
#     score = int(input(f"{i + 1}번째 학생 점수 입력 : "))
#     scores.append(score)

# 리스트 컴프리헨션 사용
# [요소 for 변수 in 반복가능객체 if 조건식]
scores = [int(input(f"{i+1}번째 학생 점수 입력")) for i in range(5)]

print(scores)
# 학생의 총합과 평균 점수 계산

print(f"전체 총합 : {sum(scores)}점")
print(f"평균 점수 : {sum(scores) / len(scores) : .2f}")
a_list = ['above', 'cookie', 'app', 'about', 'admin', 'bisket', 'apple', 'april']
# 2) 해당 리스트에서 a를 포함하지 않은 문자열을 모두 제거한 리스트를 새로 생성

print(a_list)

for word in a_list :
    # 'a'가 word에 없는 경우
    if "a" not in word :
        # word를 리스트에서 제거
        a_list.remove(word)
        print(f"{word}가 제거됩니다")

print(a_list)
stock = {"사과" : 3, "바나나" : 0, "포도" : 5}
# 3) 위의 딕셔너리에서 재고(value)가 0인 상품은 제외하고
#    과일명과 재고 개수를 출력

for item, amount in stock.items() :
    if amount == 0 :
        continue # 다음 반복으로 넘어감
    print(f"{item} : {amount}개")
```

```python
# break문

# 1부터 출력하다가 5가 되면 반복을 즉시 종료 
for i in range(1, 10) :
    if i == 5 :
        break # for문 즉시 종료
    print(i)
# continue

# 1~5 중 짝수는 건너뛰고 홀수만 출력
for i in range(1, 6) :
    if i % 2 == 0 :
        continue
    print(i)
# pass

age = 20

if age >= 19 :
    pass # 19이상이면 나중에 채울테니 일단 넘어가
else :
    print("미성년자입니다")
```

### 리스트컴프리헨션

```python
# 0 ~ 4 '제곱' 리스트 생성

# [방법1] for문 사용
squares = []
for i in range(5) :
    squares.append(i ** 2)
print(squares)

# [방법2] 리스트컴프리헨션 -> 위 3줄을 한줄로 줄여 작성
squares =[i ** 2 for i in range(5)]
print(squares)
# 조건사용
# 1 ~ 10 중에서 짝수만 골라 리스트로 생성
# [방법1] for문 if문
evens = []
for i in range(1, 11) :
    if i % 2 == 0 :
        evens.append(i)
print(evens)

# [방법2] 리스트 컴프리헨션 사용 => 4줄을 한줄로 줄여 사용
evens = [i for i in range(1, 11) if i % 2 == 0]
print(evens)
# +) 삼항 연산자와 사용
# 삼항 연산자 : 참일때 if 조건식 else 거짓일때 

# 점수 리스트에서 60점 이상이면 "합격", 아니면 "불합격"으로 바꿔 담기
scores = [90, 45, 70, 30]
result = ["합격" if s >= 60 else "불합격" for s in scores]
print(result)

# 아래 코드를 줄여서 사용한 것이다
result = []
for s in scores :
    if s >= 60 :
        result.append("합격")
    else :
        result.append("불합격")
print(result)
```

## 문제 5) if ~ elif ~ else — 기온으로 날씨 판정
기온에 따라 30 이상이면 "더움", 20 이상이면 "적당", 10 이상이면 "쌀쌀", 그 외에는 "추움"을 정해 출력하세요.

## 문제 6) for와 range — 짝수의 합
for와 range를 이용해 1부터 20까지의 짝수만 더한 합계를 출력하세요.
(힌트: range(시작, 끝, 간격)으로 2씩 건너뛸 수 있습니다. 끝 번호는 포함하지 않습니다.)
  <출력결과>
  1~20 짝수의 합: 110

```python
total_sum = 0
for i in range(1, 21) :
    if i % 2 == 0 :
        total_sum += i
print(f"1~20 짝수의 합 : {total_sum}")
```

## 문제 7) for + 딕셔너리 + continue — 품절 상품 건너뛰기
가격표를 items()로 순회하되, 가격이 0원인(품절) 상품은 continue로 건너뛰고 판매 중인 상품만 출력하세요.
판매 중인 상품들의 총액도 함께 출력하세요.
  <데이터>
  cart = {"우유": 2500, "빵": 0, "계란": 4000, "치즈": 0}

  <출력결과>
  우유: 2500원
  계란: 4000원
  판매 상품 총액: 6500원

```python
cart = {"우유": 2500, "빵": 0, "계란": 4000, "치즈": 0}
total_price = 0
for product, price in cart.items() :
    if price == 0 :
        continue
    print(f"{product} : {price}원")
    total_price += price
print(f"판매 상품 총액 : {total_price}원")
# 이런식으로 된다
total = sum(cart.values()) 
print(total)
```
