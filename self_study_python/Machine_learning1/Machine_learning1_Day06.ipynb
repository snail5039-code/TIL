## 넘파이
### 집계 함수

```python
import numpy as np
# 1차원 배열

arr = np.array([10, 20, 30, 40, 50])
print(arr)

print(f"합계 : {arr.sum()}")
print(f"평균 : {arr.mean()}")
print(f"최대값 : {arr.max()}")
print(f"최소값 : {arr.min()}")
print(f"표준편차 : {arr.std()}")
# 2차원 배열

# 학생 3명 x (국어, 영어) 점수
scores = np.array([[90, 80],
                  [70, 60],
                  [100, 50]])
print(scores)

print("==== 집계 함수 ====")
print(scores.sum()) # axis 없음 -> 전체 합계
print(scores.mean(axis=0)) # 열별(과목별) 평균 -> [국어 평균 영어 평균]
print(scores.mean(axis=1)) # 행별(학생별) 평균
```

### 최대 / 최소 값 위치 찾기

```python
arr = np.array([3, 7, 2, 9, 5])

print(f"최대값 : {arr.max()}")
print(f"최대값 위치 : {arr.argmax()}") # 3번 인덱스
print(f"최소값 위치 : {arr.argmin()}") # 2번 인덱스
# 분류 모델은 "각 정답일 확률"을 배열로 가져온다
# ex) [0.1 0.7 0.2]면 argmax()로 가장 확률 높은 위치(1번)을 골라
# "이 데이터는 1번 종류다"라고 예측한다

pred = np.array([0.1, 0.7, 0.2])
print(f"예측 : {pred.argmax()}번")
# <넘파이 배열 실습>
# 1) 2차원 배열(학생 3명의 국어, 영어) 
#  첫번째 학생 : 국어 80점, 영어 90점
# 두번째 학생 : 국어 70점, 영어 60점
# 세번째 학생 : 국어 100점, 영어 85점
# 과목별 평균과 학생별 평균을 출력

data = np.array([[80, 90],
                [70, 60],
                [100,85]])

print(f"전체 평균 : {data.mean() : .2f}")
print(f"과목별 평균 : {data.mean(axis=0).round(2)}")
print(f"학생별 평균 : {data.mean(axis=1)}")
```

## 시리즈(Series)

```python
import pandas as pd

# 시리즈 생성
sr1 = pd.Series([10, 20, 30, 40])
print(sr1)
# 인덱스를 설정하지 않으면 0부터 자동으로 붙는다

# index로 각 값에 이름(라벨)을 붙인다

sr2 = pd.Series([90, 85, 95], index=["국어", "영어", "수학"], name="학생1")
print(sr2)
# 시리즈의 구조
# : '인덱스', '값(데이터)', '열이름'으로 구성됨
print(sr2.index) # 인덱스(이름표)들
print(sr2.values, type(sr2.values)) # 데이터 => 넘파이 배열 형태
print(sr2.name) # 열이름
print(sr2.dtype) # 값들의 자료형
```

### 시리즈 데이터 접근

```python
# 시리즈 생성
sr = pd.Series([90, 85, 95], index=["국어", "영어", "수학"])
print(sr)

# 1) 이름(라벨) 사용
print("="*10)
print(sr.loc["국어"])
print(sr["국어"])
# 2) 위치 인덱스 번호
print(sr.iloc[0])
# 3) 여러개를 리스트로 골라 가져오기
print(sr[["국어", "수학"]])

# 4) 슬라이싱(이름(라벨)으로 슬라이싱은 "끝"도 포함한다!)
print(sr["국어" : "영어"])

# +) 위치 슬라이싱([0:2])은 끝을 포함하지 않지만,
#    이름 슬라이싱(["국어":"영어"])은 끝을 포함한다!
```

### 시리즈 연산

```python
print(sr)

# 벡터 연산(개별 값과 연산)
print("="*10)
print(sr + 5) # 모든 값에 5가 더해짐(인덱스 유지됨)
print(sr * 2)
# 시리즈 + 시리즈
sr_a = pd.Series([1, 2, 3], index=['a', 'b', 'c'])
sr_b = pd.Series([10, 20, 30], index=['b', 'c', 'd'])

print(sr_a)
print(sr_b)
print("="*10)

print(sr_a + sr_b)  # b, c만 짝이 맞음. a, d는 짝이 없어 NaN
# NaN(Not a Number) : 결측치(결측값) => 빠져 있는 데이터

# 라벨(이름) 기준으로 자동 정렬되는 것이 pandas의 장점
# : 순서가 뒤섞인 데이터도 이름만 맞으면 알아서 짝지어서 계산한다
```

### 조건 필터링

```python
sr = pd.Series([90, 85, 95, 100, 70],
              index=["국어", "영어", "수학", "과학", "사회"])
print(sr)

print("="*10)
# 1) 90점 이상인 과목만
# print(sr >= 90) # 조건에 맞으면 True, 아니면 False가 들어간 시리즈
print(sr[sr >= 90])
# 2) 조건이 여러개인 경우
# & (and)
# / (or)

# 점수가 90점 이상 이거나(or) 80점 미만인 과목만
print(sr[(sr >= 90) | (sr < 80)])
```

### 시리즈 통계

```python
scores = pd.Series([85, 90, 78, 92, 88, 76])

print(f"개수 : {scores.count()}")
print(f"중앙값 : {scores.median()}")
# 데이터가 짝수일때, 가운데 2개의 값의 평균
print("="*10)
print(scores.describe())
fruits = pd.Series(["사과", "바나나", "사과", "포도", "바나나", "사과"])

# +) value_counts() : 각 값이 몇번 나오는지 센 다음,
# 데이터가 많은 순으로 정렬
print(fruits.value_counts())
print(f"최빈값 : {fruits.mode()[0]}")
# +) sort_values() : 값 기준으로 오름차순 정렬한다
#    ascending=False 옵션 설정으로 하면 내림차순 정렬을 한다

print(sr)

print("=== 오름차순 정렬 ===")
print(sr.sort_values())

print("=== 내림차순 정렬 ===")
print(sr.sort_values(ascending=False))
import numpy as np
# +) 결측값 : 결측지, 비어 있는 값, NaN으로 표현

# 결측값이 포한된 시리즈 생성
sr_nan = pd.Series([10, np.nan, 30, np.nan, 50])
# np.nan => 결측치
print(sr_nan)
# 결측치가 포함된 시리즈로 통계 메소드 사용
print(f"평균 : {sr_nan.mean()}") # 통계 함수는 NaN은 빼고 계산한다
# => 평균은 10, 30, 50의 평균 => 30
# <시리즈 실습>
# 1) 점수를 저장한 시리즈 생성 후
# 점수 : 88, 92, 45, 70, 95
# 80점 이상인 값만 필터링 후 출력, 평균도 출력 

scores = pd.Series([88, 92, 45, 70, 95])
print(scores[scores >= 80])
print(f"평균 : {scores.mean()}")
# 2) 위 시리즈를 점수가 높은 순으로 정렬 후 출력(내림차순 정렬)
print(scores.sort_values(ascending=False))
# 3) 색상이 저장된 시리즈 생성 후
# 색상 : 빨강, 파랑, 빨강, 초록, 파랑, 빨강
# 각 색이 몇 번 나오는지 출력, 최빈값도 출력

colors = pd.Series(["빨강", "파랑", "빨강", "초록", "파랑", "빨강"])
print(colors.value_counts())
print(f"최빈값 : {colors.mode()[0]}")
```

## 문제 7) value_counts와 통계

1. Series `pd.Series(["A","O","A","B","O","O","AB"])` (혈액형)에서 **각 값이 몇 번 나오는지** 세어 출력하기 (`value_counts`)
2. Series `pd.Series([85, 90, 78, 92, 88, 76])` (점수)의 **평균·중앙값**을 출력하고, `describe()` 로 통계 요약을 한 번에 출력하기

        <출력결과>
        O     3
        A     2
        B     1
        AB    1
        Name: count, dtype: int64
        평균 : 84.83333333333333
        중앙값 : 86.5
        count     6.000000
        mean     84.833333
        std       6.524314
        min      76.000000
        25%      79.750000
        50%      86.500000
        75%      89.500000
        max      92.000000
