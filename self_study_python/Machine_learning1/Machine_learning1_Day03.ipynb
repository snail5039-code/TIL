## 사용자 정의 함수

```python
# def 키워드로 함수를 정의 
def say_hello() :
    print("안녕하세요!")
    print("Hello~!")
# 함수 호출 : 정의한 함수를 사용 
# => 함수명()
say_hello()
# 매개변수가 포함된 함수 정의

def greet(name) :
    # name : 인수를 저장할 매개변수 
    print(f"{name}님, 환영합니다~!")
# 함수 호출
greet("짱구")
greet("훈이")
# 숫자 2개를 전달받아, 전달받은 숫자들의 합을 출력하는 함수 정의 
def add(x, y) :
    print(f"{x} + {y} = {x + y}")
    # +) 매개변수나 함수 안에서 만든 변수는 그 함수 안에서만 사용할 수 있다

# 함수 호출 
add(10, 20)
add(y=20, x=10)
# print(f"함수 밖 : {x}") # 매개변수 x는 함수 밖에서 사용 불가 
# 디폴트 매개변수

def order(coffe, size = "중간") :
    print(f"{size} 사이즈 {coffe}")

order("아메리카노") # size 생략 -> 기본값 "중간"
order("라떼", "큰")
# return (반환값)

def add(x, y) :
    return x + y # 두 숫자를 반환하고 함수 종료

result = add(10, 20) # 함수를 호출하여 반환 받은 값 30을 변수에 저장
print(f"결과 : {result}")
print(add(10, 20) * 2) # 30 * 2
print(add(10, 20) + add(30, 40)) # 30 + 70
# 여러 값 반환
# return 뒤에 ,(콤마)로 값을 나열하면 튜플로 패킹되어 반환된다
# => 함수를 호출하는 쪽에서는 언패킹으로 여러 변수에 나눠 사용할 수 있다

# 두개의 숫자를 전달받아, 전달받은 숫자의 합과 차이를 반환하는 함수 정의 
def calc(x, y) :
    return x + y, x - y # 튜플의 패킹

print(calc(20, 10), type(calc(20, 10)))

# 튜플의 언패킹으로 합과 차이를 두 변수에 각각 저장
add_result, sub_result = calc(20, 10)
print(f"합 : {add_result}")
print(f"차 : {sub_result}")
# <함수 실습1>
# 1) 숫자요소를 가지는 리스트를 전달받아,
#    리스트 요소들의 합과 평균을 반환하는 summary() 함수를 정의 후 호출 
def summary(numbers) :
    total_sum = sum(numbers)
    avg = total_sum / len(numbers)
    return total_sum, avg

li = [1, 2, 3, 4, 5]
summary(li)

total_sum, avg = summary([10, 20, 30, 40])
print(f"합계 : {total_sum}, 평균 : {avg : .2f}")
# 2) 두개의 숫자를 전달받아 두 숫자를 나눈 몫과 나머지를 반환하는
#    divide()함수 정의 후 호출

def divide(x, y) :
    quo = x // y
    mod = x % y
    return quo, mod
divide(10, 2)

quo, mod = divide(17, 5)
print(f"몫 : {quo}, 나머지 : {mod}")
```

## 람다(lambda)

```python
# def로 정의한 함수
def add(x, y) :
    return x + y

# 람다로 사용
print((lambda x, y : x + y)(30, 40))
add_lambda = lambda x, y : x + y # 람다를 변수에 저장
print(add_lambda(3, 5))

print(add(10, 20))
# 람다 활용 sorted()

# (이름, 나이) 튜플이 저장된 리스트
people = [("철수", 25), ("영희", 19), ("민수", 32)]
# 나이 기준으로 오름차순 정렬
# key=lambda : 각 요소에서 '정렬에 사용할 값'을 뽑아준다
sorted_people = sorted(people, key=lambda person : person[1])
print(sorted_people)
# 기준만 바꾸면 다른 방식으로도 정렬할 수 있다

words = ["apple", 'kiwi', 'banana', 'fig']
# 문자열의 길이(글자 수) 기준으로 오름차순 정렬 

sorted_words = sorted(words, key=lambda w : len(w))
print(sorted_words)
# 람다 활용 map()

numbers = [1, 2, 3, 4]

# 리스트의 각 요소에 제곱한 결과를 한꺼번에 저장한 리스트 생성 
result = list(map(lambda num : num ** 2, numbers))
print(result)
# 람다 활용 filter()

numbers = [1,2, 3, 4, 5, 6]

# 리스트에서 짝수인 값으로만 구성된 리스트 생성 
result = list(filter(lambda num : num % 2 == 0, numbers))
print(result)
# <람다 실습>
scores = [("수학", 80), ("영어", 95), ("국어", 70)]
# 1) 위의 리스트를 점수가 높은 순(내림차순)으로 정렬한 결과를 출력 
sorted_scores = sorted(scores, reverse=True, key=lambda tu : tu[1])
print(sorted_scores)
# 2) map()과 람다로
li = [1, 2, 3, 4, 5]
# 리스트의 각 값을 3배로 바꿔서 출력
result = list(map(lambda num : num * 3, li))
print(result)
# 3) filter()와 람다로 
li = [15, 3, 22, 8, 40, 6]
# 위의 리스트에서 10이상인 값들만 새로운 리스트에 담아 출력 
result = list(filter(lambda x : x >= 10, li))
print(result)
```

### 자주 사용하는 기본 모듈

## 문제 2) 기본값 매개변수
가격과 세율을 받아 세금 포함 가격을 return하는 함수 price_with_tax를 만드세요. 세율의 기본값은 0.1(10%) 입니다.
세율을 생략한 호출과, 0.2로 지정한 호출을 각각 출력하세요.
  <출력결과>
  기본 세율: 11000.0
  지정 세율: 12000.0

```python
def price_with_tax(price, tax=0.1) : 
    return price * (1 + tax)
print(f"기본 세율: {price_with_tax(10000)}")

print(f"지정 세율: {price_with_tax(10000, 0.2)}")
```

## 문제 3) 여러 값 반환 — 최저·최고
숫자 리스트를 받아 (최솟값, 최댓값)을 함께 return 하는 함수 min_max를 만드세요.
아래 리스트를 넣어 언패킹으로 받아 출력하세요.
  <데이터>
  numbers = [34, 12, 89, 5, 67]

  <출력결과>
  최저: 5, 최고: 89

```python
numbers = [34, 12, 89, 5, 67]

def min_max(number) :
    return max(number), min(number)
result_max, result_min = min_max(numbers)

print(f"최저 : {result_min}")
print(f"최고 : {result_max}")
```
