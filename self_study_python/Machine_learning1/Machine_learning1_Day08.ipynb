## 데이터 프레임
### 열 추가

````python
# 데이터 불러오기
import seaborn as sns
import pandas as pd

# seaborn의 tips 데이터 불러오기
tips = sns.load_dataset("tips")
tips.head()
# 모든 행에 "미국"이 채워진 "나라"컬럼 추가
# df["컬럼명(새로운)"] = 값
tips["나라"] = "미국"
tips.head()
# 팁 비율(%) 컬럼 추가 소수점 두번째 자리까지
# = 팁 / 총액 x 100 
tips ["tip_pct"] = (tips["tip"] / tips["total_bill"] * 100).round(2)
tips.head()
# 기존 열로 새 정보를 만들어 내는 것은 데이터 분석 및 머신러닝에서 데이터 준비
# 단계에서 자주하는 작
````

### 열 삭제

````python
# 열 삭제 : columns에 지울 열 이름
tips.drop(columns=["나라"])
tips.head() # 원본에는 삭제가 되어 있지 않음
# 삭제한 결과를 원본에 반영하는 방법
# 1) 다시 변수에 저장
# 2) inplace=True 옵션 사용
tips_drop = tips.drop(columns=["나라"])
tips_drop.head()
tips.drop(columns=["나라"], inplace=True)
tips.head()
# 행 삭제(0번, 1번 행을 삭제) => 원본 삭제 x
tips.drop(index=[0, 1])
````

### 정렬

````python
# 총 결제 금액이 큰 순서로 정렬(내림차순)
tips.sort_values(by="total_bill", ascending= False)
# +) 여러 열 기준 정렬 : by= 에 열 이름을 리스트로 작성하면,
# 첫번째 열로 정렬하고 값이 같을 때 다음 열 순서로 정렬한다
# ascending도 [True, False]처럼 열마다 방향을 정할 수 있다
# - 인원(size)이 적은 순으로 정렬하되, 인원이 같으면 팁(tip) 많은 순으로
tips.sort_values(by=["size", "tip"], ascending=[True, False])
````

## 데이터 프레임 통계

````python
tips.head()
# 데이터프레임 전체에 통계 메소드 사용 
# => 숫자 열마다 평균이 들어간 시리즈가 반환
tips.mean(numeric_only=True)
# max(), min(), sum(), median() 도 똑같이 열마다 적용된다
tips.max(numeric_only=True)
# 특정 열의 통계
print(f"평균 결제금액 : {tips["total_bill"].mean() : .2f}")

# 팁과 인원수의 중앙값
tips[["tip", "size"]].median()
````

### 히스토그램

````python
import numpy as np

np.random.seed(0) # 난수의 시작값 고정

# np.random.normal(평균, 표준편차, 개수)
scores = np.random.normal(70, 10, 100)
# 평균은 70점, 표준편자 10점 근처로 흩어진 점수 100개를 생성
# print(scores)

# bins=8 : 점수 범위를 8개 구간으로 나눠 각 구간의 인원(데이터 개수)을 센다
plt.hist(scores, bins=8, color="skyblue", edgecolor="black")
plt.title("점수 분포")
plt.xlabel("점수")
plt.ylabel("학생 수")
plt.show()
# tips의 결재액 분포 시각화
import seaborn as sns

tips = sns.load_dataset("tips")

plt.hist(tips["total_bill"], bins=20, color="orange", edgecolor="black")
plt.title("총 결제액 분포")
plt.xlabel("결제액")
plt.ylabel("손님 수")
plt.show()
````
