3) 자료형(data type)
 : 데이터의 종류 

(1) 숫자형 : 정수(int)와 실수(float)
- int : 소수점이 없는 숫자
- float : 소수점이 있는 숫자 

(2) 문자열 : str
- 문자열 : 문자들의 나열

	슬라이싱 -> 문자열[시작:끝:간격] 
		: 시작 번호부터 끝 번호 앞까지 잘라낸다
		(끝 번호는 포함하지 않음)

(3) 논리형 : bool
- True/False 딱 두가지 값만 갖는 자료형

4) 형변환
- 자료형을 다른 종류의 자료형으로 변환하는 것 
- int() : 정수로 변환, int("3") -> 3
- float() : 실수로 변환, float(3) -> 3.0
- str() : 문자열로 변환, str(3) -> "3"
- bool() : 논리형으로 변환, bool(0) -> False
	True로 변환되는 것 : False로 변환되는 것을 제외한 전부
	False로 변환 되는 것 : 0, 0.0, ""(빈문자열), (), [], {}
			       ...

5) input() : 사용자에게 입력받는 함수
- input("메시지") : 실행하는 사람에게 값을 직접 입력받는다
** input()으로 받은 값은 무조건 문자열!
 => 다른 값으로 입력받을 때는 형변환을 해야함
	ex) int(input("정수 입력 : "))

3. 연산자
 : 특수한 기능을 가진 특수 문자

1) 최우선 연산자 
- () 소괄호 안의 연산을 가장 먼저 한다 
ex) 2 + 2 * 4 = 14
   (2 + 3) * 4 => 20

2) 단항 연산자 
- 부호 연산자 : +(양수), -(음수)
- not 연산자 : not True -> False
	       not False -> True

3) 산술 연산자
- 기본 사칙 연산(+, -, *, /)
- / : 나누기(결과는 실수), 분모에 0 사용 불가
	ex) 10 / 2 => 5.0
	    10 / 0 => 오류 발생(0으로 나눌 수 없음)
- % : 나머지 연산자 7 % 2 => 1
- // : 몫 연산자 7 // 2 => 3
- ** : 거듭 제곱 2 ** 3 => 8

4) 관계 연산자(비교 연산자)
- 두 값을 비교하여 논리형으로 결과가 나옴
- >, <, >=, <=, ==, !=
ex) 20 == 20 => 20과 20이 같니? => True
    20 != 20 => 20이랑 20이 다르니 => False

5) 논리 연산자 
- 참/거짓을 여러개 연결해서 사용하는 연산자
- 조건 여러개를 묶을 때 사용
- and : 둘다 참이여야 참
- or : 하나라도 참이면 참 

6) 삼항 연산자 
- 참일때 값 if 조건식 else 거짓일때 값 
- 조건에 따라 둘 중 하나의 값이 나옴 
- 조건 하나로 값을 정할때 코드를 간단하게 작성할 수 있다 

7) 산술 대입 연산자(누적 연산자)
- 대입 연산자와 산술 연산을 동시에 사용
- +=, -=, *=, /=, //=, %=, **=

## 자료형

```python
# 숫자형
price = 1000 # 정수(int)
ratio = 3.14 # 실수(float)
print(price, type(price))
print(ratio, type(ratio))

print(price + 500)
print(ratio * 2)

# 문자열(str)
word = "ptyhon"
print(word, type(word))

# 인덱싱 : 글자 하나 꺼내기
print(word[0]) # 'p'
print(word[-1]) # 'n'

# 슬라이싱 : 범위로 잘라내기
print(word[0:2]) # idx 0 ~ 1번
print(word[:2])

print(word[2:]) # idx 2 ~ 끝까지
print(word[::2]) # 처음 ~ 끝까지, 간격 2씩 슬라이싱

# 문자열 시퀀스 연산자
# - 문자열 + 문자열 => 연결
# - 문자열 * int => 반복

first = "홍"
last = "길동"
print(first + last)

print("=" * 10) # '='을 10번 반복

# 논리형
print(10 > 3)
print(10 == 3)

# 입력 함수
age = int(input("나이 입력 : "))
print(age, type(age))

print(f"내년이면 {age + 1}살이 됩니다")

# <기초 문법 실습>
# 1) 문자열 변수 email에 "test@gmail.com"을 저장하고,
#    슬라이싱으로 @ 앞의 아이디 부분(test)만 잘라서 출력
# (1) 이메일을 변수에 저장
email = "test@gmail.com"
print(email)

user_id = email[:4] # 4는 포함하지 않음
print(user_id)

# 2) 사용자로부터 2개의 숫자를 각각 입력받아,
#    두수의 합을 "두 수의 합은 00입니다"형태로 출력

num1 = int(input("첫번째 숫자 입력 : "))
num2 = int(input("두번째 숫자 입력 : "))

print(f"두 수의 합은 {num1 + num2}입니다")
```

## 연산자

```python
# 최우선 연산자
print((2 + 3) * 4) # 20

# 단항 연산자
# 1) 부호 연산자
temp = 8
print(-temp) # -8
print(-(-temp)) # -(-8) => 8

# 2) not 연산자
print(10 > 3)
print(not(10 > 3)) # not True => False

# 관계 연산자와 논리 연산자

age = 20
has_ticket = True

# and : 두 조건이 모두 참이어야 True
print(age >= 19 and has_ticket)

# or : 둘중 하나라도 참이면 True
print(age < 8 or age >= 65) # False or False => False

# 삼항 연산자

score = 75
# 60점 이상이면 "합격" 그게 아니라면 "불합격" 출력
result = "합격" if score >= 60 else "불합격"
print(f"{score}점은 {result}입니다")

# 산술 대입 연산자

total = 0
total += 1000 # total = total + 1000
total += 500 # total = total(1000) + 500
print(total)

count = 10
count -= 3
print(count)

# <연산자 실습>
# 1) total_seconds에 150(초)을 저장한 후,
#   초를 사용하여 분과 초를 계산하고 "2분 30초 형태로 출력"

total_seconds = 150

# 1분 = 60초
minutes = total_seconds // 60 # 몫 => 분
seconds = total_seconds % 60 # 나머지 => 초
print(f"{minutes}분 {seconds}초")

# 주피터 노트북 코드 자동 완성 단축키 : Tab키

# 2) 사용자로부터 정수를 입력받아,
#    입력받은 정수가 5의 배수라면 "5의 배수입니다" 출력
#    그게 아니라면 "5의 배수가 아닙니다" 출력

# 삼항 연산자
num = int(input("정수 입력 : "))
print("5의 배수 입니다" if num % 5 == 0 else "5의 배수가 아닙니다")

# 3) 사용자로부터 가격을 입력받아,
#    입력받은 가격이 2000원 이상이라면, 30% 할인한 가격을 출력
#    그게 아니라면 입력받은 가격을 그대로 출력

# (1) 가격 입력받기
price = int(input("가격 입력 : "))

#(2) 입력받은 가격에 따라 할인된 가격이나 가격 그대로 계산
# 삼항 연산자, 산술 대입 연산자
price *= 0.7 if price >= 2000 else 1

print(f"결과 : {price}")
```

## 문제 6) 산술 연산자 — 시간 변환

- 전체 초를 `//`(몫)와 `%`(나머지)를 이용해 **시·분·초**로 나누어 출력하세요.
- (힌트: 1시간 = 3600초, 1분 = 60초)

        <데이터>
        total_seconds = 3725

        <출력결과>
        1시간 2분 5초

```python
total_seconds = 3725

h = total_seconds // 3600
re_h = total_seconds % 3600
m = re_h // 60
s = total_seconds % 60

print(f"{h}시간 {m}분 {s}초")
```

## 문제 8) 삼항 연산자와 산술 대입 연산자

- 장바구니 금액이 30000원 이상이면 배송비 0원, 아니면 3000원입니다. **삼항 연산자**로 배송비를 정해 출력하세요.
- 산술 대입 연산자(`+=`)로 장바구니 금액에 배송비를 더해 **최종 결제금액**을 출력하세요.

        cart = 25000

        <출력결과>
        배송비: 3000
        최종 결제금액: 28000

```python
cart = 25000

result = 0 if cart >= 30000 else 3000

print(f"배송비 : {result}")
print(f"최종결제 금액 : {result + cart}")
```
