## 클래스의 속성

#### 1) 데이터 속성의 종류
##### (1) 인스턴스 변수
- self 키워드를 붙여서 사용하는 변수
	ex) self.age, self.name,...

#### 2) 메소드의 종류
##### (1) 인스턴스 메소드
- 첫번째 매개변수에 self를 사용

##### (2) 클래스 메소드

##### (3) 정적 메소드

#### 1) 예외 처리(Exception Handling)
- 파이썬에서 발생하는 각종 예외를 처리하기 위함
- 발생된 문제로 인해 프로그램이 비정상적으로 종료되는 것을 막고
  사용자에게 발생한 문제에 대한 정보를 전달하기 위해 사용

#### 2) 예외 처리 구조
##### (1) try ~ except
try : 
(들여쓰기, teb) 예외가 발생할 수 있는 문장
except 예외명 : 
(들여쓰기, teb) try영역에서 해당 예외가 발생했을 때 실행할 문장
	
##### (2) try ~ except ~ else ~ finally
try :
	예외가 발생할 수 있는 문장
except :
	try영역에서 예외가 발생할 때 실행할 문장
else : 
	try영역에서 예외가 발생하지 않았을 대 실행할 문장
finally :
	예외 발생 유무에 상관없이 무조건 실행할 문장

#### 3) 강제로 예외 발생
- 파이썬이 발생시키는 오류가 아니라 사용자가 발생시키고 
  싶은 경우에 사용
- raise 예외클래스명("예외 메세지 작성")

#### 4) 사용자 정의 예외 클래스

## 상속

```python
# 부모 클래스는 여러 자식 클래스를 가질 수 있다
# => 하나의 클래스가 여러 클래스에 상속을 할 수 있다 

# 부모 클래스
class Person :
    # 생성자
    def __init__(self, name) :
        self.name = name
    # 메소드
    def sleep(self) :
        print(f"{self.name}은/는 잠을 잔다")

# Person 클래스의 속성을 상속받는 자식 클래스
class Student(Person) :
    # 생성자
    def __init__(self, name, school) :
        # 부모 클래스의 생성자 호출
        super().__init__(name) # super() : 부모의 메소드 호출
        self.school = school # Student 클래스에서 추가된 데이터 속성
    # Student에서 추가할 메소드
    def study(self) :
        print(f"{self.name}이/가 {self.school}에서 공부한다")

# Person 클래스를 상속받는 Employee 클래스 정의 
class Employee(Person) :
    # 생성자
    def __init__(self, name, commpany) :
        super().__init__(name)
        self.commpany = commpany # Employee에서 추가된 데이터 속성
    # Employee 클래스에서 추가할 메소드
    def work(self) : 
        print(f"{self.name}은/는 {self.commpany}에서 근무한다")

# Student 클래스 객체 생성
st = Student("짱구", "코리아IT학교")
st.sleep() # 부모 클래스에서 상속받은 메소드 호출
st.study() # Student 클래스에서 추가된 메소드 호출

# Employee 클래스 객체 생성
em = Employee("철수", "IT회사")
em.sleep() # 부모 클래스에서 상속받은 메소드 호출
em.work() # Employee 클래서에서 추가된 메소드 호출 
```

## 오버라이딩(재정의)

```python
# 부모 클래스
class Animal :
    def __init__(self, name) :
        self.name = name 

    def make_sound(self) :
        print("동물은 어떤 소리를 낼까요?")

# 자식 클래스
class Dog(Animal) :
    def __init__(self, name) :
        super().__init__(name) 
    
    # 부모에게 상속받은 make_sound 재정의(오버라이딩)
    def make_sound(self):
        super().make_sound() # 부모 클래스의 make_sound() 호출
        print(f"강아지 {self.name}은/는 멍멍!") 
    
class Cat(Animal) :
    def __init__(self, name) :
        super().__init__(name)
    def make_sound(self):
        super().make_sound()
        print(f"고양이 {self.name}은/는 야옹!")

# Dog 클래스 객체 생성
dog = Dog("멍멍이")
dog.make_sound() # Dog클래스에서 재정의한 기능이 실행됨

# Cat 클래스 객체 생성
cat = Cat("야옹이")
cat.make_sound() # Cat클래스에서 재정의한 기능이 실행됨
```

## 예외 처리 구조

```python
# try ~ except
try : 
    # 예외가 발생될 것으로 예상되는 문장
    num1 = int(input("첫번째 정수 입력 : "))
    num2 = int(input("두번째 정수 입력 : "))
    print(f"{num1} / {num2} = {num1 / num2}")
except ZeroDivisionError as e :
    # try 영역에서 ZeroDivisionError가 발생했을 때 실행할 문장
    print("0으로 나눌 수 없습니다")
    print(f"예외 메세지 : {e}")
except ValueError as e :
    print("정수만 입력할 수 있습니다")
    print(f"예외 메세지 : {e}")
except Exception : # Exception은 모든 예외 클래스의 부모 클래스
    print("알 수 없는 예외가 발생했습니다")

# <예외 클래스> ZeroDivisionError, ValueError 
# 1) 파이썬에서 발생할 수 있는 다양한 예외를 처리할 수 있는 클래스
# 2) 모든 예외의 클래스의 부모 클래스는 Exception 클래스이다

li = [1, 2, 3, 4]
print(li[0])
print(li[10])

# try ~ except ~ else finally
try :
    li = [1, 2, 3, 4]
    print(li[0])
    # print(li[10])
except :
    print("except문 실행")
else : 
    # try블록에서 예외가 발생하지 않을때 실행할 문장
    print("else문 실행")
finally :
    # 예외 발생 유무 상관없이 실행할 문장
    print("finally문 실행")
```

## 강제로 예외 발생

```python
# raise

num = int(input("1 ~ 5사이의 정수 입력 : "))

# 범위를 벗어나면 예외 발생
if not(1 <= num <= 5) :
    raise ValueError("범위에 벗어났습니다") 

# try ~ except로 발생시킨 예외 처리
try :
    num = int(input("1 ~ 5사이의 정수 입력 : "))

    if not(1 <= num <= 5) :
        raise Exception("에러 발생!")
    print(f"입력한 값은 {num}입니다")
except ValueError as e :
    print("정수를 입력해주세요!")
    print(f"예외 메세지 : {e}")
except Exception as e :
    print("1 ~ 5사이의 정수를 입력해주세요")
    print(f"예외 메세지 : {e}")
```

## 사용자 정의 예외

```python
# 홀수를 입력받을 때 발생시킬 예외 클래스 정의

class CheckEven(Exception) :
    def __init__(self) :
        super().__init__("짝수만 입력해주세요")

# 사용자 정의 예외처리
try :
    num = int(input("정수 입력 : "))
    # 입력한 값이 짝수가 아닐 경우 예외 발생!
    if num % 2 != 0 :
        raise CheckEven # 사용자 정의 예외 발생
except CheckEven as e :
    print(e) # 예외 메세지 출력
else : 
    # 예외가 발생하지 않았을 때 
    print(f"입력한 정수는 {num}입니다!")
finally :
    # 예외 상관없이 실행할 문장
    print("프로그램 종료")

# <예외처리 실습>
# 1) IsFail 클래스 정의
# : 예외가 발생했을 경우 합격하지 못했다는 예외 메세지 출력

# 2) 사용자로부터 점수를 입력받아,
#    입력받은 점수가 70점 이상이라면 합격 문구 출력
#    70점 미만이라면 IsFail이라는 예외 발생 후 예외 처리

# 예외 클래스 정의
class IsFail(Exception) :
    def __init__(self, score) :
        super().__init__(f"{score}점은 합격 조건에 맞지 않습니다")

# 예외 처리
try : 
    score = int(input("점수 입력 : "))
    if score < 70 :
        # IsFail 예외를 발생
        raise IsFail(score)
except IsFail as e :
    print(e) # 예외 메세지 출력
else : 
    # score가 70이상인 경우
    print(f"축하드립니다! {score}점은 합격 점수입니다!")
finally :
    print("프로그램 종료!~")
```

### 문제 1) RPG 게임 프로그램(상속)

- Character클래스 선언
    - 1) 생성자 : 캐릭터의 닉네임, 레벨을 초기화
    - 2) info() 메소드 : 캐릭터의 닉네임과 레벨 출력

- Worrior 클래스  선언
    - 1) Character클래스의 자식 클래스
    - 2) 생성자 : 캐릭터의 닉네임, 레벨, 무기이름 초기화
    - 3) attack() 메소드 : 닉네임과 무기이름을 활용해 공격 메세지 출력

- Magician 클래스 선언
    - 1) Character클래스의 자식 클래스
    - 2) 생성자 : 캐릭터의 닉네임, 레벨, 마법이름 초기화
    - 3) cast() 메소드 : 닉네임과 마법을 활용해 공격 메세지 출력

- Worrior 클래스와 Magician 클래스의 객체 생성 후,
각각  info() 메소드와 자식 클래스에서 추가된 메소드 호출

        <객체 생성후 출력 결과>
        캐릭터명 : 고길동
        레벨 : 40Lv
        전사 고길동이/가 생선가시(으)로 공격!

        캐릭터명 : 둘리
        레벨 : 10Lv
        마법사 둘리이/가 호이호이(으)로 마법공격!

```python
class Character :
    def __init__(self, name, lvel) :
        self.name = name
        self.lvel = lvel
    def info(self) :
        print(f"캐릭터명 : {self.name}")
        print(f"레벨 : {self.lvel}Lv")
class Worrior(Character) :
    def __init__(self, name, lvel, weapon) :
        super().__init__(name, lvel)
        self.weapon = weapon
    def attack(self) :
        print(f"전사 {self.name}이/가 {self.weapon}(으)로 공격!") 
class Magician(Character) :
    def __init__(self, name, lvel, magic_name) :
        super().__init__(name, lvel)
        self.magic_name = magic_name
    def cast(self) :
        print(f"마법사 {self.name}이/가 {self.magic_name}(으)로 마법 공격!")  
wor = Worrior("고길동", 40 ,"생선가시")
magic = Magician("둘리", 10 ,"호이호이")
wor.info()
wor.attack()
magic.info()
magic.cast()
```

### 문제 2) 성인 여부 확인 프로그램(예외처리, 사용자 정의 예외)
- 사용자로부터 나이를 입력받아
- 입력받은 나이가 20살 이상이면 나이와 성인여부 문구 출력
- 입력받은 나이가 20살 미만이면 AgeError 발생
- AgeError 예외 클래스
    - 해당 예외 발생시 입력받은 나이와 성인이 아니라는 문구 출력

            <출력예시1>
            나이를 입력해주세요 >> 15
            15세는 성인이 아닙니다!
            프로그램 종료~!

            <출력예시2>
            나이를 입력해주세요 >> 21
            21세는 성인입니다!
            프로그램 종료~!

```python
class AgeError(Exception) :
    def __init__(self, age) :
        super().__init__(f"{age}세는 성인이 아닙니다!")
try :
    age = int(input("나이를 입력해주세요 >> "))
    if age < 20 :
        raise AgeError(age)
except AgeError as e :
    print(e)
else :
    print(f"{age}세는 성인입니다!")
finally :
    print("프로그램 종료~!")
```
