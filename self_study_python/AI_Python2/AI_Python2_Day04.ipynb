### 재귀함수

않고 종료된다
  - 기본 조건이 충족되면 함수는 더 이상 자기 자신을 호출하지 않고
결과를 반환
- (2) 재귀 호출
  - 함수가 자기 자신을 호출하는 부분
  - 이 호출은 문제를 점차 더 작은 크기로 줄여가며,
최종적으로는 기본 조건에 도달하게 된다

### 람다(lambda)
- : 람다는 익명 함수를 만들기 위해 사용
- 익명이라는 것은 이름이 없다는 것이다
- 익명함수란 이름이 없는 함수로 재사용하지 않는다는 것이다(일회성)
- 람다는 함수를 간편하게 작성할 수 있어서
다른 함수의 인수로 전달할 때 사용

#### 사용 방법
- [일반적인 함수 정의]
def 함수명(매개변수, ...) :
return 반환값
- [람다(lambda)]
lambda 매개변수 : 반환값

#### 람다의 활용
- (1) map() 함수
  - map(함수, 반복가능객체)
  - 첫번째 매개변수에는 함수,
두번째 매개변수로는 반복가능 객체(리스트, 튜플, range(), ...)가 온다
  - map()함수의 반환 값은 map객체이기 때문에
list(), tuple()로 형변환을 해야한다
  - map()함수 동작
: 반복가능 객체 요소들을 첫번째 인수로 작성한 함수에 하나씩 전달해서
반환받은 값들을 하나로 묶음
  - map(적용시킬 함수, 적용할 값들)

- (2) reduce() 함수
  - reduce(함수, 반복가능객체)
  - 파이썬 내장 함수가 아니기 때문에
from functools import reduce 을 작성해야 한다
  - reduce()함수 동작 :
반복가능 객체의 요소를 지정된 함수로 처리한 뒤
이전 결과와 누적한 결과를 반환하는 함수

- (3) filter() 함수
  - filter(함수, 반복가능객체)
  - filter() 함수 동작 :
반복가능 객체 요소 중 조건에 맞는 요소만 하나로 묶음
전달 된 함수가 True를 반환하는 값들을 묶음
  - 전달할 함수는 조건함수로 작성
  - 조건함수 : 반환(return)값이 True 혹은 False인 함수

### 재귀함수

````python
# 재귀함수 : 자기 자신을 호출하는 함수

def recur_print(count) :
    print(f"{count}번 함수 호출")
    # 기본조건 : 재귀 호출을 멈추는 조건
    if count == 3 :
        print(f"{count}번 함수 종료")
        return

    # 재귀호출 : 자기 자신을 호출하는 부분
    recur_print(count + 1)
    print(f"{count}번 함수 종료")

# recur_print(0) => recur_print(1) => recur_print(2) => recur_print(3)
# recur_print(0) <= recur_print(1) <= recur_print(2) <= return

# 함수 호출
recur_print(0)

# <재귀함수 예제>
# 재귀함수로 1 ~ n까지의 합을 계산

def sum_recur(num) :
    # 기본조건 : 재귀호출을 멈추는 조건
    if num == 1 :
        return 1 # 1을 반환하면서 함수 종료
    # 재귀호출 : 자기자신을 호출
    return num + sum_recur(num-1)
    #       3 + sum_recur(2)
    #               2 + sum_recur(1)
    #                       1 (함수 종료)

# 함수 호출
print(f"1부터 3까지의 합 : {sum_recur(3)}")

# <재귀함수 실습>
# 1 ~ n까지의 곱을 반환하는 factorial() 함수 정의 후 호출 

def factorial(num) :
    # 기본조건 : num이 1이거나 0이면 1이나 0을 반환
    if num == 0 :
        return 0
    if num == 1 :
        return 1 # 1을 반환하면서 함수 종료
    # 재귀 호출    
    return num * factorial(num - 1)
    #       3  * factorial(2)
    #               2   * factorial(1)
    #                           1

# 함수 호출
print(factorial(3))
print(factorial(1))
print(factorial(0))
````

## 1) map()함수

````python
# map(함수, 반복가능객체)

# 1) 1 ~ 10까지의 정수에 +2를 한 값들을 저장한 리스트
result = list(map(lambda x : x+2 , range(1, 11)))
print(result, type(result))

# 2) 1 ~ 5까지의 정수에 제곱한 값들을 저장한 튜플
result = tuple(map(lambda x : x**2, range(1, 6)))
print(result)
````

## 2) reduce() 함수

````python
from functools import reduce

# reduce(함수, 반복가능객체) : 반복가능 객체의 요소를 함수에 적용한 뒤 이전 결과와 누적

# 1) 1 ~ 3까지의 합을 반환하는 reduce()함수 작성
result = reduce(lambda x, y : x + y, range(1, 4))
# range(1, 4) : 1 ~ 3
#                  x   +    y
# 첫번째 값        1   +    2    두번째 값
# 이전 결과(1+2)   3   +    3    세번째 값
# return               6 

print(result)

# 2) 1 ~ 5까지의 합을 reduce()로 계산
result = reduce(lambda x, y : x + y, range(1, 6))
#               x   +   y
# 첫번째        1   +   2   두번째 값
# 이전 결과     3   +   3   세번째 값
# 이전 결과     6   +   4   네번째 값
# 이전 결과     10  +   5   마지막 값    
# return            15
print(result)
````

## 3) filter()함수

````python
# filter(함수, 반복가능객체) : 반복가능 객체 요소 중 조건에 맞는 값들을 묶음

# 1) 1 ~ 10까지 정수 중 5보다 작은 값들로 구성된 리스트 생성
result = list(filter(lambda x : x < 5, range(1, 11)))
print(result) # 람다가 True를 반환하는 값들을 묶음

# 2) 1 ~ 50까지 정수 중 3의 배수로만 구성된 리스트 생성
result = list(filter(lambda x : x % 3 == 0, range(1, 51)))
print(result)

# <람다 실습>
# 학생들의 점수를 저장한 리스트
score_li = [80, 76, 50, 60, 45, 95]
# 가 있다

# 1) 람다와 map()함수를 사용하여, 점수가 70점 이상이면 "합격" 아니면 "불합격"
#   값으로 저장된 새로운 리스트 생성
# 결과 : ['합격', '합격', '불합격', '불합격', '불합격', '합격']

result = list(map(lambda x : "합격" if x >= 70 else "불합격" , score_li))
print(result)

# 삼항 연산자 : 조건식이 True일때 if 조건식 else 조건식이 False일때
print("합격" if 90 > 70 else "불합격")

# 2) 위의 score_li에서 70점 이상인 점수들의 합계 계산
#   람다와 filter() / 람다와 reduce() 

# 방법 1) filter()함수 사용
# (1) filter()함수 사용하여 70점 이상인 점수만 가져오기
result = list(filter(lambda x : x >= 70, score_li))
print(result)
# (2) 리스트의 합계 계산
total = sum(result)
print(total)

from functools import reduce
# 방법 2) reduce()함수와 삼항 연산자 사용

total = reduce(lambda x, y : (x if x >= 70 else 0) + (y if y >= 70 else 0), score_li)
# score_li에 가져온 요소가 70점 이상일 경우 : x + y
#                          70점 미만일 경우 : x + 0 
print(total)

# 3) 과일 이름을 저장한 리스트
fruits = ["apple", "banana", "melon", "kiwi"]
# 람다와 fiter()사용하여 글자가 5글자 이하인 과일명으로 저장된 새로운 리스트 생성

result = list(filter(lambda x : len(x) <= 5, fruits))
print(result)
````

## 문제 1) 리스트의 각 요소 제곱하기
- 람다 함수를 사용하여 리스트의 각 요소를 제곱하는 코드를 작성
- map() 함수와 람다 함수를 사용하여 리스트의 각 숫자를 제곱
- 리스트 [1, 2, 3, 4, 5]를 입력하면 각 요소가 제곱된 새 리스트를 반환

```
        <출력예시>
        제곱하기 전 : [1, 2, 3, 4, 5]
        제곱한 후 : [1, 4, 9, 16, 25]
```

````python
li = [1, 2, 3, 4, 5]

result = list(map(lambda x : x ** 2, li))
print(f"제곱하기 전 : {li}")
print(f"제곱한 후 : {result}")
````

## 문제 2) 짝수와 홀수 판별하기
- filter() 함수와 람다 함수를 사용하여 주어진 리스트에서 짝수만 추출
- 리스트에서 짝수로만 구성된 새로운 리스트를 반환

```
        <출력예시>
        람다 적용 전 : [10, 15, 20, 25, 30, 35]
        람다 적용 후 : [10, 20, 30]
```

````python
li = [10, 15, 20, 25, 30, 35]
result = list(filter(lambda x : x % 2 == 0, li))
print(f"람다 적용 전 : {li}")
print(f"람다 적용 후 : {result}")
````

## 문제 3) 리스트 요소의 곱 구하기
- reduce()와 lambda를 사용하여 리스트에 있는 모든 숫자의 곱을 계산하는 코드를 작성

````python
from functools import reduce
li = [1, 2, 3, 4, 5]
result = reduce(lambda x, y : x * y, li)
print(f"리스트 : {li}")
print(f"리스트 요소들의 곱 : {result}")
````

## 문제 4) 리스트 내 최댓값 구하기
- reduce()와 lambda를 사용하여 리스트에서 최대값 구하기

```
        <출력예시>
        리스트 : [42, 17, 68, 99, 23, 10]
        최대값 : 99
```

````python
li = [42, 17, 68, 99, 23, 10]

result = reduce(lambda x, y : (x if x > y else y), li)
print(f"리스트 : {li}")
print(f"최대값 : {result}")
````
