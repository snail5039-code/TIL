## 2. 클래스의 속성
- 데이터 속성 : 값으로서 데이터를 정의하는 것
- 메소드 : 기능(행위)로서의 처리를 정의하는 것

#### 1) 데이터 속성의 종류
##### (1) 인스턴스 변수
- self키워드 붙여서 사용하는 변수
	self.데이터속성
- 객체들마다 다른 값을 저장할 수 있는 변수 
- 객체를 통해서만 접근 간으
- 객체명.인스턴스 변수명
	ex) pr.name

##### (2) 클래스 변수
- 한 클래스에서 만들어진 객체가 동일한 값을 가지는 변수
- 객체와 클래스명으로 접근 가능 
- 클래스명.클래스변수명 or 객체명.클래스변수명

#### 2) 메소드의 종류
##### (1) 인스턴스 메소드
- 인스턴스 변수를 사용하는 메소드를 의미 
- 첫번째 매개변수에 self를 작성하여 인스턴스 변수에 접근 
- 객체를 통해서만 호출 가능 
- 객체명.인스턴스메소드명()

##### (2) 클래스 메소드
- 클래스 변수를 사용하는 메소드를 의미 
- 첫번째 매개변수에 cls(class)를 작성하여 클래스 변수에 접근
- 객체나 클래스명으로 호출 가능 
- 메소드 정의 위에 @classmethod 데코레이터를 표시하고 작성

##### (3) 정적 메소드
- 인스턴스 변수, 클래스 변수를 모두 사용하지 않는 메소드를 의미 
- 반드시 작성해야 하는 매개변수가 없음
- 메소드 위에 @staticmethod 데코레이토를 표시하고 작성

## 생성자

```python
# 생성자가 추가된 Person 클래스 정의

class Person :
    # 생성자 : Person 객체를 만들때 실핼할 내용
    # => 객체를 생성할때 name, age 데이터 속성 초기화
    def __init__(self, name, age) :
        # print("Person클래스 객체 생성")
        self.name = name
        self.age = age

    def introduce(self) :
        print(f"안녕하세요 저는 {self.name}입니다")
        print(f"제 나이는 {self.age}살 입니다")

# Person클래서 객체 생성
p1 = Person("둘리", 10) # Person() : 생성자 호출
p2 = Person("도우", 11)

p1.introduce()
p2.introduce()

# <생성자 실습>
# Player() 클래스 정의
# - 데이터 속성 : 이름(name), 나이(age), 직업(job)
# - 생성자 : 이름, 나이, 직업 초기화
# - show() 메소드 : 이름, 나이, 직업을 출력
# 축구 선수, 야구 선수 2개의 객체 생성 후 각각 show() 메소드 호출

class Player :
    def __init__(self, name, age, job) : 
        self.name = name
        self.age = age
        self.job = job
    def show(self) :
        print(f"제 이름은 {self.name}입니다")
        print(f"제 나이는 {self.age}살 입니다")
        print(f"제 직업은 {self.job}입니다")

p1 = Player("둘리", 20, "축구 선수")
p2 = Player("철수", 22, "야구 선수")

p1.show()
p2.show()

# 객체 별로 속성을 추가할 수 있다
# : 객체명.속성명(기존 x) = 값
p1.country = "한국"
print(p1.country)
# print(p2.country)
# => 오류 발생 p2객체에는 country 데이터 속성을 추가하지 않아 오류 발생

# <생성자 실습2>
# Student 클래스 정의
# - 생성자 : 학생의 이름, 나이, python점수, java점수, C언어 점수를 받아
#            데이터 속성 초기화
# - 메소드 :
# 1) info() : 학생의 이름, 나이를 출력하는 기능
# 2) report() : 학생의 python, java, C언어 점수를 출력한 후, 
#               3과목의 총점과 평균 계산 후 출력
#
# - Student 클래스 객체 생성 후 메소드 호출

class Student :
    def __init__(self, name, age, python, java, c) :
        self.name = name
        self.age = age
        # 과목명과 점수가 저장된 딕셔너리를 데이터 속성으로 생성
        self.subject = {"python" : python, "java" : java, "C언어" : c}
    def info(self) :
        print(f"이름 : {self.name}, 나이 : {self.age}")
    # 성적표 출력
    def report(self) :
        print(f"===== {self.name}의 성적표 =====")
        # 과목 점수 출력(딕셔너리)
        for key, value in self.subject.items() :
            print(f"{key}의 점수 : {value}점")
        # 총점과 평균 계산 후 출력 
        # 과목마다의 점수는 self.subject의 value값
        total = sum(self.subject.values()) # value값들의 핪
        # 평균 = 총점 / 과목 수(딕셔너리의 길이)
        avg = total / len(self.subject)
        print(f"==== 총점과 평균 ====")
        print(f"총점 : {total}")
        print(f"평균 : {avg : .2f}")

s1 = Student("둘리", 11, 50, 70, 100)
s1.info()
s1.report()
```

## 클래스의 속성

```python
from ast import NameConstant
class Student :
    school_name = "python 학교" # 클래스 변수, 모든 객체들이 같은 값을 가지고 시작 

    # 생성자 : 학생의 이름, 나이를 초기화
    def __init__(self, name, age) :
        self.name = name # 인스턴스 변수 : 객체들마다 다른 값을 저장할 수 있는 변수
        self.age = age
    
    # 학생의 이름, 나이를 출력하는 메소드(인스턴스 변수 사용)
    def print_info(self) : # 인스턴스 메소드
        print(f"학생 이름 : {self.name}")
        print(f"학생 나이 : {self.age}")
    
    # 학생들의 학교명을 출력하는 메소드(클래스 변수 사용)
    @classmethod
    def print_school(cls) : # 클래스 메소드
        print(f"학생들은 {cls.school_name}에 다닌다!")

    # 등교하기 위한 교통수단을 출력하는 메소드
    # => 인스턴스, 클래스 변수 둘다 사용 x
    @staticmethod
    def print_trans() : # 정적 메소드
        print("학생들은 버스를 타고 등교한다!")

# Student 클래스 객체 생성
st1 = Student("둘리", 12)
st2 = Student("도우너", 11)

print("==== 데이터 속성 ====")
# 인스턴수 변수 : 객체를 통해서만 사용 가능 
print(st1.name) # 객체들마다 다른 값을 저장함
print(st2.name)
# print(Student.name) # 클래스명으로 인스턴스 변수 사용 불가능

# 클래스 변수 : 객체와 클래스명을 통해 접근 가능
print(st1.school_name) # 객체들 모두 같은 값을 가지고 생성
print(st2.school_name)
print(Student.school_name)
# => 인스턴스 변수와 구분하기 위해 클래스명을 통해 사용하는 것이 권장사항

print("\n==== 메소드 호출 ====")

# 인스턴스 메소드 호출 : 객체를 통해서만 호출 가능 
st1.print_info()
# Student.print_info() # 클래스명으로 사용 불가

# 클래스 메소드 호출 : 객체와 클래스명을 통해 호출 가능 
st1.print_school()
Student.print_school()

# 정적 메소드 호출 : 객체와 클래스명을 통해 호출 가능
st1.print_trans()
Student.print_trans()
```

## 상속

```python
# 부모 클래스
class Person :
    def __init__(self, name) :
        self.name = name
    def eat(self, food) :
        print(f"{self.name}이/가 {food}를 먹습니다.")
# Person 클래스의 속성
# - 데이터 속성 : name
# - 메소드 : eat() 

# 자식 클래스
class Student(Person) :
    # Person클래스의 자식 클래스(Person의 속성을 상속받음)

    # 자식 클래스의 생성자
    def __init__(self, name, school) :
        # self.name = name
        # 부모의 생성자를 호출하여 => 초기화 코드 재활용 가능
        super().__init__(name) # 부모의 생성자 매개변수 name에 값 전달
        self.school = school # 자식 클래스에서 추가된 데이터 속성
    # 자식 클래스에서 추가할 메소드
    def study(self) :
        print(f"{self.name}은/는 {self.school}에서 공부합니다")

# Student 클래스의 속성
# - 데이터 속성 : name(상속받은 것), school(추가된 것)
# - 메소드 : eat()(상속받은 것), study()(추가된 것)

# 부모 클래스 객체
pr = Person("홍길동")
pr.eat("떡볶이")

# 자식 클래스 객체
st = Student("해리포터", "호그와트")
st.study()
st.eat("감자") # 상속받은 메소드 
```

### 문제 2) 사각형(Rectangle) 클래스
- 데이터 속성 : 가로(width), 세로(height)
- 사각형의 넓이를 계산하는  메소드 : calculate_area
- 사각형의 둘레를 계산하는 메소드 :  calculate_perimeter
- 사각형의 가로, 세로를 사용자로부터 입력받아 메소드 호출

```python
class Rectangle :
    def __init__(self, width, height) :
        self.width = width
        self.height = height
    def calculate_area(self) :
        print(f"사격형의 넓이 : {self.width * self.height}")
    def calculate_perimeter(self) :
        print(f"사각형의 둘레 : {(self.width + self.height) * 2}")

in_width = int(input("가로 : "))
in_height = int(input("세로 : "))

re = Rectangle(in_width, in_height)
re.calculate_area()
re.calculate_perimeter()
```

```python
class Calculator :
    def __init__(self, num1, num2) :
        self.num1 = num1
        self.num2 = num2
    def plus(self) :
        print(f"덧셈 결과 : {self.num1 + self.num2}")
    def minus(self) :
        if self.num1 >= self.num2 :
            print(f"뺄셈 결과 : {self.num1 - self.num2}")
        else : 
            print(f"뺄셈 결과 : {self.num2 - self.num1}")
    def mul(self) :
        print(f"곱셈 결과 : {self.num1 * self.num2}")
    def div(self) :
        if self.num1 == 0 or self.num2 == 0 :                     
            print("두개의 숫자 중 0이 있다면 나눗셈을 실행할 수 없음!") 
        elif self.num1 >= self.num2 :                             
            print(f"나눗셈 결과 : {self.num1 / self.num2}")       
        else :                                                   
            print(f"나눗셈 결과 : {self.num2 / self.num1}")

in_num1 = int(input("첫번째 숫자 입력 : "))
in_num2 = int(input("두번째 숫자 입력 : "))

cal = Calculator(in_num1, in_num2)
print("****계산기 프로그램 실행****")

while True :
    print("연산내용을 선택해 주세요!")
    print("1번 : 덧셈, 2번 : 뺄셈, 3번 : 곱셈, 4번 : 나눗셈")
    choice = input("선택지 입력(1~4) :")

    if choice == "1" :
        cal.plus()
        break
    elif choice == "2" :                                          
        cal.minus()                                               
        break                                                     
    elif choice == "3" :                                          
        cal.mul()                                                 
        break                                                     
    elif choice == "4" :                                          
        cal.div()                                                 
        break
    else :
        print("1~4 사이의 번호를 다시 입력해주세요")
```
