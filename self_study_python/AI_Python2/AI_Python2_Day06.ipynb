### 타입 힌트(type hint)
- 파이썬 3.5버전 부터 도입된 기능
- 변수나 함수의 매개변수, 반환값의 타입을 명시하는 목적으로 사용
- 다른 타입의 값을 저장해도 오류 발생 x
    - => 오로지 타입 명시 목적으로 사용
#### 타입 힌트 문법
##### (1) 변수의 타입 힌트 문법
- 변수명 :타입 = 값
- 콜론(:)뒤에 타입을 명시
- 값은 나중에 저장 가능
- 타입과 저장할 값이 일치하지 않아도 실행된다
##### (2) 함수의 타입 힌트 문법
- def 함수명(매개변수: 매개변수 타입, ...) -> 반환타입 :
    함수 내용
    return 반환값
- 매개변수는 콜론(:) 뒤에 타입을 명시
- 반환 타입은 화살표(->) 뒤에 명시
- 반환값이 없는 경우 -> None 사용
#### 타입의 종류
##### (1) 컬렉션의 타입 힌트
: 대괄호 []를 사용하여 컬렉션이 담을 타입 명시
- List[타입] # 리스트가 넣을 타입
- Dict[키타입, 값타입] # 딕셔너리의 키와 값의 타입
- Tuple[타입, ...] # 튜플의 각 위치 타입
- Set[타입] # 세트가 넣을 타입
##### (2) 특수 타입
- Optional : None이 될 수 있는 타입
- Union : 여러 타입 중 하나
    - 변수명 : Union[str, int] => 문자열 혹은 정수가 저장될 변수
    - 변수명 : str | int # Union타입을 | 기호로 간단히 표현 가능
- Any : 모든 타입 허용
- None : 값이 없음
### 클래스(class)
- 공통된 요소들이 모여 있는 하나의 틀
- 여러 번수와 함수(메소드)를 묶어서 처리할 수 있는 것
#### 클래스 기본 개념
##### (1) 데이터 속성(data attribute)
- 값으로서 데이터를 정의하는 것
- (self.변수에 데이터가 될 값을 대입해서 사용 가능)
##### (2) 메소드(method)
- 클래스 안에 있는 함수는 메소드(method)라고 함
- 기능(행위)로서의 처리를 정의하는 것
##### (3) 속성(attribute)
- 데이터 속성과 메소드를 합쳐서 부르는 말
##### (4) 객체(Object) / 인스턴스(instance)
- 정의한 클래스를 실제로 사용하기 위해 사용
- 클래스로부터 생성되는 각각의 존재가 인스턴스
#### 클래스와 객체
- 클래스(class) : 똑같은 무엇인가를 계속 만들어 낼 수 있는 틀
    (붕어빵 틀)
- 객체(Object) : 클래스에 의해 만들어진 결과물
    (팥붕어빵, 슈크림 붕어빵, ...)
- 붕어빵 하나의 틀에서 나왔지만 붕어빵 하나 하나는 별도의 객체
- 붕어빵을 만드는 재료가 붕어빵의 특징을 결정하듯,
객체는 지정한 속성 값에 따라 고유한 특징을 갖게 됨
- 객체는 속성(상태, 특징)을 가지고 있으며 행위(동작, 기능)을
할 수 있는 대상
- 객체의 데이터 속성은 변수로 구현, 행위는 함수(메소드)로 구현
    - => 클래스를 통해서 객체를 만들고 활용하는 프로그래밍 언어를
    "객체 지향 프로그래밍" 언어 라고 함
#### 클래스 정의
: 새로운 클래스를 만드는 것을 의미
```
class 클래스명 :
    # 속성 : 데이터 속성, 메소드
    def 메소드명(self, 매개변수, ...) :
        # 처리를 나타내는 함수(메소드)
        self.변수 = 값 # 데이터를 나타내는 변수(데이터 속성)
        ...
        return ...
```
- 클래스명은 첫글자 대문자로 작성
- 해당 클래스의 속성은 들여쓰기(tab)안에서 작성
- self : 객체 구별용 매개변수(인스턴스 자신을 나타냄)
- self.변수명 => 여러 객체도 한번의 정의로 구별하여 다룰 수 있게됨
#### 클래스의 사용
- 추상적인 개념의 클래스를 사용하기 위해서는 객체를 만들어야 한다
- 객체(인스턴스)를 만드는 작업을 객체화라고 함
##### [객체화 방법]
- 객체명 = 클래스명()
    - ex) pr = Person()
- 만들어진 객체를 사용하여 클래스의 속성에 접근 가능하다
##### [각 속성에 접근]
- 데이터 속성에 접근
    - 객체명. 데이터속성명
    - ex) pr.name = "짱구"
    print(pr.name)
    - 객체명.메소드명()
    - ex) pr.set_name("짱구")
#### 클래스를 사용하는 이유
: 하나의 클래스를 통해 여러개의 객체를 생성하므로
코드의 반복을 줄일 수 있다
- 클래스의 메소드를 하나만 변경하더라도
모든 객체에 영향을 줄 수 있다

````python
# 변수에 타입 힌트 사용
name : str # name이라는 변수는 str(문자열)을 저장할 변수
name = "짱구"
print(name, type(name))

name = 100 
print(name) # 다른 타입의 값을 저장해도 오류 발생 x, 활용 타입 명시하는 목적으로 사용 

age : int = 10 # 정수 타입 명시
height : float = 150.5 # 실수 타입 명시
is_student : bool = True # 논리형 타입 명시 
# 함수에 타입힌트 사용

def greet(name : str) -> str : # 매개변수는 문자열, 반환값도 문자열
    return f"안녕하세요, {name}님!"

# 함수 호출
print(greet("짱구"))
````

````python
from typing import List, Dict, Tuple, Set
# 컬렉션의 타입 힌트

# 1) 리스트 타입 힌트
numbers : List[int] = [1, 2, 3, 4, 5] # 정수로 구성된 리스트
names : List[str] = ["짱구", "철수", "훈이"] # 문자열로 구성된 리스트

# 2) 딕셔너리 타입 힌트
student : Dict[str, str] = {"name" : "철수", "grade" : "A"}
# key와 value가 모두 문자열인 딕셔너리

# 3) 튜플 타입 힌트
tu : Tuple[int, int] = (10, 20) # 두개의 정수로 구성된 튜플

# 4) 세트 타입 힌트
se : Set[int] = {1, 2, 3} # 정수로 구성된 세트

print(numbers)
print(names)
print(student)
print(tu)
print(se)
````

````python
from typing import Optional, Union
# 특수 타입 명시

# Optional 타입(None이 될 수 있는 타입)
def get_name(user_id : int) -> Optional[str]:
    # 사용자를 찾으면 이름을 반환, 못 찾으면 None을 반환
    if user_id == 1 :
        return "둘리"
    return None

# Union 타입(여러 타입 중 하나)
def process_data(data : Union[str, int]) -> str :
    # 문자열이나 정수를 받아서 문자열로 반환
    return str(data)

# 함수 호출
print(get_name(1))
print(process_data(1))
````

## 클래스(class)

````python
# 사람의 표현하는 Person 클래스 정의
class Person :
    # 클래스명은 앞글자 대문자로 작성!
    # Person 클래스의 속성(데이터 속성, 메소드)

    # 1) 이름을 정의하는 메소드
    def set_name(self, name) :
        self.name = name # self.데이터속성 = 값
    #2) 나이를 정의하는 메소드
    def set_age(self, age) :
        self.age = age # self.데이터 속성 = 값
    # 3) 자기 자신을 소개하는 메소드(클래스의 데이터 속성 사용)
    def introduce(self) :
        print(f"안녕하세요. 저는 {self.name}입니다")
        print(f"제 나이는 {self.age}살 입니다")
    # 4) 객체의 self값을 출력하는 메소드
    def print_self(self) :
        print(self)
# Person클래스를 사용하기 위한 객체 생성
# : 객체명 = 클래스명()
pr = Person()

# 객체를 통한 속성 접근
# 1) 데이터 속성 : 객체.데이터속성
pr.name = "짱구" # self.name 데이터 속성에 "짱구" 저장
print(pr.name) # 데이터 속성에 저장된 값 사용

# 2) 메소드 : 객체.메소드명()
pr.set_age(5)
print(pr.age) # self.age에 저장된 값 출력
pr.introduce()
# 클레스라는 틀을 통해 여러 객체를 생성할 수 있다
p1 = Person()
p2 = Person()
p3 = Person()
p4 = Person()

print(p1) # 클래스를 복사한 값이 저장된 메모리 주소
p1.print_self() # self : 객체 자기 자신(메모리 주소)
print(p2)
print(p3)
print(p4)

# 객체들의 속성 접근
p1.name = "짱구"
# .(~의, ~안에)
# p1.name => p1의 메모리 주소 안에 있는 name
p1.age = 5

p2.name = "철수"
# p2.name => p2의 메모리 주소 안에 있는 name
p2.age = 6
# name, age 같은 이름을 쓰는데 다른 값을 넣을 수 있는 이유?
# => 객체가 저장된 메모리 주소가 서로 다르기 때문에

p3.set_name("훈이")
p3.set_age(8)

p4.set_name("짱아")
p4.set_age(2)

p1.introduce()
p2.introduce()
p3.introduce()
p4.introduce()
# <클래스 실습>
# 강아지를 표현하는 Dog클래스 정의
# Dog 클래스의 속성 : 
# - 강아지의 이름을 정하는 메소드 set_name() 정의
# - 강아지의 나이를 정의하는 메소드 set_age() 정의
# - 강아지의 이름과 나이를 사용하여 출력하는 메소드 get_walk() 정의
# => 출력예시 : 강아지 000은/는 00살이고 산책하는 것을 좋아한다 
# - 데이터 속성 : 이름(name), 나이(age)
# - Dog클레스의 객체 생성 후 메소드 호출

class Dog :
    def set_name(self, name) :
        self.name = name
    def set_age(self, age) :
        self.age = age
    def get_walk(self) :
        print(f"강아지 {self.name}은/는 {self.age}살이고 산책하는 것을 좋아한다")
dog = Dog()

dog.set_name("멍")
dog.set_age(5)

dog.get_walk()
````

### 문제1) 학생 클래스

- 학생의 이름과 점수를 속성으로 가지는 Student 클래스를 작성
- 데이터 속성 : 나이, 점수
- 메소드 3개
    - 학생의 이름을 정의하는 set_name() 정의
    - 학생의 점수를 정의하는 set_score() 정의
    - 학생의 이름과 정보를 출력하는 print_info() 정의
- 학생의 정보를 출력하는 메소드 호출까지

        <출력예시>
        학생 이름 입력 : 둘리
        학생 점수 입력 : 80
        이름:둘리
        점수: 80점


````python
class Student : 
    def set_name(self, name) :
        self.name = name
    def set_score(self, score) :
        self.score = score
    def print_info(self) :
        print(f"이름 : {self.name}")
        print(f"점수 : {self.score}")
st = Student()
st_name = input("학생 이름 입력 : ")
st_score = input("학생 점수 입력 : ")
st.name = st_name
st.score = st_score
st.print_info()
````

### 문제 2) 타입 힌트를 사용한 계산 함수 작성

- 두 개의 정수를 입력받아, 덧셈과 곱셈의 결과를 튜플로 반환하는 함수를 작성
- 이 함수에는 타입 힌트를 사용
- 타입 힌트를 활용하여 함수의 매개변수와 반환값의 타입을 명시할 것

        <함수 실행 및 결과 출력>
        print(calculate(3, 5))  # (8, 15)
        print(calculate(10, 2))  # (12, 20)


````python
from typing import Tuple
def calculate(num1 : int, num2 : int) -> Tuple[int, int] :
    add = num1 + num2
    mul = num1 * num2
    return (add, mul)
print(calculate(3, 5))
print(calculate(12, 20))
````

### 문제 3) 타입 힌트를 활용한 리스트 필터링 함수 작성
- 정수 리스트와 기준 값을 입력받아, 기준 값보다 큰 숫자들로만 이루어진 새로운 리스트를 반환하는 함수를 작성.
- 이 함수에는 타입 힌트를 사용

- 요구사항
    - 첫 번째 매개변수는 정수 리스트
    - 두 번째 매개변수는 비교할 기준 값(정수)
    - 기준 값보다 큰 숫자들만 포함된 리스트를 반환

- 타입 힌트를 사용하여 함수의 매개변수와 반환값을 명시할 것.

        <함수 실행 및 결과 출력>
        print(filter_numbers([1, 5, 8, 10], 5))  # [8, 10]
        print(filter_numbers([3, 7, 2, 9, 4], 4))  # [7, 9]


````python
from typing import List
def filter_numbers(numbers : List[int], number : int) -> List[int] :
    filter_li = []
    for num in numbers :
        if num > number :
            filter_li.append(num)
    return filter_li

print(filter_numbers([1, 5, 8, 10], 5))
print(filter_numbers([3, 7, 2, 9, 4], 4))
````
