### None타입
파이썬에서 None은 '아무것도 없음' 또는 '값이 없음'을
나타내는 특별한 타입
#### None의 특징
- 다른 타입과 연산 불가(오류 발생)
- None은 0이나 빈 문자열(""), 빈 컬렉션([], {}, ...) 등과는 다르다
- None은 정말로 아무것도 없음을 의미
#### None이 사용되는 경우
- 함수에서 아무것도 반환하지 않을 때
- 아직 저장할 값이 정해지지 않은 변수를 만들 때
- 찾는 값이 없을 때
#### None 확인 방법
- 값 is None : 값이 None이면 True, 아니면 False
- 값 == None도 기능하지만, is None이 권장사항
## 콜백함수

```python
# 파이썬에서 함수를 변수에 저장할 수 있다

# 함수 정의
def hello() :
    print("안녕~!")

# 함수를 변수에 저장
greeting = hello # 함수 이름 뒤에 ()를 붙이지 않고 저장
# print(greeting)

# 함수 호출
hello()
greeting() # greeting에 저장된 hello()함수 호출됨

# ** 파이썬에서 함수를 변수에 저장할 수 있는 이유
# - 파이썬은 함수를 일급 객체로 다루기 때문에
# - 일급객체 : 값처럼 사용되는 것, 변수에 저장(할당)할 수 있고
#              매개변수에 전달될 수 있고, 함수가 반활할 수 있는 객체를 가리키는 용어 
# 콜백함수

# 1) 콜백함수를 인수로 전달받아 사용하는 주함수 정의
def call_greeting(callback) :
    print("콜백함수를 실행하기 전입니다")
    callback() # callback변수에 저장된 함수 호출
    print("콜백함수를 실행한 후입니다")

# 2) 콜백함수 정의 : 다른 함수의 인수로 사용되는 함수
def hello() :
    print("안녕하세요")

def bye() :
    print("안녕히가세요!")

# 함수 호출 
call_greeting(hello) # 주함수를 호출하면서, hello()를 콜백함수로 사용, callback매개변수에 전달
#                       callback = hello
print("--------------------------")
call_greeting(bye) # bye 함수를 콜백함수로 사용 => callback = bye

# 콜백함수 : 동일한 함수를 다양한 상황에서 다르게 활용할 수 있다.
```

### 매개변수가 있는 콜백함수

```python
# 콜백함수도 일반 함수처럼 매개변수를 가질 수 있음

# 1) 주함수 정의 
def calculate(callback, x, y) :
    result = callback(x, y)
    # 전달받은 콜백함수를 호출하면서 x, y를 콜백함수에 전달
    # => 반환받은 값을 변수에 저장 
    print(f"결과 : {result}")

# 2) 콜백으로 사용할 함수(콜백함수)
def add(x, y) : 
    return x + y

def multiply(x, y) :
    return x * y 

# 함수 호출
calculate(add, 10, 20) # add를 콜백함수로 사용, 주함수에서는 add(10, 20)를 호출하여 반환받은
#                        값을 출력
calculate(multiply, 10, 20)
```

### 파이썬 내장 함수 중 콜백함수 사용하는 경우

```python
# 1) sorted()함수와 콜백함수

# sorted()
li = [1, 3, 5, 4]
print(sorted(li)) # 오름차순 정렬
print(sorted(li, reverse = True)) # 내림차순 정렬
# sorted(정렬할 데이터, key = 정렬 기준이 되는 값을 반환하는함수(콜백함수))

students = [
    # 학생들의 정보를 저장한 딕셔너리
    {"name" : "영희", "age" : 15},
    {"name" : "철수", "age" : 13},
    {"name" : "민수", "age" : 17} 
]
print(students, type(students))

# 리스트에 저장된 딕셔너리들을 학생의 나이 기준으로 정렬

# sorted()에 전달할 콜백함수 정의
# => 학생의 나이를 반환하는 함수 정의 
def get_age(student) :
    # student : 리스트 안에 있는 딕셔너리를 전달받을 매개변수
    return student["age"] # sorted()함수는 이 값을 기준으로 정렬

# sorted()함수에 콜백함수 get_age전달
# sorted()에서 리스트에 있는 딕셔너리를 하나씩 get_age()함수에 전달
# get_age()(콜백함수)가 반환하는 나이 값들을 기준으로 오름차순 정렬

sorted_stus = sorted(students, key = get_age)
print(f"나이 순으로 정렬된 학생들 : {sorted_stus}")

# 람다를 활용한 방법 (lambda 매개변수 : 반환값)

sorted_stus = sorted(students, key = lambda student : student['age'])
print(sorted_stus)
# 2) filter()함수와 콜백함수
# filter(조건함수, 반복가능객체)
# 반복가능객체 요소들을 하나씩 조건함수에 전달하여, True를 반환하는 요소들만 묶음 

numbers = [x for x in range(1, 11)]
print(numbers) # 1 ~ 10까지 저장된 리스트

# filter()와 콜백함수를 사용하여 numbers에서 짝수만 저장한 리스트 생성

# 짝수만 걸러내는 콜백함수
def is_even(num) :
    # num : numbers에 있는 정수를 하나씩 전달받을 매개변수
    return num % 2 == 0 # 짝수면 True, 아니면 False를 반환 

# filter()함수에 콜백함수 is_even전달
# numbers에 있는 요소들이 is_even에 하나씩 전달 
# is_even()함수가 반환하는 값이 True일 경우들의 결과를 하나로 묶음
even_nums = list(filter(is_even, numbers))
print(even_nums)

# 람다 활용
even_nums = list(filter(lambda x : x % 2 == 0, numbers))
print(even_nums)
# <콜백함수 실습>
fruits = ["apple", "banana", "orange", "kiwi", "watermelon"]
# 1) 콜백함수와 filter()를 사용하여 위의 리스트에서 6글자 이상인
# 과일만 필터링한 리스트 생성 후 출력

def len_fruits(le) :
    return len(le) >= 6
even_fruits = list(filter(len_fruits, fruits))

print(even_fruits)
students = [
    {"name" : "영희", "score" : 85},
    {"name" : "철수", "score" : 92},
    {"name" : "민수", "score" : 78},
    {"name" : "수진", "score" : 95}
]

# 2) 콜백함수를 사용하여 점수가 90점 이상인 학생들의 이름을 저장한 리스트 생성

# (1) 콜백함수와  filter()를 사용하여 90점 이상인 학생들
def is_high(student) :
    # student : 리스트 안의 딕셔너리를 전달받는 매개변수
    return student["score"] >= 90

high_stus = list(filter(is_high, students))
print(high_stus)

# (2) 고득점자 필터링 후 이름만 추출
names = [stu["name"] for stu in high_stus]
print(names)
# None : 비어 있음, 아무것도 없음을 나타내는 타입
print(None, type(None))

# 1) 다른 타입과 연산 불가
# print(None + 10) # 오류 발생
# print(None * 2)

# 2) None타입은 빈 문자열, 0, 빈 컬렉션과는 다르다
print(None == 0) # False
print(None == "")
print(None == [])

# +) None타입 확인 방법
# 값 is None : 특정 값이 None이니??? => 맞으면 True, 아니면 False
var = None
print(var is None)
```

### None이 사용되는 경우

```python
# 1) 함수에서 반환하는 값이 없을 때

def say_hello() :
    print("안녕하세요")
    # return None

result = say_hello() # 반환하는 값을 변수에 저장
print(result) # None
# 2) 나중에 값을 넣을 변수를 만들 때
name = None
print(name) # None

# 나중에 값 넣기
name = "철수"
print(name)
# 3) 찾는 값이 없을 때 

# 학생의 이름을 전달받아, 이름이 학생 명단에 있으면 "찾았습니다" 반환
# 못찾았으면 None을 반환하는 함수 정의
def find_student(name) :
    # 학생 명단
    students = ["철수", "영희", "민수"]

    if name in students :
        return "찾았습니다"
    return None

print(find_student("철수"))
print(find_student("짱구"))
# is None

di = {"a" : "apple", "b" : "banana"}
# 딕셔너리의 value 접근 방법
print(di["a"])
# print(di["c"]) => key가 없으면 오류 발생
print(di.get("a"))
print(di.get("c")) # => 없는 key를 작성하면 None이 반환됨
# 이름(문자열)을 전달받아, 이름에 해당하는 별명이 있으면
# 별명을 반환, 아니라면 None을 반환하는 함수 정의

def get_nickname(name) :
    # 이름과 별명이 저장된 딕셔너리 생성
    di = {"짱구" : "짱짱", "철수" : "철철"}
    return di.get(name) # .get(key) : key가 없으면 None을 반환

nickname = get_nickname("길동")
if nickname is None :
    print("별명이 없습니다")
else :
    print(f"별명 : {nickname}")
```

### 문제 1) 콜백함수 사용 사칙 연산 함수 정의
- 콜백 함수를 사용하여 사칙연산을 수행하는 operate() 함수를 정의
- operate() 함수는 두 개의 숫자와 연산을 수행할 함수(콜백함수)를 매개변수로 받아 연산결과를 반환
- 필요 콜백함수 정의
    - add(x, y) : 덧셈 반환 (x + y)
    - subtract(x, y) : 뺄셈 반환 (x - y)
    - multiply(x, y) : 곱셈 반환 (x * y)
    - divide(x, y) : 나눗셈 반환 (x / y)
- operate(x, y, 콜백함수) 함수를 정의

        <출력예시>
        --- 함수 호출 및 결과 출력 ---
        10 + 5 = 15
        10 - 5 = 5
        10 x 5 = 50
        10 / 5 = 2.0


```python
def operate(callback, x, y) :
    result = callback(x, y)
    print(result)

def add(x, y) :
    return (f"{x} + {y} = {x + y}")

operate(add, 10, 5)

def subtract(x, y) :
    return (f"{x} - {y} = {x - y}")

operate(subtract, 10, 5)

def multiply(x, y) :
    return (f"{x} * {y} = {x * y}")

operate(multiply, 10, 5)

def divide(x, y) :
    return (f"{x} / {y} = {x / y}")

operate(divide, 10, 5)
```

### 문제 2) 학생정보 점수순으로 정렬
- 학생들의 이름과 점수를 튜플 리스트로 저장한 후,
- 점수를 기준으로 오름차순 정렬하는 프로그램을 작성
- 필요 콜백함수
    - key_func(student) : 학생 한명의 정보를
    저장하고 있는 튜플을 전달받아 점수만 반환하는 콜백함수 정의
- sort_students(student_list, key_func) 함수를 정의
    - 학생들의 정보를 저장하고 있는 리스트와
    콜백함수를 전달받아, 정렬된 결과를 반환하는 함수 정의

            <출력 예시>
            정렬 전 : [('둘리', 85), ('또치', 92), ('도우너', 78)]
            정렬 후 : [('도우너', 78), ('둘리', 85), ('또치', 92)]


```python
students = [
    ('둘리', 85),
    ('또치', 92),
    ('도우너', 78)
]
print(f"정렬 전 : {students}")

def key_func(student) :
    return student[1]

sort_students = sorted(students, key = key_func)

print(f"정렬 후 : {sort_students}")
```

### 문제 3) 학생리스트에서 특정 학생 찾기(None타입)
- 학생들의 이름을 저장하고 있는 리스트와
찾고자하는 학생의 이름(문자열)을 전달받아
- 학생의 이름이 리스트에 있으면 학생의 이름을, 없으면 None을 반환하는 search_name() 함수 정의
- search_name() 함수 호출 후,
    - 반환받은 값이 None값이라면 "{이름}이 없습니다" 출력
    - 반환받은 값이 있다면 "{이름}을 찾았습니다" 출력

            <출력예시1>
            학생 리스트 : ['둘리', '또치', '도우너']
            둘리을/를 찾았습니다!

            출력예시2>
            학생 리스트 : ['둘리', '또치', '도우너']
            길동이/가 없습니다


```python
names = ['둘리', '또치', '도우너']

def search_name(name) :
    if name in names :
        return name
    return None

se_name = ("둘리")
name = search_name(se_name)
if name is None :
    print(f"{se_name}이/가 없습니다")
else :
    print(f"{name}을/를 찾았습니다")

se_name = ("길동")
name = search_name(se_name)
if name is None :
    print(f"{se_name}이/가 없습니다")
else :
    print(f"{name}을/를 찾았습니다")
```
