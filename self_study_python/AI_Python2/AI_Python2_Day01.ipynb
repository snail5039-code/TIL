### 내장함수

#### 내장함수의 종류

##### 숫자 내장함수 (정수 or 실수)
- abs() : 전달된 값의 절대값을 반환
    - -10의 절대값 => 10
    - 10의 절대값 => 10
- divmod() : 전달된 두 숫자를 나누어서 몫과 나머지를
  튜플 형태로 반환 (몫, 나머지)
- max() : 전달된 값들 중 가장 큰 값을 반환
- min() : 전달된 값들 중 가장 작은 값을 반환
- sum() : 전달된 리스트나 튜플과 같은 반복가능 객체 요소들의 합을 반환
- pow() : 전달된 두 숫자의 거듭제곱 결과를 반환(** 연산자와 동일)
- round() : 전달된 값의 반올림한 값을 반환
    - 전달된 값이 1개라면, 정수로 반올림한 값을 반환
    - 전달된 값이 2개라면, 두번째로 전달된 값만큼 소수점을 남겨둠

##### 시퀀스 내장 함수
- : 시퀀스 자료형과 사용할 수 있는 내장 함수
- 시퀀스 자료형 : 값이 연속적으로 이어진 자료형
    - ex) 문자열, 리스트, 튜플, ...
- enumerate() : 리스트와 함께 사용할 수 있음
    - 리스트의 요소와 해당 인덱스 번호가 튜플 형태로 묶임
    - (idx, 값)
- range() : 전달된 값에 따라 연속된 정수 범위를 생성
    - 특정 범위의 값을 생성하기 때문에 범위 생성자
- len() : 전달된 객체의 길이(항목 수)를 반환하는 함수
- sorted() : 전달된 반복 가능 객체 요소들을 오름차순 정렬
    - sorted(반복가능객체, reverse = True) :
        - 내림차순 정렬
    - 리스트.sort() => 오름차순 정렬 메소드
    - 리스트.sort(reverse = True) => 내림차순 정렬 메소드
- zip() : 여러개의 반복가능 객체들을 하나로 압축
    - 각 순서에 맞는 요소들을 튜플로 묶어서 반환

## 내장 함수
### 숫자형 내장 함수

```python
# abs() : 절대값 반환
print(abs(-10)) # 10
print(abs(10))
print(abs(-3.14)) # 3.14

# divmod() : (몫, 나머지)

# 연산자
print(10 // 3) # 3(몫)
print(10 % 3) # 1(나머지)

print(divmod(10, 3), type(divmod(10, 3)))
print(divmod(20, 4))

# min(), max(), sum()

li = [10, 20, 30]
print(f"{li}안에서 가장 큰 값 : {max(li)}")
print(f"{li}안에서 가장 작은 값 : {min(li)}")
print(f"{li}안의 요소들의 합 : {sum(li)}")

print(max(1, 2, 3))
print(min(1, 2, 3))
# print(sum(1, 2, 3)) # sum()은 리스트나 튜플 같은 반복가능 객체를 전달받음 
# pow() : 거듭제곱 함수
# ** 연산자
print(10 ** 2) # 10 * 10
print(pow(10, 2)) # 10 * 10

print(9 ** 0.5) # 9의 제곱근
print(pow(9, 0.5)) 
# round() : 반올림한 값을 반환하는 함수
print(round(1.5)) # 2
print(round(3.14, 1)) # 두번째 숫자는 소수점 자리수

print(round(2.5)) # 앞의 자리수가 짝수라면 버림 : 오사오입
```

## 시퀀스 내장 함수

```python
# enumerate() : 리스트와 사용   

li = ["가위", "바위", "보"]
print(li, type(li))

for i in enumerate(li) :
    print(i, type(i)) # (idx, 값) 튜플 형태

for idx, element in enumerate(li) :
    print(f"idx : {idx}, element : {element}")
# range()

print(range(1, 10, 2), type(range(1, 10, 2)))

# 연속된 정수 값이 저장된 반복 가능 객체 생성 
li = list(range(1, 10, 2)) # 리스트로 형변환
print(li, type(li))

# 빈 리스트 생성
li1 = []
li2 = list()

print(list(range(0)))
# len() 
li = ["a", "b", "c"]
str1 = "hello"
di = {"a" : "apple", "b" : "banana"}

print(len(li)) # 3
print(len(str1)) # 5(문자열의 길이)
print(len(di)) # 2, 딕셔너리는 key와 value가 항상 한쌍이기 때문에
print(len(range(100))) # 0 ~ 99
# sorted()
li = [5, 2, 7, 1, 8, 2, 3]
print(li, type(li))

print(sorted(li)) # 원본은 변경되지 않음
print(li)

# 리스트.sort() : 원본 자체가 정렬됨
li.sort()
print(li)

# 내림차순
print(sorted(li, reverse = True))
# zip()

name_li = ["짱구", "철수", "훈이", "맹구"]
score_li = [74, 100, 80, 50]

# 학생들의 이름과 점수를 짝지어서 출력
print(f"{name_li[0]}의 점수는 {score_li[0]}점 입니다")
print(f"{name_li[1]}의 점수는 {score_li[1]}점 입니다")
print(f"{name_li[2]}의 점수는 {score_li[2]}점 입니다")
print(f"{name_li[3]}의 점수는 {score_li[3]}점 입니다")

# for문과 zip() 함수 사용
print("===== for와 zip() =====")
for i in zip(name_li, score_li) :
    print(i, type(i))

print("===== 튜플의 언패킹 =====")
for name, score in zip(name_li, score_li) :
    print(f"{name}의 점수는 {score}점 입니다")
# <내장 함수 실습>
# 월별 일수 리스트가 있을 때 1월 ~ 12월의 월과 마지막일 같이 출력 
day = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

# <출력예시>
# 1월 : 31일
# 2월 : 28일 
# ...

# enumerate(), zip()을 활용

# 방법 1) enumerate()
for idx, element in enumerate(day, start = 1) :
    print(f"{idx}월 : {element}일")

# 방법 2) zip()
print("===== zip() =====")
#for i in zip(range(1, 13), day) :
#    print(i)

for i, j in zip(range(1, 13), day) :
    print(f"{i}월 : {j}일")
```

## 리스트 컴프리헨션

```python
# [요소 for 변수 in 집합 if 조건식]

# 1) 리스트 컴프리헨션으로 1 ~10까지의 정수가 저장된 리스트 생성
li = [x for x in range(1, 11)]
print(li)

# 아래 내용을 줄여서 작성한 것 
li = []
for i in range(1, 11) :
    li.append(i)
print(li)
# 2) 1 ~ 10까지의 정수에 * 2를 하고 리스트에 저장
li = [x * 2 for x in range(1, 11)]
print(li)

# 아래 코드를 줄여서 작성한 것
li = []
for i in range(1, 11) :
    li.append(i * 2)
print(li)

# 3) 학생1 ~ 학생10까지의 문자열이 저장된 리스트 생성
li = [f"학생{x}" for x in range(1, 11)]
print(li)
# 4) 1 ~ 10까지 정수 중 짝수만 저장된 리스트 생성
li = [x for x in range(1, 11) if x % 2 == 0]
print(li)

# 아래 코드를 줄여서 작성한 것
li = []
for i in range(1, 11) :
    if i % 2 == 0 : 
        li.append(i)
print(li)
# <리스트 컴프리헨션 실습>
# 리스트 컴프리헨션으로 1 ~ 20까지 정수 중 3의 배수로만 구성된 리스트 생성
li = [x for x in range(1, 21) if x % 3 == 0]
print(li)

li = [x for x in range(3, 21, 3)]
print(li)
```

## 문제2) 리스트에 저장된 데이터를 아래와 같이 출력하기.
## apart = [ [101, 102], [201, 202], [301, 302] ]

        [결과]
        301 호
        302 호
        201 호
        202 호
        101 호
        102 호

```python
apart = [ [101, 102], [201, 202], [301, 302] ]
print(apart)

for i in sorted(apart, reverse = True) :
    for j in i :
        print(f"{j} 호")
```

## 문제3) 무지개의 일곱 색깔을 저장하는 rainbow 리스트를 생성하고 다음과 같이 출력하기

        <출력결과>
        무지개의 1번째 색은 red입니다.
        무지개의 2번째 색은 orange입니다.
        무지개의 3번째 색은 yellow입니다.

## 문제4) 학생들의 점수를 담고 있는 리스트

scores = [75, 90, 45, 60, 30, 80, 50]
