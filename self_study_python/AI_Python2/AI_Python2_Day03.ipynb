````python
# return(반환값, 리턴값)이 있는 함수 정의

# 1) 2개의 정수를 전달 받아, 두 정수의 합을 반환하는 add()함수 정의
def add(num1, num2) :
    return num1 + num2

# 함수 호출(return이 있는 함수)
# : 함수명(인수, ...)
print(add(10, 20))
# add()라는 함수를 호출할 때, 10과 20(인수)를 함수에 전달하고
# 함수에서는 전달받은 값으로 10 + 20 한 다음 함수를 호출한 곳에 반환한다

# ** 반환 받은 값은 값처럼 사용할 수 있다 
# 1) 변수 = 값 
result = add(10, 20) # 함수를 호출하고 반환 받은 값 30을 변수에 저장 => result = 30
print(result)

# 2) 값 + 값 : 연산 가능
print(add(30, 40) + add(10, 20)) # 70 + 30

# +) return이 있는 함수만 함수를 호출하는 쪽을 값처럼 사용할 수 있다 

# <컬렉션을 반환하는 함수>
# 1) 3개의 정수를 전달받아, 전달받은 정수들을 저장한 리스트를 반환하는 get_list()함수 정의
def get_list(num1, num2, num3) :
    li = [num1, num2, num3] # 전달받은 정수 3개를 저장한 리스트
    return li
# 함수 호출
print(get_list(1, 2, 3), type(get_list(1, 2, 3)))
result = get_list(1, 2, 3) # 반환받은 리스트를 변수에 저장 
print(result)

# 2) 2개의 정수를 전달받아, 전달받은 정수들의 합과 곱을 반환하는 get_nums() 함수 정의

def get_nums(num1, num2) :
    return num1 + num2, num1 * num2 # 튜플의 패킹 

# 함수 호출
print(get_nums(10, 20), type(get_nums(10, 20)))
# => (합, 곱) 튜플을 반환받음
add_result, mul_result = get_nums(10, 20) # 튜플의 언패킹
print(add_result)
print(mul_result)

# 3) 이름과 나이를 전달받아, 전달받은 이름과 나이로 구성된 딕셔너리를 반환하는 get_dict() 함수 정의

def get_dict(name, age) :
    return {"이름" : name, "나이" : age}

print(get_dict("짱구", 5))
di = get_dict("유리", 7) # 반환받은 딕셔너리를 저장
print(di, type(di))
print(di["이름"])

# +) return은 함수의 종료를 의미
def exit() :
    print("return은")
    return
    print("함수의 종료를 의미한다")

exit()

# 정수를 1개를 전달받아, 정수가 짝수라면 "짝수" 반환 그게 아니라면 "홀수" 반환
def is_even(num) :
    if num % 2 == 0 :
        return "짝수" # "짝수"를 반환하면서 함수 종료
    return "홀수"

print(is_even(10))
print(is_even(9))

# <return이 있는 함수 실습>
# 1) 2개의 정수를 전달받아, 전달받은 정수들 중
# 큰수 - 작은수, 큰수 / 작은수 2개의 값을 반환하는 get_cal()함수 정의 후 호출 

def get_cal(num1, num2) :
    if num1 > num2 :
        return num1 - num2, num1 / num2
    return num2 - num1, num2 / num1
print(get_cal(4, 3))
sub, div = get_cal(20, 10) # 튜플의 언패킹
print(f"뺄셈 결과 : {sub}")
print(f"나눗셈 결과 : {div}")

# 2) 정수들이 저장된 리스트 1개를 전달받아, 전달받은 리스트 요소들의 합과 평균이 
# 저장된 딕셔너리를 반환하는 get_di()함수 정의 후 호출

def get_di(li) : 
    # li : 정수들이 저장된 르스트
    li_sum = sum(li)
    li_avg = sum(li) / len(li)
    return {"합계" : li_sum, "평균" : li_avg}

# 함수 호출
li = [x for x in range(1, 11)] # 1 ~ 10이 저장된 리스트(리스트 컴프리헨션)
print(li)

di = get_di(li)
print(di, type(di))
print(di["합계"])
print(di["평균"])
````

## 전역변수 vs 지역변수

````python
glo_val = 100 # 전역변수(어디서든 사용 가능)

# 함수 정의
def print_value(param) :
    loc_val = 200 # 지역변수(해당 함수 안에서만 사용 가능)
    print(f"전역변수 : {glo_val}")
    print(f"지역변수 : {loc_val}")
    print(f"매개변수 : {param}") # 매개변수(일종의 지역변수)

# 함수 호출
print_value(300)
print(f"함수 밖 전역변수 : {glo_val}")
# print(f"함수 밖 지역변수 : {loc_val}") # 지역변수는 함수 밖에서 사용 불가
# print(f"함수 밖 매개변수 : {param}") # 매개변수도 함수 밖에서 사용 불가

glo_val2 = 100 # 전역변수

# 함수 정의
def print_value2() :
    glo_val2 = 300 # 지역변수(함수 안에서 지역변수와 전역변수의 이름이 같다면 지역변수가 우선)
    print(glo_val2)

# 함수 호출
print_value2()
print(f"함수 밖 : {glo_val2}") # 100(전역변수에 저장된 값 변경 x) 

# 함수를 사용하여 전역변수에 저장된 값 수정
# 1) 함수 내에서 global 키워드 사용 
glo_val3 = 100 # 전역변수

# 함수 정의
def print_value3() :
    global glo_val3 # 전역변수를 그대로 사용한다
    glo_val3 = 300 # 전역변수에 저장된 값이 300으로 수정됨
    print(f"함수 안 : {glo_val3}")

# 함수 호출
print_value3()
print(f"함수 밖 : {glo_val3}")

# 2) 매개변수와 리턴을 사용하는 방법
glo_val4 = 100

# 함수 정의
def change_value(value) : 
    value += 100
    return value # 100을 증가 후 반환

glo_val4 = change_value(glo_val4) # 100을 증가 후 반환받은 값을 변수에 다시 저장
print(glo_val4)
````

## 모듈

````python
import random
# 랜덤(난수)과 관련된 기능을 가진 모듈

# 1) randint(a, b) : a이상 b이하의 숫자 중 랜덤한 숫자를 반환
print(random.randint(1, 10)) # 1 ~ 10 중 랜던한 숫자를 반환
# .(~의, ~안에) random.randint() => random 모듈 안의 randint()함수 호출

# 2) randrange()
# randrange(a) : 0 ~ a-1 사이의 숫자 중 랜덤한 겂을 return
# randrange(a, b) : a ~ b-1 사이의 숫자 중 랜덤한 값을 return
# rangrange(a, b, c) : a, a+c, a + 2c, ... b-1 사이의 숫자 중 랜덤한 값을 return
print(random.randrange(10)) # 0 ~ 9
print(random.randrange(2, 10)) # 2 ~ 9
print(random.randrange(2, 10, 3)) # 2, 5, 8

# 3) choice(반복가능객체) : 반복가능 객체 요소 중 랜덤한 값 1개를 return
li = ["봄", "여름", "가을", "겨울"]
print(random.choice(li))

# 4) sample(반복가능객체, 개수) : 반복가능 객체 요소 중 지정한 개수만큼 return
print(random.sample(li, 2))

# 5) shuffle(반복가능객체) : 반복가능객체 요소들의 순서를 랜덤하게 섞는 함수
random.shuffle(li)
print(li)

# 모듈에서 특정 함수만 import
from random import randint, randrange

print(randint(1, 5))
# 특정 함수만 import 하는 경우는 모듈명 없이 호출
print(randrange(10))
# print(choice([1, 2, 3]))

import time
# 시간과 관련된 기능을 가진 모듈 

# 1) sleep(초) : 해당 초만큼 실행을 멈추는 함수
print("안녕하세요")
# time.sleep(3) # 3초 딜레이
print("안녕히가세요")

# 2) time() : 1970년 1월 1일 0시 0분 0초 부터 현재까지 흐른 초
print(time.time())

# time()함수 활용 : 프로그램 실행시간 계산
import time

start = time.time()
# 실행 시간을 계산할 코드
li = []
for i in range(100000) : 
    li.insert(0, i) # 0번째 위치에 i값을 저장

end = time.time()

print(f"실행 시간 : {end - start}")

import math
# 수학 연산과 관련된 기능을 가진 모듈

# 1) 올림, 내림 함수
print(math.ceil(3.14)) # 4
print(math.floor(3.79)) # 3

# 2) 팩토리얼 반환 함수
# 3! => 1 x 2 x 3 => 6
print(math.factorial(5)) # 5! 1 x .. x 5 => 120

# 3) 제곱, 제곱근 반환 함수
print(math.pow(3, 2)) # 9 (모듈)
print(pow(3, 2)) # 9(내장 함수)
print(3 ** 2) # 9(연산자)

print(math.sqrt(9)) # 9의 제곱근 => 3
print(math.sqrt(25)) # 25의 제곱근 => 5
````

## 문제 1)  뺄셈 연산을 하는 함수 정의
- 2개의 숫자를 전달받아
- 전달받은 숫자들의 뺄셈 연산 결과를 반환하는 함수 sub() 정의 후 호출
- 단, 2개의 정수 중 큰수에서 작은수를 뺀 결과를 반환하도록 작성(큰수 - 작은수)

```
        <출력예시>
        첫번째 정수 입력 : 10
        두번째 정수 입력 : 5
        10과 5의 차이는 5
```

````python
def sub(num1, num2) :
    if num1 > num2 :
        return num1 - num2
    return num2 - num1
num1 = int(input("첫번째 정수 입력 : "))
num2 = int(input("두번째 정수 입력 : "))
print(f"{num1}과 {num2}의 차이는 : {sub(num1, num2)}")
````

## 문제 3) Lotto 프로그램 만들기(사용자 정의함수, 모듈)

- 로또는 최대 5번의 게임을 진행할 수 있다
- 한 게임당 번호는 1 ~ 45까지의 숫자 중 7자리의 숫자를 랜덤으로 가져온다
- 게임당 숫자는 중복되지 않으며 게임당 7개의 숫자는 오름차순 정렬을 한다

- <게임 진행 순서>
    - 사용자로부터 게임 횟수를 입력받는다(1 ~ 5)
    - 1 ~ 5를 입력받으면 로또번호를 출력한다
    - 입력받은 정수가 범위를 벗어나면 다시 횟수를 입력받는다

```
            <출력예시1>
            Lotto 프로그램을 시작 합니다
            게임 횟수 입력 >> 5
            [12, 18, 21, 23, 32, 35, 44]
            [8, 15, 19, 27, 30, 33, 41]
            [4, 8, 21, 31, 35, 40, 43]
            [4, 11, 16, 23, 27, 31, 33]
            [3, 7, 20, 24, 32, 34, 43]
            프로그램 종료

            <출력예시2>
            Lotto 프로그램을 시작 합니다
            게임 횟수 입력 >> 10
            게임은 최소 1번 최대 5번 진행 가능!
            게임 횟수 입력 >>
```

````python
import random

print("Lotto 프로그램을 시작 합니다")
while True :
    cnt = int(input("게임 횟수 입력 >> "))
    if 0 < cnt < 6 :
        break
    else :
        print("게임은 최소 1번 최대 5번 진행 가능!")
for i in range(cnt) :
    lotto = random.sample(range(1, 46), 7)
    lotto.sort()
    print(lotto)
print("프로그램 종료")
````

## 문제 4) 카페 쿠폰 스탬프 프로그램 만들기(지역변수, 전역변수)
- 주문한 음료수의 개수 만큼 스템프를 찍어주는 프로그램
- 전역변수 stamp의 초기값은 0으로 저장
- 사용자로부터 음료수의 개수를 입력받고 입력받은 정수만큼
  스템프를 찍어준다
- 스템프를 찍어주는 함수 change_stamp() 정의
    - 전역변수 stamp에 저장된 값을 1씩 증가

```
            <출력예시1>
            주문 전 스템프 : 0
            주문할 음료수 개수 입력 : 5
            적립된 스템프 : 5

            <출력예시2>
            주문 전 스템프 : 0
            주문할 음료수 개수 입력 : 3
            적립된 스템프 : 3
```

````python
stamp = 0
def order() :
    global stamp
    print(f"주문 전 스템프 : {stamp}")
    cnt = int(input("주문할 음료수 개수 입력 : "))
    stamp += cnt
order()
print(f"적립된 스템프 : {stamp}")
````
