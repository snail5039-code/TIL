#### 매개변수
    f(x) = x   +   2

#### 용어
- (1) 함수 정의 : 사용자 정의 함수를 새로 만드는 것을 의미
- (2) 인수 : 함수에 전달되는 실제 값(input)을 의미
- (3) 매개변수 : 인수를 받아서 저장하는 변수를 의미, parameter
- (4) 반환값(리턴값) : 사용자 정의 함수의 결과(output)를 의미, return
- (5) 함수 호출 : 함수를 실제로 사용하는 것을 의미

#### 사용자 정의 함수 정의
- : 함수를 만드는 것을 의미, def 키워드를 사용

        def 함수명(매개변수, ...) :
            (들여쓰기, tab) 함수의 기능
            ...
            return 반환값(리턴값)

- 함수명 : 사용자 임의로 작성(동사로 작성)
- 매개변수 : 함수에 전달되는 인수를 저장할 변수
    - 전달되는 값이 없다면 매개변수를 작성하지 않아도됨
- 함수의 기능은 들여쓰기 안에 작성
- return(반환값) : 함수를 호출(사용)하는 쪽에 반환하는 값
    - 반환할 값이 없다면 작성하지 않아도됨

      디폴트 값을 설정한다!
- (2) 가변 매개변수

```python
# 사용자 정의 함수 정의

# def 함수명() :
# (들여쓰기) 기능

# 간단한 인사말을 출력하는 welcome() 함수 정의
def welcome() :
    print("hello~!")
    print("안녕~!")

# 함수 호출 : 정의한 함수를 사용 
# => 함수명()
welcome()
welcome()
# 코드의 재사용성을 높이기 위해 사용 
# 매개변수가 포함된 함수 정의
# 1) 정수 1개를 전달받아, 전달된 정수에 + 10을 한 결과를 출력하는 add_ten() 함수 정의
def add_ten(num) : 
    # num : 전달받을 정수를 저장할 매개변수
    print(num + 10)

# 2) 정수 2개를 전달받아, 전달받은 정수들의 합을 출력하는 add()함수 정의
def add(num1, num2) :
    # num1, num2 : 전달받을 2개를 저장할 매개변수
    print(num1 + num2) 

# 함수 호출(매개변수가 있는 함수)
# => 함수명(인수, ...)

add_ten(20) # 함수를 호출하면서 함수의 매개변수 num에 20(인수)을 전달하면 
# 함수에서는 전달 받은 값(20)을 + 10을 한 결과를 출력

add(10, 20) # num1과 num2에 각각 10과 20을 전달하여 더한 값을 출력
# add(30) # 매개변수와 인수의 개수는 일치해야 한다
# 3) 이름과 나이를 전달받아 전달받은 이름과 나이를 출력하는 introduce()함수 정의
def introduce(name, age) :
    print(f"이름 : {name}")
    print(f"나이 : {age}살")

# 함수 호출
introduce("짱구", "5") # 매개변수 name, age에 순차적으로 인수가 전달됨
# introduce(5, "짱구")
introduce(age = 5, name = "짱구") 
# 매개변수에 전달될 인수를 지정할 수 있음 => 순서를 맞추지 않아도됨
# 디폴트 매개변수
# : 디폴트 값은 맨 뒤의 매개변수 부터 설정을 해줘야 한다!
# 숫자 4개를 더한 결과를 출력하는 함수 정의
def add(num1, num2, num3 = 30, num4 = 40) :
    print(num1 + num2 + num3 + num4)

# 함수 호출
add(10, 20, 30, 40)
add(10, 20, 30) # num4에는 기본값 40이 사용됨
add(10, 20, 40)
# 가변 매개변수 : 몇 개의 인수를 사용할지 정확히 모를 때 사용

def show(*args) :
    print(f"args : {args}") # 전달되는 인수들을 하나의 튜플로 묶음
    print(f"타입 : {type(args)}")

# 함수 호출
show(1, "abc", 40)
show("python", 100)
# <사용자 정희 함수 실습>
# 1) 3개의 정수를 전달받아, 전달받은 정수들의 평균을 출력하는 print_avg()함수 정의 후 호출

# def print_avg(num1, num2, num3) :
#     print((num1 + num2 + num3) // 3)

def print_avg(num1, num2, num3) :
    result = (num1 + num2 + num3) / 3
    print(f"평균 : {result : .2f}")

def print_avg2(li) :
    # li : 3개의 정수가 저장된 리스트를 전달받을 매개변수
    # => 매개변수에 컬렉션, 시퀀스 자료형과 같은 반복가능 객체도 전달가능
    result = sum(li) / len(li) # 합계 / 길이(요소의 개수)
    print(f"평균 : {result : .2f}")

print_avg(10, 20, 30)
print_avg2([11, 12, 15])
# 2) 문자열 1개를 전달받아, 전달받은 문자열을 거꾸로 출력하는 reverse_str()함수 정의 후 호출

def reverse_str(str1) :
    print(str1[::-1]) # 슬라이싱 간격에 음수를 설정하면 거꾸로 슬라이싱

reverse_str("안녕하세요")
```

## 문제2) 자판기 프로그램

- 700원짜리 음료수를 뽑을 수 있는 자판기 프로그램 구현(함수 정의)
- 돈을 넣으면 몇잔의 음료수를 뽑을 수 있는지
- 그리고 잔돈은 얼마인지 모든 경우의 수를 출력하도록 구현

- 함수 정의
    - 반환값 : 없음
    - 함수 이름 : vending_machine()
    - 매개변수 : 정수 money

            <출력예시>
            투입할 금액 입력 : 3000
            음료수 = 0개, 잔돈 = 3000원
            음료수 = 1개, 잔돈 = 2300원
            음료수 = 2개, 잔돈 = 1600원
            음료수 = 3개, 잔돈 = 900원
            음료수 = 4개, 잔돈 = 200원

```python
def vending_machine(money) :
    max_count = money // 700
    for count in range(0, max_count + 1) :
        change = money - (count * 700 )
        print(f"음료수 = {count}개, 잔돈 = {change}원")

in_money = int(input("투입할 금액 입력 : "))
vending_machine(in_money)
```

## 문제3) 점수의 합계, 평균구하기
- 학생들의 인원수와 인원수 만큼의 점수를 입력받아 리스트에 저장
- get_sum_avg() 함수 정의
    - 학생들의 점수를 저장하고 있는 리스트를 전달받아,
    - 점수의 총합과 평균점수를 반환하는 함수
- 함수 호출 후 점수의 총합과 평균 점수를 출력

        <출력예시>
        학생들의 인원수 : 3
        1번째 학생의 점수 입력 >> 85
        2번째 학생의 점수 입력 >> 90
        3번째 학생의 점수 입력 >> 76
        ========결과 출력======
        (251, 83.66666666666667) <class 'tuple'>
        학생 3명의 총합 점수 : 251점
        학생 3명의 평균 점수 :  83.67점

```python
count = int(input("학생들의 인원수 : "))

score_li = []

for i in range(1, count + 1) :
    score = int(input(f"{i}번째 학생의 점수 입력 >> "))
    score_li.append(score)
print("=====결과 출력=====")

def get_sum_avg(li) :
    total = sum(li)
    avg = total / len(li)
    return total, avg

result = get_sum_avg(score_li)
print(result, type(result))

print(f"학생 3명의 총합 점수 : {result[0]}")
print(f"학생 3명의 평균 점수 : {result[1] : .2f2}")
```

## 문제4) 전화번호 형식 변경하는 함수 정의
- "010-1234-5678" 형태의 전화번호를 전달받아
- 전달받은 전화번호를 "-"을 "."으로 변경하여
- "010.1234.5678" 형태로 출력하는 프로그램을 작성

        <출력예시>
        변경 전 : 010-1234-5678
        변경 후 : 010.1234.5678

```python
phone = input("전화번호 입력 : ")

def change(phone) :
    return phone.replace("-", ".")

print(f"변경 전 : {phone}")

print(f"변경 후 : {change(phone)}")
```
