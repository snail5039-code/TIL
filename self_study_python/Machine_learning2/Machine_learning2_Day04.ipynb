```python
day04

0. 복습

이상치 처리
- 이상치 : 다른 값들과 동떨어지게 유난히 크거나 작은 값

1) 이상치 찾기
(1) 상자 그래프 시각화(boxplot)

(2) 통계적인 방법 - IQR
- IQR = Q3 - Q1
- Q3 + 1.5 x IQR 보다 크거나
  Q1 - 1.5 x IQR 보다 작은 값을 이상치로 본다

2) 이상치 처리
(1) 제거
(2) 대체
(3) 유지

구간분할
 : 연속된 숫자 데이터를 몇 개의 구간으로 나눠 범주로 바꾸는 것

1) pd.cut() : 값의 크기 기준으로 나눈다
	ex) 0 ~ 18, 18 ~ 35
2) pd.qcut() : 개수 기준으로 나눈다(각 구간에 같은 인원수가 들어가게)

1. 스케일링(scailing)
 : 열마다 다른 값의 크기 범위를 비슷하게 맞춰주는 작업
- 데이터의 각 열(특성) 단위와 범위가 다르다
- 값의 크기가 큰 열은 그 자체로 "더 중요한 것처럼" 취급되기 쉽다
- 많은 머신러닝 모델은 값이 큰 데이터에 크게 영향을 받는다 
ex) 키(cm), 몸무게(kg), 연봉(원)을 그대로 놓고 비교하면,
    숫자 큰 연봉만 눈에 들어온다.
    키 1cm 차이와 연봉 1원 차이를 같은 무게로 볼수는 없다
=> 스케일링은 이렇게 단위가 다른 값들을 공정하게 비교할 수 있도록
   크기를 맞춰주는 것

1) 정규화 (Min-Max-Scaling)
 : 각 열의 값을 0과 1사이로 눌러서 담는 방법
- 가장 작은 값은 0, 가장 큰 값은 1이 되고
  나머지는 그 사이에 들어간다
- 공식 : (값 - 최소값) / (최대값 - 최소값)
         최소값이면 0, 최대값이 1이면 딱 가운데 값이 0.5
- scikit-lean의MinMaxScalar를 사용한다

2) 표준화 (Standardization)
 : 값을 평균 0, 표준편차 1이 되도록 스케일링
- 공식 : (값 - 평균) / 표준편차
  평균이면 0, 평균보다 크면 양수, 작으면 음수가 된다
- 정규화에 달리 0 ~ 1 같은 정해진 범위가 없다

3) Robust 스케일링
- 이상치에 강한 스케일링(이상치에 영향을 덜 받음)
- 위의 두 스케일링은 이상치에 약하다는 단점이 있다
	정규화는 최대값을 기준으로 사용하는데,
		이상치가 최대값이면 나머지 정상 값들이
		0 근처로 다 몰려버린다
	표준화는 평균, 표준편차를 쓰는데 이 둘 자체가 이상치에
		영향을 크게 받는다
- Robust 스케일링은 이상치에 영향을 덜 받는 중앙값과 IQR(사분위 범위)
	을 기준으로 계산한다
		공식 : (값 - 중앙값) / IQR

4) 스케일링 사용

==========================================================================
정규화		값의 범위를 0~1로 딱 정하고 싶다
표준화		대부분의 일반적인 상황
Robust		이상치가 많은 경우

2. 데이터프레임 결합
 : 여러개의 데이터 프레임을 하나의 표로 합치는 것
- 데이터가 한 파일/표에 다 들어 있는 경우는 오히려 드물다
		ex) 회원 + 주문(누가 무엇을 샀는지)
- 나뉜 채로 "짱구가 무엇을 샀나?" 같은 질문에 답하기 어렵다

1) pd.concat()
- 데이터를 그냥 단순히 이어 붙이는 결합
- 두 방향이 있다
	axis = 0 : (기본값) 데이터를 위/아래로 연결 => 행이 늘어난다
		* 같은 구조의 데이터를 합칠 때
	axis = 1 : 데이터를 좌/우로 연결 => 열이 늘어난다

2) pd.merge()
- 공통 열(키)로 합치기
- merge()는 공통 열의 값이 같은 행끼리 찾아 짝지어 합친다
- ex) 회원 데이터와 주문 데이터를 회원ID가 같은 것끼리 합쳐
      주문 옆에 주문자 정보를 붙여볼 수 있다
- 옵션
	on : 기준이 되는 공통 열
	how : 어떻게 합칠지
		(inner/left/right/outer)
	left_in/right_on 두 데이터의 키 열 이름이 다를때 각각 지정
		ex) left_in = "회원ID", right_on = "ID"

+) how 옵션 - 어떻게 합칠까?
  옵션 값		뜻			결과
=====================================================================
inner(기본값)		양쪽 모두에 있는 키만	교집합
left			왼쪽 표는 다 남기고	왼쪽 전부
			오른쪽은 있으면 붙임
right			오른쪽 표는 다 남기고	오른쪽 전부
			왼쪽은 있으면 붙임
outer 			어느 한쪽에라도 있으면 다 합칩함

* 짝을 못 찾은 칸은 NaN으로 채워진다

+) concat vs merge
	pd.concat()			pd.merge()
==========================================================================
방식	위치로 그냥 이어 붙이기		공통 열(키)를 맞춰 합치기
기준열  필요 없음			있어야함
언제 	같은 구조의 데이터를		서로 다른 데이터를
	위/아래로 쌓을때		연결할때 사용
ex) 	1월 데이터 + 2월 데이터		회원 + 주문(회원ID로 연결)
```

```python
## 스케일링
import pandas as pd
import seaborn as sns

# 타이타닉 데이터 불러오기
titanic = sns.load_dataset("titanic")

# 스케일링 전에 age의 결측치를 중앙값으로 채우기
titanic["age"] = titanic['age'].fillna(titanic['age'].median())

print(titanic['age'].isnull().sum())

print(titanic[['age', 'fare']].describe())
# age는 최대 80인데 fare는 최대 512이다.
# 두 열이 표현하는 범위가 다르다
### 정규화
from sklearn.preprocessing import MinMaxScaler

# 1) 스케일러 준비
scaler = MinMaxScaler()

# 2) age, fare 두 열을 한꺼번에 정규화(fit_transform)
scaled = scaler.fit_transform(titanic[["age", "fare"]])

# 3) 결과 확인
# 결과는 numpy배열이라 열 이름이 없다
scaled_minmax = pd.DataFrame(scaled, columns=["age", "fare"])
scaled_minmax.head()
# 80까지 가던 나이와 512까지 가던 요금이 모두 0 ~ 1 사이의 값으로 바뀐다
# => 이제 두 열의 범위가 같아짐
# 정말 최소값은 0, 최대값은 1인지 확인
print(f"각 열의 최대값 : {scaled_minmax.max().to_dict()}")
print(f"각 열의 최소값 : {scaled_minmax.min().to_dict()}")
# 공식 : (값 - 최소값) / (최대값 - 최소값)
age = titanic['age']
minmax = (age - age.min()) / (age.max() - age.min())
print(minmax.head())
### 표준화
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled = scaler.fit_transform(titanic[['age', 'fare']])
print(scaled)

scaled_std = pd.DataFrame(scaled, columns=['age', 'fare'])
scaled_std
# 값이 0을 중심으로 양수 음수로 퍼진다
# 음수는 평균보다 작다는 뜻, 양수는 크다는 뜻이다
# 평균 0, 표준편차 1이 되었는지 확인
print(f"각 열의 평균 : {scaled_std.mean().round(2).to_dict()}")
print(f"각 열의 표준편자 : {scaled_std.std().round(2).to_dict()}")
# 두 열 모두 평균0, 표준편차 1로 맞춰져 있다
## Robust 스케일링
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()
scaled = scaler.fit_transform(titanic[['age', 'fare']])

scaled_robust = pd.DataFrame(scaled, columns= ['age', 'fare'])
scaled_robust.head()
# 중앙값 0기준이라 중앙값 보다 크면 양수, 작으면 음수
# <스케일링 실습>
import seaborn as sns
# 펭귄 데이터 불러오기
df = sns.load_dataset("penguins")

# 결측치 처리(중앙값으로 채우기)
df['flipper_length_mm'] = df['flipper_length_mm'].fillna(df['flipper_length_mm'].median())
df['body_mass_g'] = df['body_mass_g'].fillna(df['flipper_length_mm'].median())

# 스케일링할 두 열을 선택
features = df[['flipper_length_mm', 'body_mass_g']]

df.head()
# 1) 팽귄의 무게와 발의 길이 두 열을 표준화하고, 
#    각 열의 평균과 표준편차 확인
std_scaler = StandardScaler()
std_result = pd.DataFrame(
    std_scaler.fit_transform(features), # 표준화 실행
    columns=['flipper_length_mm', 'body_mass_g']
)

print('==== 표준화 결과 ====')
print(f"평균 : {std_result.mean().round(2).to_dict()}")
print(f"표준편차 : {std_result.std().round(2).to_dict()}")

std_result.head()
# 2) 같은 두 열을 정규화하고,
#    각 열의 최대값과 최소값 확인
mm_sacler = MinMaxScaler()
mm_result = pd.DataFrame(
    mm_sacler.fit_transform(features), # 정규화 실행
    columns=['flipper_length_mm', 'body_mass_g']
)

# 결과 확인
print("==== 정규화 결과 ====")
print(f"최소값 : {mm_result.min().to_dict()}")
print(f"최대값 : {mm_result.max().to_dict()}")

mm_result.head()
## 데이터프레임 결합
import pandas as pd

# csv 파일 불러오기
members = pd.read_csv("./회원.csv")
orders = pd.read_csv("./주문.csv")
# 회원 데이터
members
# 주문 데이터
orders
# 두 데이터에 공통으로 있는 열이 회원ID이다.
# 이 열이 두 데이터를 이어주는 다리(키 key)역할을 한다
### concat()
# 같은 구조(이름, 점수)의 두 데이터프레임 생성
class1 = pd.DataFrame({
    "이름" : ["김하늘", "이바다"],
    "점수" : [90, 85]
})
class1
class2 = pd.DataFrame({
    "이름" : ["박구름", "최산들"],
    "점수" : [70, 95]
})
class2
# 세로로 이어붙이기(axis=0이 기본이라 생략 가능)
pd.concat([class1, class2])
# 행 번호를 0부터 새로 매기기
pd.concat([class1, class2], ignore_index=True)
# 가로 뱡향으로 늘리기
kor = pd.DataFrame({
    "이름" : ["짱구", "철수", "유리"],
    "국어" : [70, 85, 90]
})
kor
eng = pd.DataFrame({
    "영어" : [60, 80, 95]
})
eng
# axis = 1 : 좌우로 붙여 열을 늘린다
pd.concat([kor, eng], axis=1)
### pd.merge()
print(members)
print(orders)
# 두 데이터를 공통 열 '회원ID'기준으로 합치기
pd.merge(members, orders, on="회원ID")
# 회원 정보와 주문 정보가 회원ID를 키로 하나의 데이터로 합쳐짐
# M01(김하늘)은 주문이 2개라 두줄로 나온다
# 그런데, 주문이 없던 M03, M04와 회원 명단에 없던 주문 M05는 결과에서 빠졌다
# => how옵션 때문에! how 옵션은 기본값이 inner(양쪽 모두에 있는 것만)라서
# left : 왼쪽(회원) 데이터는 모두 남긴다 -> 주문이 없는 M03, M04도 포함
pd.merge(members, orders, on="회원ID", how="left")
# 없는 주문 정보는 NaN이 된다
# "모든 회원을 기준으로, 있으면 주문을 붙여라"
# right : 오른쪽(주문) 데이터는 모두 남긴다
pd.merge(members, orders, on="회원ID", how="right")
# 모든 주문을 기준으로 삼아, 회원 명단에 없던 M05의 주문도 남아있다
# (이름, 지역은 NaN)
# outer : 어느 한쪽에라도 있으면 모두 남긴다(합칩)
pd.merge(members, orders, on="회원ID", how="outer")
# 주문 없는 회원(M03, M04)도, 회원 아닌 주문(M05)도 모두 남고,
# 빈 곳은 NaN
# 키 열 이름이 다를때 - left_on / right_on
# 두 표의 기준 열 이름이 서로 다르면 on하나로는 안된다
# ex) 왼쪽은 회원ID, 오른쪽은 ID라면, 각각 어떤 열이 키인지 알려줘야 한다
orders2 = orders.rename(columns={"회원ID" : 'id'})
# : 회원 ID열 이름을 id로 바꿈
orders2
members
# 왼쪽은 회원ID, 오른쪽은 id를 키로 지정해 합치기
pd.merge(members, orders2, left_on="회원ID", right_on="id")
# 이름이 다른 두 열 (회원ID, id)을 짝지어 합쳤다.
# 결과엔 두 열이 모두 남는다
```
