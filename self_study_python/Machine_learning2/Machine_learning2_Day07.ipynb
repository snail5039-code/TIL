2. 기초통계와 상관관계
 : 데이터를 몇개의 대표 숫자로 요약"(기초 통계)하고
   두 값이 함께 움직이는 정도를 나타내는 (상관관계)
- 기초통계(기술 통계) : 데이터의 생김새를 요약한다
	대략 어디에 몰려 있는지, 얼마나 흩어져 있는지를 숫자로 본다
- 상관관계 : "A가 커지면 B도 커지나?"처럼
	두 변수가 같이 움직이는지를 -1 ~ 1 사이 숫자로 나타낸다

1) 대표값 - 평균, 중앙값, 최빈값
 : 데이터를 하나의 숫자로 대표하는 값이다
- mean()	평균	극단값(이상치)에 영향을 많이 받음
- median()	중앙값	이상치에 영향을 덜 받음(강함)
- mode()	최빈값 	가장 자주 나온 값

2) 산포토(퍼진 정도) = 표준편차, 분산, 분위수
 : 데이터가 얼마나 흩어졌나를 확인
- std()		표준편차 값들이 평균에서 평균적으로 얼마나
			떨어졌나(클수록 넓게 퍼짐)

- var()		분산	표준편차에 제곱한 값
- quantile(q)	분위수	하위 q지점의 값(0.25, 0.5, 0.75)

3) 상관계수 - 두 데이터가 같이 움직이는 정도
 : 두 숫자 데이터가 얼마나 함께 움직이는지를 -1 ~ 1 사이의
   숫자로 표현한다
** 상관계수 해석
   +1에 가까움	 한쪽이 커지면 다른 쪽도 커짐(양의 상관)
   0에 가까움	 서로 별 관계 없음
   -1에 가까움	 한쪽이 커지면 다른 쪽도 작아짐(음의 상관)
- df.corr()는 모든 숫자 열의 상관계수를 표(상관행렬)로 만들어준다
- numeric_only = True를 붙여 숫자 열만 계산가능

+) 상관관계는 중요하다!!
- 머신러닝에서 "생존과 관련된 열은 무엇인가?"를 상관계수로 미리 봐서
  모델에 넣을 후보 특성을 고를 출발점으로 사용한다

```python
import pandas as pd

cafe = pd.read_csv("./카페주문.csv")
print(cafe.dtypes)
cafe
# 주문일시와 회원가입일의 타입이 object => 그냥 문자열이다
# 판다스는 이걸 날짜로 인식하지 못한다 => 먼저 날짜로 바꿔야 한다
# 문자열 열을 날짜 타입으로 변환해서 원래 열에 덮어쓰기
cafe['주문일시'] = pd.to_datetime(cafe["주문일시"])
cafe['회원가입일'] = pd.to_datetime(cafe["회원가입일"])

print(cafe.dtypes)
cafe
# 날짜 뽑기
# 주문일시에서 '월', '요일', '몇시'를 뽑기
# 1) 월
print(cafe['주문일시'].dt.month)
# 2) 요일 번호(0=월요일...6=일요일)
print(cafe['주문일시'].dt.dayofweek)
print('='*15)

# 요일을 이름(영어)로 확인
print(cafe['주문일시'].dt.day_name())
# 3) 주문한 시간 뽑기
print(cafe['주문일시'].dt.hour)
# +) 추출한 날짜 및 시간을 활용하면, 그룹연산과 필터링에 사용 가능
cafe['월'] = cafe["주문일시"].dt.month
cafe['요일'] = cafe["주문일시"].dt.day_name()
cafe['시'] = cafe["주문일시"].dt.hour

cafe
### 날짜 계산
# 1) 기준이 되는 날짜 하나 만들기
# date = pd.Timestamp("2024-12-31")
date = pd.Timestamp.now()
print(date)

# 2) 기준일에서 회원가입을 빼면 기간(Timedelta)이 나온다
date - cafe["회원가입일"]
# 3) 기간에서 '일 수'만 정수로 뽑아 새 열로 저장
cafe['가입경과일'] = (date - cafe["회원가입일"]).dt.days
cafe

# +) 오늘을 기준으로 하려면
# pd.Timestamp.now() -> 지금시간
### 기간 필터링
# 주문일시가 2024-04-01 이후인 행만 필터링
cafe[cafe['주문일시'] >= "2024-04-01"]
# 날짜 범위를 시작 ~ 끝 범위로 필터링할때
# between이 편하다
# 4월 1일 ~ 4월 30일 사이의 주문만
mask = cafe["주문일시"].between("2024-04-01", "2024-04-30 23:59:59")
cafe[mask]
# 월별 매출 합계 계산
print(cafe.groupby(cafe["주문일시"].dt.month)["금액"].sum())

cafe
# 요일별 주문 건수를 계산
print(cafe.groupby(cafe["주문일시"].dt.day_name())['금액'].count())
# <날짜 시간 데이터 실습>
import pandas as pd

delivery = pd.read_csv("./택배배송.csv")
delivery
# 1) 주문일과 배송완료일을 날짜 타입으로 바꾸기(to_datetime)
delivery['주문일'] = pd.to_datetime(delivery['주문일'])
delivery['배송완료일'] = pd.to_datetime(delivery['배송완료일'])
delivery.dtypes
# 2) 배송에 며칠 걸렸는지 계산해 배송요일 열 만들기
#    (두 날짜 뺄셈 -> dt.dats)
delivery['배송소요일'] = (delivery['배송완료일'] - delivery["주문일"]).dt.days
delivery
# 3) 배송이 3일을 초과해 걸린(지연된)건만 출력
result = delivery[delivery["배송소요일"] > 3]
result
## 기초 통계와 상관관계
import pandas as pd
import seaborn as sns

# 타이타닉 데이터 불러오기
titanic = sns.load_dataset("titanic")
titanic.head()
# describe() : 숫자 열들의 주요 통계를 한 표로 보여준다
titanic.describe().round(2)
# 나이의 3가지 대표값 
print(f"나이 평균 : {titanic['age'].mean() : .2f}")
print(f"나이 중앙값 : {titanic['age'].median() : .2f}")
print(f"나이 최빈값 : {titanic['age'].mode()[0]}")
# 나이의 3가지 대표값이 비슷하다 =>
# 데이터가 한쪽으로 크게 치우치지 않았다는 뜻
# 요금의 평균 vs 중앙값
print(f"요금 평균 : {titanic['fare'].mean() : .2f}")
print(f"요금 중앙값 : {titanic['fare'].median() : .2f}")
# 평균과 중앙값이 2배가 넘는다 => 초고가 요금 몇 건이 평균을 확 끌어올린 것이다

## 산포도
# 나이의 퍼짐 정도
print(f"나이 표준편차 : {titanic['age'].std() : .2f}")
print(f"나이 분산 : {titanic['age'].var() : .2f}")
# 표준편차가 14.53이고 나이가 평균(29.70)에서 위 아래로 대략 15정도 흩어져
# 있다는 뜻 => 15 ~ 44 근처에 몰려 있다고 대략 확인할 수 있다
# 분위수
print(titanic['age'].quantile(0.25)) # 1사분위
print(titanic['age'].quantile(0.5)) # 2사분위(중앙값)
print(titanic['age'].quantile(0.75)) # 3사분위
print("="*15)
print(titanic['age'].quantile([0.25, 0.5, 0.75])) 
## 상관계수
# 주요 숫자 열들끼리 상관계수를 구해, 생존(survived)과
# 가장 관련 있는 값이 무엇인지 확인
cols = ['survived', 'age', 'sibsp', 'parch', 'fare']
titanic[cols].corr(numeric_only = True)
# survived 행을 보면 상관계수가 1이나 -1과 가까운 값이 없다
# => 나머지 열과 별 관계가 없음
# 나이와 요금, 딱 두 열의 상관계수 확인
print(f"나이와 요금 상관계수 : {titanic['age'].corr(titanic['fare'])}")
# 거의 0이다 => 별 관계 없음을 의미
import matplotlib.pyplot as plt
# 상관관계 시각화 - 산점도와 히트맵

# 1) 산점도(scatter) - 두 데이터를 x, y축에 점을 찍는다
# 점들이 한방향으로 몰려 있으면 상관관계가 강하고, 흩어져 있으면 약하다
plt.figure(figsize=(6, 4))
plt.scatter(titanic['age'], titanic['fare'], alpha=0.4)
plt.xlabel("age")
plt.ylabel("fare")
plt.show()
# 점들이 뚜렷한 기울기 없이 흩어져 있다
# 히트맵 - 상관행렬 전체를 색으로 표현(열이 여러개일때 관계를 확인)

plt.figure(figsize=(6, 5))
sns.heatmap(titanic[cols].corr(numeric_only=True),
           annot=True, cmap="coolwarm", vmin=-1, vmax=1, fmt=".2f")
# annot = True : 각 칸에 숫자 표시
# cmap="coolwarm" : 스타일(색상)
# vmin/vmax로 -1 ~ 1 범위 고정
plt.show()
# <기초 통계 상관관계 실습>
import seaborn as sns
import pandas as pd

# 팁 데이터 불러오기
tips = sns.load_dataset("tips")
tips.head()
# 1) 숫자 열의 통계 요약 정보 확인하고,
# total_bill의 평균과 중앙값을 각각 확인
print(f"총액의 평균 : {tips["total_bill"].mean() : .2f}")
print(f"총액의 중앙값 : {tips['total_bill'].median() : .2f}")
    
tips.describe().round(2)
# 2) 세 숫자 열의 상관행렬을 구하고,
#    total_bill과 tip의 상관계수 확인
print(tips['total_bill'].corr(tips['tip']))

tips.corr(numeric_only=True)
# 총액과 팁의 상관계수가 0.68로 높게 나온다
# 총액이 클수록 팁도 커지는 양의 상관관계를 가진다
# 3) 상관행렬을 히트맵으로 시각화
plt.figure(figsize=(5, 4))
sns.heatmap(tips.corr(numeric_only=True),
           annot=True, cmap="coolwarm", vmin=-1, vmax=1, fmt=".2f")
```
