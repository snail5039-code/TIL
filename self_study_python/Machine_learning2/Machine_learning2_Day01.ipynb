1. matplotlib

1) 산점도
- 두 값을 (x, y) 좌표의 점으로 찍어,
  두 변수가 어떤 관계인지 보여준다.
ex) "공부를 많이 할수록 점수가 오르나?"
- 산점도 읽기
 - 양의 관계(우상향) : 한 값이 커질수록 다른 값도 커진다
 - 음의 관계(우하양) : 한 값이 커질수록 다른 값은 작아진다
 - 관계 없음(흩어짐) : 뚜렷한 방향 없이 점이 사방에 퍼져 있다

2. 고급 시각화

Seaborn 라이브러리
- matplotlib의 기능과 스타일을 확장한 시각화 도구 고급 버전
- tips, titanic, iris 등 다양한 데이터셋을 제공한다

1) regplot() : 회귀선이 있는 산점도
- 서로 다른 2개의 변수 사이의 산점도 그래프를 그리고
  선형회귀분석에 의한 회귀선을 같이 나타낸다
+) 선형 회귀분석
- 머신러닝 알고리즘 중 기초적인 알고리즘
- 서로 영향을 주는 2개의 변수의 관계를 파악해 종속 변수의 값을 예측

2) kdeplot() : 밀도 그래프
- 하나의 데이터의 분포를 확인할때 사용하는 그래프
- 일정 범위로 구간을 만들어 x축으로 설정하고, y축은 집계된 값을
  나타내는 그래프

3) heatmap() : 히트맵
- 2개의 범주형 변수(카테고리 데이터)를 x, y축에 설정하고
  데이터를 메트릭스 형태로 분류 
- 값에 비례하여 색상을 다르게 해서 2차원으로 시각화

3. 웹 크롤링
 : 완성된 웹 페이지에서 필요한 정보를 수집하는 과정

+) 웹(web)
 : 인터넷을 통해 사용자들이 서로의 정보를 공유할 수 있는 공간

1) 웹 서버와 클라이언트

(1) 요청(request)
 : 클라이언트가 웹 서버에 필요한 정보를 요구하거나 처리해달라고
   부탁하는 행위
ex) 검색, 로그인

(2) 응답(response)
 : 웹 클라이언트가 요청을 수신받고 필요한 데이터를 전달하는 행위
ex) 검색 결과 화면

2) HTML 
 : 웹페이지를 만드는 문법을 갖춘 언어
   태그(Tag)로 구성되어 있음
- 여는 태그(Opening Tag) : 요소의 이름과 열고 닫는 꺽쇠 괄호로 구성됨
			   <span>
- 닫는 태그(Closing Tag) : 요소의 이름 앞에 슬래쉬(/)를 써서 구성
			   </span>
- 내용(Content) : 요소의 내용이며, 단순한 텍스트
- 요소(Element) : 여는 태그와 닫는 태그 내용을 통들어서 이른말

(1) html 기본적인 구조

태그의 종류
<div class="root">내용</div>
여는태그               닫는 태그

<html>  <부모의 시작 태그>
	<div></div> <자식 태그>
</html> <부모의 닫는 태그>

(2) 태그의 속성
- class : 태그들을 그룹화하는 공통적인 속성
	  (같은 class속성명을 사용하는 요소가 여러개다)
- id : 중복되지 않는 태그의 특징적인 속성
	(웹페이지 안에 id속성명은 중복되지 않는다)

3) 웹 크롤링 패키지

(1) requests
 : 파이썬에서 동작하는 작고 빠른 브라우저
   웹 사이트의 요청을 쉽게 처리하는 라이브러리
   HTML 문서를 가져올 때 사용

- request로 웹페이지 가져오기
import requests
url = "https://www.naver.com"
response = request.get(url)

- 속성(4가지)
	.status_code : 상태코드
		2xx(웹 크롤링 가능한 상태)
	.encoding : 언어설정
	.text : 웹페이지 소스(보기 쉬운 형태로)
	.content : 웹페이지 소스(모든 문자 그대로)

(2) Beatifulsoup
 : 구문을 해석해서 필요한 내용만 추출하는 패키지
- 메소드(5가지)
	.find("태그명") : 태그명이 "태그명"인 첫번째 것
		<div class = "first">Hello</div>
	.find("태그명").text : 내용만 추출
		Hello
	.find("태그명", class_="클래스속성명")
		: "태그명"인 것 중 클래스 속성명인 것의 첫번째
	.find("태그명", id='id속성명')
		: "태그명"인 것 중 아이디 속성명인 것의 첫번째
	.find_all("태그명") : "태그명"인 모든 것들을 
				리스트로 가져옴
	.find_all("태그명")[1].text
		=> 내용만 추출하기 위해서는 인덱스 번호를 지정해야함

```python
## matplotlib
### 산점도 그래프
import matplotlib.pyplot as plt

# 한글깨짐 방지
plt.rcParams["font.family"] = "Malgun Gothic"
# 데이터 생성
study_hour = [1, 2, 3, 4, 5, 6, 7, 8] # 공부시간
scores = [40, 45, 50, 60, 65, 70, 85, 90] # 시험 점수

# 산점도 시각화 scatter()
plt.scatter(study_hour, scores, color="green")
plt.title("공부시간과 점수의 관계")
plt.xlabel("공부시간")
plt.ylabel("점수")
plt.show()
# 점들이 오른쪽 위로 올라가는 모양
# : 한 값이 커질 때 다른 값도 커지는 양의 상관 관계
# (공부시간 높을수록 점수도 높다)
import seaborn as sns

# tips 데이터 불러오기
tips = sns.load_dataset("tips")
tips.head()
# 결제액과 팁의 관계 확인
plt.scatter(tips["total_bill"], tips["tip"], alpha = 0.5)
#alpha = 0.5 (투명도)
plt.title("결제액과 팁의 관계")
plt.xlabel("결제액")
plt.ylabel("팁")
plt.show()
# 결제액이 클수록 팁도 대체로 많다
# +) 그래프 영역 나누기 : subplot()
# 다른 그래프를 한 화면에 볼 때 사용

# 그래프 전체 영역 설정
plt.figure(figsize=(10, 4)) # 가로 10, 세로 4

plt.subplot(1, 2, 1) # 1행 2열 중 1번째 칸
plt.hist(tips["total_bill"], bins=20, color="orange", edgecolor="black")
plt.title("결제액 분포")

plt.subplot(1, 2, 2) # 1행 2열 중 2번째 칸
plt.scatter(tips["total_bill"], tips["tip"], alpha= 0.5)
plt.title("결제액 vs 팁")

plt.tight_layout() # 그래프끼리 겹치지 않게 간격 자동 조정
plt.show()
## seaborn
### 1) regplot()
import seaborn as sns

# 타이타닉 데이터 셋
titanic = sns.load_dataset("titanic")

titanic.head()
# 타이타닉 승객들의 요금과 나이의 상관관계를 나타내는 산점도

# 그래프 전체 영역 설정
fig = plt.figure(figsize=(15, 5))
# 보조 그래프 위치 설정
ax1 = fig.add_subplot(1, 2, 1) # 1행 2열 1번째
ax2 = fig.add_subplot(1, 2, 2) # 1행 2열 2번째

# 스타일 테마
sns.set_style("darkgrid")

sns.regplot(x='age',
           y='fare',
           data = titanic,
           ax=ax1) # 위치 설정
sns.regplot(x='age',
           y='fare',
           data = titanic,
           fit_reg= False, # 회귀선 미표시
           color = "orange", 
           ax=ax2) # 위치 설정

plt.show()
### kdeplot()
# 그래프 영역 설정
fig = plt.figure(figsize=(15, 5))
ax1 = fig.add_subplot(1, 3, 1)
ax2 = fig.add_subplot(1, 3, 2)
ax3 = fig.add_subplot(1, 3, 3)

# 시각화
# 1) 타이타닉 승객들의 age 밀도 그래프
sns.kdeplot(x='age', data=titanic, ax = ax1)

# 2) 생존여부(survived)에 따른 나이 밀도 그래프(기준이 설정된 밀도 그래프)
sns.kdeplot(x='age', data=titanic, ax=ax2, hue='survived')

# +) 타이타닉 승객들의 나이 히스토그램
sns.histplot(titanic["age"], ax=ax3)

plt.show()
### heatmap()
# 타이타닉 승객의 나이와 요금의 상관관계를 나타내는 히트맵

ti_age_fare = titanic[['age', 'fare']]

# 히트맵 시각화
sns.heatmap(data=ti_age_fare.corr(),
           annot = True, # 숫자 표시
           cmap="coolwarm", # 색상
           linewidth=0.7) # 구분석

plt.show()
# 상관관계
ti_age_fare.corr() # 상관계수 확인 통계 메소드
## 네이버 뉴스 웹크롤링
import requests
from bs4 import BeautifulSoup as bs

url = "https://news.naver.com/"
response = requests.get(url) # url에 요청을 보냄
# 크롤링 가능한 상태인지 확인
print(response.status_code) # 200 : 2xx => 크롤링 가능한 상태
html = response.text
soup = bs(html, "html.parser") # html 코드를 추출할 수 있도록 변환
# print(soup)
# find()
span = soup.find("span")
# print(span) # 웹페이지 안에서 첫번째 나온 태그명 하나만 추출
print(span.text)

span_li = soup.find_all("span") # 웹페이지 안에 있는 span요소 전부
# print(span_li)

# 얼론사별 이라는 문자만 추출
text = soup.find("span", class_="Nitem_link_menu").text
print(text)
## 웹 크롤링으로 야구 순위 가져오기
import requests
from bs4 import BeautifulSoup as bs
url = "https://www.koreabaseball.com/Record/TeamRank/TeamRankDaily.aspx"
# 1) 상태 확인
response = requests.get(url)
print(response.status_code)

# 2) html 코드 불러오기
html = response.text
soup = bs(html, "html.parser")
# print(soup)
# 팀순위라는 글자 가져오기
data = soup.find("h4", class_= "tit-page").text
print(data)

# 순위 글자 가져오기
th_li = soup.find_all("th") # th태그로 된 요소들 리스트로 가져오기
# print(th_li)
print(th_li[0].text)
# 순위표 가져오기
table = soup.find("tbody")

# 순위표의 행들 가져오기
rows = table.find_all("tr")

# 1순위 팀의 순위와 구단명
tr = rows[0]

# 첫번째 행에서 순위와 구단명 가져오기
print(tr.find_all("td"))
print(tr.find_all("td")[0].text)
print(tr.find_all("td")[1].text)
# for문으로 1위 ~ 10위 추출
for tr in rows :
    # print(tr)
    td = tr.find_all("td")
    # 순위
    rank = td[0].text.strip() # strip() : 공백문자 제거
    # 팀명
    team = td[1].text.strip()
    print(f"순위 : {rank}위, 팀명 : {team}")
# 추출한 내용을 text 파일로 저장
with open("./ranking.txt", "w", encoding="utf-8") as f :
    for tr in rows :
        td = tr.find_all("td")
        rank = td[0].text.strip() 
        team = td[1].text.strip()
        f.write(f"{rank}위 : {team}\n")

print("순위와 팀명이 파일에 저장되었습니다")
```

문제 1) 네이버 기사 웹 크롤링 후 메모장에 저장
- 본인이 원하는 기사 선정하여 크롤링
- 메모장명 : 'python_news.txt'
- ※메모장은 과제 제출시 따로 제출할 것

```python
import requests
from bs4 import BeautifulSoup as bs
url = "https://m.sports.naver.com/kbaseball/article/076/0004430475"
response = requests.get(url)
# print(response.status_code)

html = response.text
soup = bs(html, "html.parser")
# print(soup.text)
title = soup.find("h2", class_="ArticleHead_article_title__qh8GV").text
print(title)

body = soup.find("div", class_="_article_content").get_text("\n", strip=True) # get_text : 태그 마다 줄바꿈
print(body)

with open("./python_news.txt", "w", encoding="utf-8") as f :
    f.write("네이버 기사 크롤링\n\n")
    f.write(f"기사 제목 : {title}\n\n")
    f.write(f"기사 내용 : {body}\n")
print("작성 완료")
```
